### An Agentic AI Framework for Personal Finance, Portfolio Optimization & Integrated Risk Analysis

### Product Overview

Financial Decision Intelligence is an **agentic AI-powered analytical platform** designed to support personal financial management, investment planning, and forward-looking financial decisions within a unified framework.

The project originates from a fundamental principle: **effective personal financial management requires a comprehensive approach**, combining the monitoring of income and expenses, control of cash flows, identification of potential financial risks, and corrective actions to maintain liquidity and financial resilience. Equally important is the effective management of financial surpluses, ensuring that accumulated savings are not left idle but are strategically allocated to investments, taking into account liquidity needs, risk exposure, and long-term financial objectives.

**The personal finance component uses an illustrative household budget constructed from a CSV dataset containing 12 months of income and expense records. The investment component uses historical market data for real equities belonging to the STOXX Europe 50 index.**

The integrated analysis **builds on a 12-month monitoring period of personal income and expenses to assess cash flows and savings capacity, providing the basis for a balanced and prudent allocation of accumulated savings into an optimized investment portfolio with defensive characteristics**. The integration of personal financial planning and portfolio management brings together projections of personal finances, including potential financial shocks, with simulated portfolio performance to assess the evolution of liquidity and investments under different financial and market conditions. Specifically, it combines **five personal financial scenarios** (baseline conditions, a prolonged reduction in employment income, extraordinary expenses, and increased recurring expenses) with **three investment market scenarios** (unfavorable, central, and favorable). These market scenarios are intended as **illustrative proxies for potential portfolio gains and losses**, providing a broader perspective on how investment performance could interact with personal cash flows, liquidity needs, and potential financial shocks. They are estimated using **historical bootstrap simulations**, which resample observed stock returns to capture historical market fluctuations, volatility, and extreme movements. The purpose is to incorporate potential investment outcomes into the broader financial assessment, allowing personal budgeting, financial resilience, and investment risk to be evaluated together. The resulting **integrated combinations**, evaluated over rolling horizons from 1 to 24 months, allow the joint assessment of liquidity resilience, investment performance, and overall financial sustainability under alternative future conditions.

In conclusion, the platform goes beyond monitoring financial performance. It aims to **transform financial data into actionable insights**, enabling users to understand their current financial position, **identify recurring expenses** and opportunities for improvement, **allocate savings more efficiently**, and evaluate the **sustainability of financial decisions** under alternative future conditions.

### Investment Methodology

The investment methodology follows a **structured, sequential process**, in which each analytical stage informs the subsequent modelling decisions:

I. **Historical market data analysis:** analysis of five years of daily price and return series for 50 equities belonging to the STOXX Europe 50 universe.

II. **Risk-return assessment and selection:** calculation of annualized returns and volatility, followed by a combined ranking assigning equal weight to both measures to select the top 25 securities.

III. **Return distribution analysis:** application of the Jarque-Bera normality test and excess **kurtosis analysis** revealed non-normal return distributions with heavy tails, indicating the presence of extreme market movements. These findings motivated the use of historical bootstrap simulations to better capture the distinctive risk characteristics of individual stocks.

IV. **Portfolio construction and optimization:** covariance analysis and **Monte Carlo simulation** of 5,000 candidate portfolios, selecting the maximum **Sharpe Ratio** portfolio subject to diversification constraints, including a maximum of four securities per sector.

**V. Historical bootstrap simulation:** non-parametric **resampling of observed returns over the preceding five years** to generate 1,000 potential portfolio trajectories over the **subsequent 24 months**, incorporating the **historical peaks, downturns, and extreme fluctuations** of individual stocks while preserving contemporaneous cross-asset relationships.

VI. **Forward-looking scenario analysis:** calculation of **three potential portfolio outcomes** based on the 5th, 50th, and 95th percentiles of **simulated performance**, representing:

- **Unfavorable:** downside scenario (5th percentile).

- **Central:** neutral scenario (50th percentile).

- **Favorable:** upside scenario (95th percentile).

```python
# ============================================================
# BOX 1
#
# PERSONAL FINANCE — 12 MONTHS (NOV 2025 – OCT 2026)
# MONTHLY CHARTS + STATISTICS + OVERALL SUMMARY
# ============================================================

df = pd.read_csv(
    "https://raw.githubusercontent.com/Valeria-Raele/financial-plan-intelligence/refs/heads/main/data/personal_finance_transactions_12_months_Nov2025_Oct2026.csv"
)

df["Amount_EUR"] = pd.to_numeric(df["Amount_EUR"])

# ------------------------------------------------------------
# 1. MONTHLY FINANCIAL DATA
# ------------------------------------------------------------

monthly = df.pivot_table(
    index="Month",
    columns="Type",
    values="Amount_EUR",
    aggfunc="sum",
    fill_value=0
).sort_index()

monthly["Balance"] = monthly["Income"] - monthly["Expense"]

expenses = df[df["Type"] == "Expense"].copy()
income = df[df["Type"] == "Income"].copy()

monthly_expenses = expenses.pivot_table(
    index="Month",
    columns="Category",
    values="Amount_EUR",
    aggfunc="sum",
    fill_value=0
).sort_index()

# Same category order across all months
categories = monthly_expenses.sum().sort_values().index
monthly_expenses = monthly_expenses[categories]

# ------------------------------------------------------------
# 2. MONTHLY CHARTS AND STATISTICS
# ------------------------------------------------------------

n_months = len(monthly_expenses)
n_cols = 2
n_rows = int(np.ceil(n_months / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(20, n_rows * 8)
)

axes = np.array(axes).reshape(-1)

for i, (month, values) in enumerate(monthly_expenses.iterrows()):

    ax = axes[i]

    month_income = monthly.loc[month, "Income"]
    month_expense = monthly.loc[month, "Expense"]
    month_balance = monthly.loc[month, "Balance"]

    ax.barh(
        values.index,
        values.values
    )

    ax.set_title(
        f"{month} | Total Expenses: €{month_expense:,.0f}",
        fontsize=13,
        fontweight="bold",
        pad=12
    )

    ax.set_xlim(0, max(values.max() * 1.15, 1))
    ax.tick_params(axis="y", labelsize=9)
    ax.tick_params(axis="x", labelsize=9)
    ax.grid(axis="x", alpha=0.2)
    ax.set_axisbelow(True)

    active_categories = values[values > 0].sort_values(
        ascending=False
    )

    category_lines = []

    for category, amount in active_categories.items():

        weight = amount / month_expense * 100

        category_lines.append(
            f"{category}: €{amount:,.2f} ({weight:.1f}%)"
        )

    midpoint = int(np.ceil(len(category_lines) / 2))

    left_lines = category_lines[:midpoint]
    right_lines = category_lines[midpoint:]

    summary_text = (
        f"INCOME: €{month_income:,.2f}     "
        f"EXPENSES: €{month_expense:,.2f}     "
        f"BALANCE: €{month_balance:,.2f}"
    )

    ax.text(
        0.0, -0.13,
        summary_text,
        transform=ax.transAxes,
        fontsize=10,
        fontweight="bold",
        va="top"
    )

    ax.text(
        0.0, -0.20,
        "\n".join(left_lines),
        transform=ax.transAxes,
        fontsize=9,
        va="top",
        linespacing=1.45
    )

    ax.text(
        0.53, -0.20,
        "\n".join(right_lines),
        transform=ax.transAxes,
        fontsize=9,
        va="top",
        linespacing=1.45
    )

for j in range(n_months, len(axes)):
    axes[j].axis("off")

fig.suptitle(
    "PERSONAL FINANCE — MONTHLY EXPENSES AND STATISTICS",
    fontsize=20,
    fontweight="bold",
    y=0.999
)

plt.subplots_adjust(
    left=0.13,
    right=0.97,
    top=0.975,
    bottom=0.025,
    hspace=1.15,
    wspace=0.35
)

plt.show()

# ============================================================
# 3. OVERALL STATISTICS — 12 MONTHS
# ============================================================

total_income = monthly["Income"].sum()
total_expenses = monthly["Expense"].sum()
net_balance = monthly["Balance"].sum()

savings_rate = (
    net_balance / total_income * 100
    if total_income > 0 else np.nan
)

print("\n" + "=" * 75)
print("PERSONAL FINANCE — OVERALL STATISTICS")
print("=" * 75)

print(f"Period:                   {monthly.index[0]} to {monthly.index[-1]}")
print(f"Number of months:         {n_months}")
print(f"Total income:             €{total_income:,.2f}")
print(f"Total expenses:           €{total_expenses:,.2f}")
print(f"Net balance:              €{net_balance:,.2f}")
print(f"Savings rate:             {savings_rate:.2f}%")

print("\nMONTHLY AVERAGES AND DISTRIBUTION")
print("-" * 75)

print(f"Average monthly income:   €{monthly['Income'].mean():,.2f}")
print(f"Average monthly expenses: €{monthly['Expense'].mean():,.2f}")
print(f"Average monthly balance:  €{monthly['Balance'].mean():,.2f}")

print(f"Median monthly income:    €{monthly['Income'].median():,.2f}")
print(f"Median monthly expenses:  €{monthly['Expense'].median():,.2f}")
print(f"Median monthly balance:   €{monthly['Balance'].median():,.2f}")

print(f"Highest monthly income:   €{monthly['Income'].max():,.2f}")
print(f"Highest monthly expenses: €{monthly['Expense'].max():,.2f}")
print(f"Lowest monthly expenses:  €{monthly['Expense'].min():,.2f}")

print(f"Months with surplus:      {(monthly['Balance'] > 0).sum()}")
print(f"Months with deficit:      {(monthly['Balance'] < 0).sum()}")
print(f"Months with zero balance: {(monthly['Balance'] == 0).sum()}")

# ------------------------------------------------------------
# 4. EXPENSE DISTRIBUTION BY CATEGORY
# ------------------------------------------------------------

expense_stats = (
    expenses.groupby("Category")["Amount_EUR"]
    .agg(["sum", "count"])
    .sort_values("sum", ascending=False)
)

expense_stats.columns = [
    "Total_EUR",
    "Transactions"
]

expense_stats["Weight_%"] = (
    expense_stats["Total_EUR"] / total_expenses * 100
)

expense_stats["Monthly_Average_EUR"] = (
    expense_stats["Total_EUR"] / n_months
)

print("\n" + "=" * 75)
print("TOTAL EXPENSES BY CATEGORY — 12 MONTHS")
print("=" * 75)

print(expense_stats.round(2).to_string())

# ------------------------------------------------------------
# 5. INCOME DISTRIBUTION BY CATEGORY
# ------------------------------------------------------------

income_stats = (
    income.groupby("Category")["Amount_EUR"]
    .agg(["sum", "count"])
    .sort_values("sum", ascending=False)
)

income_stats.columns = [
    "Total_EUR",
    "Transactions"
]

income_stats["Weight_%"] = (
    income_stats["Total_EUR"] / total_income * 100
)

print("\n" + "=" * 75)
print("TOTAL INCOME BY CATEGORY — 12 MONTHS")
print("=" * 75)

print(income_stats.round(2).to_string())

# ------------------------------------------------------------
# 6. OVERALL CHART — EXPENSE DISTRIBUTION
# ------------------------------------------------------------

expense_plot = expense_stats.sort_values("Total_EUR")

fig, ax = plt.subplots(figsize=(15, 10))

bars = ax.barh(
    expense_plot.index,
    expense_plot["Total_EUR"],
    height=0.65
)

for bar, (category, row) in zip(
    bars,
    expense_plot.iterrows()
):
    amount = row["Total_EUR"]
    weight = row["Weight_%"]

    ax.text(
        amount + expense_plot["Total_EUR"].max() * 0.012,
        bar.get_y() + bar.get_height() / 2,
        f"€{amount:,.0f} ({weight:.1f}%)",
        va="center",
        fontsize=11
    )

ax.set_xlim(
    0,
    expense_plot["Total_EUR"].max() * 1.35
)

ax.set_title(
    "Overall Expenses by Category — 12 Months",
    fontsize=17,
    fontweight="bold",
    pad=22
)

ax.set_xlabel("Total Expenses (EUR)")
ax.set_ylabel("")
ax.grid(axis="x", alpha=0.2)
ax.set_axisbelow(True)

fig.subplots_adjust(
    left=0.20,
    right=0.93,
    top=0.90,
    bottom=0.11
)

plt.show()

# ------------------------------------------------------------
# 7. MONTHLY INCOME VS EXPENSES — 12 MONTHS
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(16, 11))

y = np.arange(len(monthly))
bar_height = 0.36

ax.barh(
    y - bar_height / 2,
    monthly["Income"],
    height=bar_height,
    label="Income",
    color="#4472C4"
)

ax.barh(
    y + bar_height / 2,
    monthly["Expense"],
    height=bar_height,
    label="Expenses",
    color="#ED7D31"
)

ax.set_yticks(y)
ax.set_yticklabels(monthly.index, fontsize=10)
ax.invert_yaxis()

max_amount = monthly[["Income", "Expense"]].max().max()

for i, (month, row) in enumerate(monthly.iterrows()):

    ax.text(
        row["Income"] + max_amount * 0.012,
        i - bar_height / 2,
        f"€{row['Income']:,.0f}",
        va="center",
        fontsize=9
    )

    ax.text(
        row["Expense"] + max_amount * 0.012,
        i + bar_height / 2,
        f"€{row['Expense']:,.0f}",
        va="center",
        fontsize=9
    )

ax.set_xlim(0, max_amount * 1.22)

ax.set_title(
    "Monthly Income vs Expenses — 12 Months",
    fontsize=17,
    fontweight="bold",
    pad=20
)

ax.set_xlabel("EUR")
ax.legend(loc="lower right")
ax.grid(axis="x", alpha=0.15)
ax.set_axisbelow(True)

fig.subplots_adjust(
    left=0.13,
    right=0.95,
    top=0.93,
    bottom=0.08
)

plt.show()

# ------------------------------------------------------------
# 8. MONTHLY SAVINGS AND DEFICITS — 12 MONTHS
# ------------------------------------------------------------

balance = monthly["Balance"]

balance_pct = (
    balance / monthly["Income"].replace(0, np.nan) * 100
)

colors = [
    "#2E8B57" if value >= 0 else "#D9534F"
    for value in balance
]

fig, ax = plt.subplots(figsize=(15, 11))

y = np.arange(len(monthly))

ax.barh(
    y,
    balance.values,
    color=colors,
    height=0.68
)

ax.axvline(0, color="#333333", linewidth=1.2)

max_balance = max(
    abs(balance.min()),
    abs(balance.max()),
    1
)

limit = max_balance * 1.45

ax.set_xlim(-limit, limit)

ax.set_yticks(y)
ax.set_yticklabels(monthly.index, fontsize=10)
ax.invert_yaxis()

for i, value in enumerate(balance.values):

    pct = balance_pct.iloc[i]

    label = (
        f"€{value:+,.0f} ({pct:+.1f}%)"
        if pd.notna(pct)
        else f"€{value:+,.0f}"
    )

    offset = max_balance * 0.025

    ax.text(
        value + offset if value >= 0 else value - offset,
        i,
        label,
        va="center",
        ha="left" if value >= 0 else "right",
        fontsize=9,
        fontweight="bold"
    )

ax.set_title(
    "Monthly Savings and Deficits — 12 Months",
    fontsize=17,
    fontweight="bold",
    pad=20
)

ax.set_xlabel("Monthly Net Balance (EUR)")
ax.grid(axis="x", alpha=0.15)
ax.set_axisbelow(True)

total_surpluses = balance[balance > 0].sum()
total_deficits = balance[balance < 0].sum()

summary = (
    f"TOTAL SURPLUSES: €{total_surpluses:,.2f}     "
    f"TOTAL DEFICITS: €{total_deficits:,.2f}     "
    f"NET BALANCE: €{net_balance:,.2f}"
)

fig.text(
    0.5,
    0.025,
    summary,
    ha="center",
    fontsize=11,
    fontweight="bold"
)

fig.subplots_adjust(
    left=0.14,
    right=0.95,
    top=0.91,
    bottom=0.10
)

plt.show()
```

```python
# ============================================================
# BOX 3
#
# Task and Tool Definitions
# PERSONAL FINANCE — ANALYTICAL TOOLS
# ============================================================
#
# TOOL 1 — analyze_finance
# General financial analysis:
# - Monthly and annual income, expenses and balances
# - Overall financial statistics
# - Savings and savings rate
# - Income and expense categories
# - Top 3 income and expense items
# - Identification of surplus and deficit months
#
# TOOL 2 — analyze_deficit_expenses
# Detailed analysis of deficit months:
# - Identification of months with negative balances
# - Top 5 expenses in each deficit month
# - Expense category and description
# - Percentage contribution to monthly expenses
#
# TOOL 3 — compare_income_expenses
# Comparative financial analysis:
# - Income vs expenses across 24 months
# - Comparison between two months
# - Annual comparisons
# - Expense-to-income ratios
# - Savings rates and financial balance variations
#
# All tools:
# - Read the existing DataFrame df
# - Perform independent calculations
# - Do not depend on shared global variables
# - Return text results compatible with LangGraph
#
# REQUIRED DATAFRAME COLUMNS:
# Month, Transaction Type, Amount, Category, Description
#
# ============================================================


# Flexible recognition: accents, punctuation, and lexical variants.

def _finance_normalize(value):
    value = unicodedata.normalize("NFKD", str(value).lower())
    value = "".join(ch for ch in value if not unicodedata.combining(ch))
    return re.sub(r"\s+", " ", re.sub(r"[^a-z0-9%]+", " ", value)).strip()

def _finance_any(query, *phrases):
    q = " " + _finance_normalize(query) + " "
    return any(" " + _finance_normalize(p) + " " in q for p in phrases)

def _finance_has(query, *stems):
    tokens = _finance_normalize(query).split()
    return any(any(token.startswith(stem) for token in tokens) for stem in stems)

# ============================================================
# ROLLING PERIOD — COMMON DATA FILTER
# ============================================================

def get_rolling_data(rolling_months: int = 24):

    if not 1 <= rolling_months <= 24:
        raise ValueError("Allowed period: 1 to 24 months.")

    data = df.copy()
    if "Amount" not in data.columns:
        data["Amount"] = pd.to_numeric(data["Amount_EUR"])
    if "Transaction Type" not in data.columns:
        data["Transaction Type"] = data["Type"].map({
            "Income": "credit", "Expense": "debit"
        })
    if "Month" not in data.columns:
        dates = data["Date"] if "Date" in data.columns else data.index
        data["Month"] = pd.to_datetime(dates).to_period("M").to_timestamp()
    data["Month"] = pd.to_datetime(data["Month"])

    last_month = data["Month"].max().to_period("M")
    first_month = last_month - (rolling_months - 1)

    data = data[
        (data["Month"].dt.to_period("M") >= first_month) &
        (data["Month"].dt.to_period("M") <= last_month)
    ].copy()

    return data


# ============================================================
# TOOL 1 — GENERAL FINANCIAL ANALYSIS
# ============================================================

@tool
def analyze_finance(query: str, rolling_months: int = 24) -> str:
    """
    General personal finance analysis.

    Use for financial summaries, monthly and annual totals,
    income, expenses, balances, savings, income categories,
    expense categories, rankings and deficit-month identification.

    Supported requests:
    - "monthly totals"
    - "overall summary"
    - "overall financial statistics"
    - "annual summary"
    - "average monthly savings"
    - "expense categories"
    - "income categories"
    - "top 3"
    - "deficit months"

    For detailed expenses within deficit months, use
    analyze_deficit_expenses.

    For comparisons between periods or financial ratios,
    use compare_income_expenses.
    """

    q = _finance_normalize(query)
    data = get_rolling_data(rolling_months)

    monthly = (
        data.groupby(["Month", "Transaction Type"])["Amount"]
        .sum()
        .unstack(fill_value=0)
        .sort_index()
    )

    for col in ["credit", "debit"]:
        if col not in monthly.columns:
            monthly[col] = 0.0

    if monthly.empty:
        return "No transactions available."

    monthly["Balance"] = monthly["credit"] - monthly["debit"]

    total_income = monthly["credit"].sum()
    total_expenses = monthly["debit"].sum()
    net_balance = monthly["Balance"].sum()
    n_months = len(monthly)

    savings_rate = (
        net_balance / total_income * 100
        if total_income > 0 else np.nan
    )

    expenses = data[data["Transaction Type"] == "debit"]
    income = data[data["Transaction Type"] == "credit"]

    # --------------------------------------------------------
    # 1. MONTHLY TOTALS
    # --------------------------------------------------------

    if _finance_any(q, 'monthly totals', 'monthly summary', 'monthly income', 'monthly expenses', 'monthly balance', 'monthly net cash flow', 'monthly aggregation'):

        return (
            "MONTHLY FINANCIAL SUMMARY\n\n"
            + monthly.to_string(
                float_format=lambda x: f"{x:,.2f}"
            )
        )

    # --------------------------------------------------------
    # 2. TOP 3 INCOME AND EXPENSE ITEMS
    # --------------------------------------------------------

    elif _finance_any(q, "top 3", "top three", "largest income items", "largest expenses", "highest income", "highest expenses", "income and expense ranking"):

        results = []

        for ttype, label in [
            ("credit", "INCOME"),
            ("debit", "EXPENSES")
        ]:

            subset = data[
                data["Transaction Type"] == ttype
            ]

            top = (
                subset.groupby(["Month", "Description"])["Amount"]
                .sum()
                .reset_index()
                .sort_values(
                    ["Month", "Amount"],
                    ascending=[True, False]
                )
                .groupby("Month")
                .head(3)
            )

            results.append(
                f"\nTOP 3 {label} BY MONTH\n"
                + top.to_string(index=False)
            )

        return "\n".join(results)

    # --------------------------------------------------------
    # 3. SAVINGS ANALYSIS
    # --------------------------------------------------------

    elif any(x in q for x in [
        'average monthly savings',
        'mean monthly savings',
        'average savings',
        'monthly surplus average',
        'average net cash flow',
        'average amount saved',
    ]):

        return (
            "SAVINGS ANALYSIS\n\n"
            f"Average Monthly Savings: "
            f"€{monthly['Balance'].mean():,.2f}\n"
            f"Net Savings: €{net_balance:,.2f}\n"
            f"Savings Rate: {savings_rate:.2f}%\n"
            f"Months with Surplus: "
            f"{(monthly['Balance'] > 0).sum()}\n"
            f"Months with Deficit: "
            f"{(monthly['Balance'] < 0).sum()}"
        )

    # --------------------------------------------------------
    # 4. EXPENSE DISTRIBUTION BY CATEGORY
    # --------------------------------------------------------

    elif any(x in q for x in [
        'expense categories',
        'spending categories',
        'expenditure breakdown',
        'expenses by category',
        'spending distribution',
        'category expenses',
    ]):

        stats = (
            expenses.groupby("Category")["Amount"]
            .agg(["sum", "count"])
            .sort_values("sum", ascending=False)
        )

        stats.columns = ["Total_EUR", "Transactions"]

        stats["Weight_%"] = (
            stats["Total_EUR"] / total_expenses * 100
            if total_expenses > 0 else 0
        )

        stats["Monthly_Average_EUR"] = (
            stats["Total_EUR"] / n_months
        )

        return (
            "EXPENSE DISTRIBUTION BY CATEGORY\n\n"
            + stats.round(2).to_string()
        )

    # --------------------------------------------------------
    # 5. INCOME DISTRIBUTION BY CATEGORY
    # --------------------------------------------------------

    elif any(x in q for x in [
        'income categories',
        'income sources',
        'income breakdown',
        'income distribution',
        'income by category',
        'revenue categories',
    ]):

        stats = (
            income.groupby("Category")["Amount"]
            .agg(["sum", "count"])
            .sort_values("sum", ascending=False)
        )

        stats.columns = ["Total_EUR", "Transactions"]

        stats["Weight_%"] = (
            stats["Total_EUR"] / total_income * 100
            if total_income > 0 else 0
        )

        stats["Monthly_Average_EUR"] = (
            stats["Total_EUR"] / n_months
        )

        return (
            "INCOME DISTRIBUTION BY CATEGORY\n\n"
            + stats.round(2).to_string()
        )

    # --------------------------------------------------------
    # 6. DEFICIT MONTH IDENTIFICATION
    # --------------------------------------------------------

    elif any(x in q for x in [
        'which months have a deficit',
        'months have a deficit',
        'deficit months',
        'months in deficit',
        'negative cash flow',
        'negative monthly balance',
        'expenses exceed income',
        'monthly shortfall',
    ]):

        negative = monthly[
            monthly["Balance"] < 0
        ]

        if negative.empty:
            return "There are no deficit months."

        return (
            "MONTHS WITH NEGATIVE BALANCE\n\n"
            + negative.to_string(
                float_format=lambda x: f"{x:,.2f}"
            )
        )

    # --------------------------------------------------------
    # 7. ANNUAL SUMMARY
    # --------------------------------------------------------

    elif any(x in q for x in [
        'annual summary',
        'yearly summary',
        'annual totals',
        'yearly income',
        'yearly expenses',
        'annual net savings',
        'annual financial overview',
    ]):

        annual = monthly[["credit", "debit", "Balance"]].copy()
        annual["Year"] = annual.index.year

        annual = annual.groupby("Year").sum()

        return (
            "ANNUAL FINANCIAL SUMMARY\n"
            "(Available months only)\n\n"
            + annual.to_string(
                float_format=lambda x: f"{x:,.2f}"
            )
        )

    # --------------------------------------------------------
    # 8. OVERALL FINANCIAL STATISTICS
    # --------------------------------------------------------

    elif any(x in q for x in [
        'overall summary',
        'general summary',
        'financial overview',
        'total income',
        'total expenses',
        'total savings',
        'overall cash flow',
    ]):

        return (
            "PERSONAL FINANCE — OVERALL STATISTICS\n\n"
            f"Period: {monthly.index.min():%Y-%m} "
            f"to {monthly.index.max():%Y-%m}\n"
            f"Months: {n_months}\n\n"
            f"Total Income: €{total_income:,.2f}\n"
            f"Total Expenses: €{total_expenses:,.2f}\n"
            f"Net Balance: €{net_balance:,.2f}\n"
            f"Savings Rate: {savings_rate:.2f}%\n\n"
            f"Average Monthly Income: "
            f"€{monthly['credit'].mean():,.2f}\n"
            f"Average Monthly Expenses: "
            f"€{monthly['debit'].mean():,.2f}\n"
            f"Average Monthly Balance: "
            f"€{monthly['Balance'].mean():,.2f}\n\n"
            f"Median Monthly Income: "
            f"€{monthly['credit'].median():,.2f}\n"
            f"Median Monthly Expenses: "
            f"€{monthly['debit'].median():,.2f}\n"
            f"Median Monthly Balance: "
            f"€{monthly['Balance'].median():,.2f}\n\n"
            f"Months with Surplus: "
            f"{(monthly['Balance'] > 0).sum()}\n"
            f"Months with Deficit: "
            f"{(monthly['Balance'] < 0).sum()}\n"
            f"Balanced Months: "
            f"{(monthly['Balance'] == 0).sum()}"
        )

    return (
        "Query not recognized.\n"
        "Available: monthly totals, top 3, average monthly savings, "
        "overall summary, expense categories, income categories, "
        "deficit months, annual summary."
    )


# ============================================================
# TOOL 2 — DEFICIT EXPENSE ANALYSIS
# ============================================================

@tool
def analyze_deficit_expenses(query: str, rolling_months: int = 24) -> str:
    """
    Detailed expense analysis for months with negative balances.

    Use this tool when the user asks which expenses
    contributed to spending during deficit months.

    Identifies deficit months independently and reports:
    - Monthly income, expenses and negative balance
    - Top 5 expense items in each deficit month
    - Category and description
    - Amount and percentage of monthly expenses

    Examples:
    - "analyze the main expenses in deficit months"
    - "top 5 expenses in deficit months"
    - "main expenses in deficit months"
    """

    q = _finance_normalize(query)

    if not any(x in q for x in [
        'deficit',
        'negative months',
        'negative balance',
        'deficit drivers',
        'deficit analysis',
        'expenses in deficit months',
        'largest expenses during deficits',
        'spending drivers',
        'negative balance causes',
        'monthly shortfall',
    ]):
        return (
            "Invalid request. "
            "Specify the expenses in deficit months."
        )

    data = get_rolling_data(rolling_months)

    # --------------------------------------------------------
    # 1. IDENTIFY DEFICIT MONTHS
    # --------------------------------------------------------

    monthly = (
        data.groupby(["Month", "Transaction Type"])["Amount"]
        .sum()
        .unstack(fill_value=0)
        .sort_index()
    )

    for col in ["credit", "debit"]:
        if col not in monthly.columns:
            monthly[col] = 0.0

    monthly["Balance"] = monthly["credit"] - monthly["debit"]

    negative = monthly[
        monthly["Balance"] < 0
    ]

    if negative.empty:
        return "There are no deficit months."

    # --------------------------------------------------------
    # 2. SELECT EXPENSES IN DEFICIT MONTHS
    # --------------------------------------------------------

    negative_expenses = data[
        data["Month"].isin(negative.index)
        & (data["Transaction Type"] == "debit")
    ]

    results = [
        "DEFICIT MONTHS — DETAILED EXPENSE ANALYSIS",
        f"Number of Deficit Months: {len(negative)}"
    ]

    # --------------------------------------------------------
    # 3. TOP 5 EXPENSE ITEMS PER DEFICIT MONTH
    # --------------------------------------------------------

    for month, row in negative.iterrows():

        month_data = negative_expenses[
            negative_expenses["Month"] == month
        ]

        top = (
            month_data.groupby(
                ["Category", "Description"]
            )["Amount"]
            .sum()
            .sort_values(ascending=False)
            .head(5)
        )

        results.append(
            f"\n{'=' * 60}\n"
            f"MONTH: {month:%Y-%m}\n"
            f"Income: €{row['credit']:,.2f}\n"
            f"Expenses: €{row['debit']:,.2f}\n"
            f"Deficit: €{row['Balance']:,.2f}\n"
            f"{'-' * 60}\n"
            "TOP 5 EXPENSES"
        )

        for rank, ((category, description), amount) in enumerate(
            top.items(), start=1
        ):

            weight = (
                amount / row["debit"] * 100
                if row["debit"] > 0 else 0
            )

            results.append(
                f"{rank}. {category} — {description}: "
                f"€{amount:,.2f} "
                f"({weight:.1f}% of monthly expenses)"
            )

    return "\n".join(results)


# ============================================================
# TOOL 3 — COMPARATIVE FINANCIAL ANALYSIS
# ============================================================

@tool
def compare_income_expenses(query: str, rolling_months: int = 24) -> str:
    """
    Comparative analysis of income, expenses and balances.

    Use this tool for:
    - Income versus expenses
    - Monthly financial comparisons
    - Annual financial comparisons
    - Comparison between two specified months
    - Expense-to-income ratios
    - Savings rates
    - Financial balance variations

    Examples:
    - "overall comparison"
    - "monthly comparison"
    - "annual comparison"
    - "expense-to-income ratio"
    - "compare 2026-01 and 2026-02"
    """

    q = _finance_normalize(query)
    data = get_rolling_data(rolling_months)

    # --------------------------------------------------------
    # 1. PREPARE FINANCIAL DATA
    # --------------------------------------------------------

    monthly = (
        data.groupby(["Month", "Transaction Type"])["Amount"]
        .sum()
        .unstack(fill_value=0)
        .sort_index()
    )

    for col in ["credit", "debit"]:
        if col not in monthly.columns:
            monthly[col] = 0.0

    if monthly.empty:
        return "No transactions available."

    monthly["Balance"] = monthly["credit"] - monthly["debit"]

    monthly["Expense_Income_%"] = (
        monthly["debit"]
        / monthly["credit"].replace(0, np.nan)
        * 100
    )

    monthly["Savings_Rate_%"] = (
        monthly["Balance"]
        / monthly["credit"].replace(0, np.nan)
        * 100
    )

    monthly["Status"] = monthly["Balance"].apply(
        lambda x: "SURPLUS" if x > 0
        else "DEFICIT" if x < 0
        else "BALANCED"
    )

    total_income = monthly["credit"].sum()
    total_expenses = monthly["debit"].sum()
    net_balance = total_income - total_expenses

    expense_income_ratio = (
        total_expenses / total_income * 100
        if total_income > 0 else np.nan
    )

    savings_rate = (
        net_balance / total_income * 100
        if total_income > 0 else np.nan
    )

    # --------------------------------------------------------
    # 2. COMPARE TWO SPECIFIC MONTHS
    # --------------------------------------------------------

    months_found = re.findall(
        r"\b\d{4}-\d{2}\b",
        query
    )

    if _finance_has(q, "compare", "comparison", "comparing") and len(months_found) >= 2:

        selected = pd.to_datetime(
            months_found[:2],
            format="%Y-%m",
            errors="coerce"
        )

        if selected.isna().any():
            return "Invalid month format."

        missing = [
            m.strftime("%Y-%m")
            for m in selected
            if m not in monthly.index
        ]

        if missing:
            return (
                "Months not available: "
                + ", ".join(missing)
            )

        columns = [
            "credit",
            "debit",
            "Balance",
            "Expense_Income_%",
            "Savings_Rate_%"
        ]

        comparison = monthly.loc[selected, columns]
        variation = comparison.iloc[1] - comparison.iloc[0]

        return (
            "MONTHLY FINANCIAL COMPARISON\n\n"
            + comparison.to_string(
                float_format=lambda x: f"{x:,.2f}"
            )
            + "\n\nVARIATION (SECOND MONTH - FIRST MONTH)\n\n"
            + variation.to_string(
                float_format=lambda x: f"{x:,.2f}"
            )
        )

    # --------------------------------------------------------
    # 3. ANNUAL COMPARISON
    # --------------------------------------------------------

    elif _finance_has(q, "annual", "yearly") or _finance_any(q, "year over year"):

        annual = monthly[["credit", "debit"]].copy()
        annual["Year"] = annual.index.year

        annual = annual.groupby("Year")[
            ["credit", "debit"]
        ].sum()

        annual["Balance"] = (
            annual["credit"] - annual["debit"]
        )

        annual["Expense_Income_%"] = (
            annual["debit"]
            / annual["credit"].replace(0, np.nan)
            * 100
        )

        annual["Savings_Rate_%"] = (
            annual["Balance"]
            / annual["credit"].replace(0, np.nan)
            * 100
        )

        return (
            "ANNUAL FINANCIAL COMPARISON\n"
            "(Available months only)\n\n"
            + annual.to_string(
                float_format=lambda x: f"{x:,.2f}"
            )
        )

    # --------------------------------------------------------
    # 4. MONTHLY COMPARISON AND FINANCIAL RATIOS
    # --------------------------------------------------------

    elif any(_finance_normalize(x) in q for x in [
        'monthly comparison',
        'monthly income versus expenses',
        'monthly income vs expenses',
        'month-by-month comparison',
        'monthly cash flow comparison',
        'expense-to-income ratio',
        'expenses to income',
        'spending-to-income ratio',
        'expense income percentage',
        'expenses divided by income',
        'spending ratio',
        'savings rate',
        'saving rate',
        'savings percentage',
        'savings-to-income ratio',
        'percentage of income saved',
        'monthly savings rate',
    ]):

        return (
            "MONTHLY INCOME VS EXPENSES\n\n"
            + monthly.to_string(
                float_format=lambda x: f"{x:,.2f}"
            )
            + "\n\nOVERALL RATIOS\n"
            + f"Expense-to-Income Ratio: "
            f"{expense_income_ratio:.2f}%\n"
            + f"Savings Rate: {savings_rate:.2f}%"
        )

    # --------------------------------------------------------
    # 5. OVERALL COMPARISON
    # --------------------------------------------------------

    elif any(x in q for x in [
        'overall comparison',
        'total income versus expenses',
        'aggregate income vs expenses',
        'overall income expense balance',
        'cumulative income expenses',
        'overall financial comparison',
    ]):

        return (
            "INCOME VS EXPENSES — OVERALL COMPARISON\n\n"
            f"Total Income: €{total_income:,.2f}\n"
            f"Total Expenses: €{total_expenses:,.2f}\n"
            f"Net Balance: €{net_balance:,.2f}\n\n"
            f"Expense-to-Income Ratio: "
            f"{expense_income_ratio:.2f}%\n"
            f"Savings Rate: {savings_rate:.2f}%\n\n"
            f"Average Monthly Balance: "
            f"€{monthly['Balance'].mean():,.2f}\n"
            f"Months with Surplus: "
            f"{(monthly['Balance'] > 0).sum()}\n"
            f"Months with Deficit: "
            f"{(monthly['Balance'] < 0).sum()}"
        )

    return (
        "Query not recognized.\n"
        "Available: overall comparison, monthly comparison, "
        "annual comparison, expense-to-income ratio, savings rate, "
        "compare 2026-01 and 2026-02."
    )


# ============================================================
# TOOL REGISTRY — PERSONAL FINANCE
# ============================================================
# Use this list when registering the financial tools
# in the LangGraph workflow.
# ============================================================

personal_finance_tools = [
    analyze_finance,
    analyze_deficit_expenses,
    compare_income_expenses
]
```

```python
# ============================================================
# BOX 4
#
# PERSONAL FINANCE — FINANCIAL CASH FLOW ANALYSIS
#
# We introduce indicators to assess financial stability:
# balance variability, savings regularity, and the
# frequency of months with negative balances.
#
# Objective:
# - Income stability
# - Expense variability
# - Monthly balance distribution
# - Cash flow volatility
# - Surplus and deficit frequency
# - Financial cash flow stability indicators
#
# Historical analysis only — no future projections.
# ============================================================

def financial_cash_flow_analysis(rolling_months: int = 24):

    # --------------------------------------------------------
    # 1. RETRIEVE HISTORICAL FINANCIAL DATA
    # --------------------------------------------------------

    data = get_rolling_data(rolling_months)

    monthly_cf = (
        data.groupby(["Month", "Transaction Type"])["Amount"]
        .sum()
        .unstack(fill_value=0)
        .sort_index()
    )

    for col in ["credit", "debit"]:
        if col not in monthly_cf.columns:
            monthly_cf[col] = 0.0

    if monthly_cf.empty:
        raise ValueError("No financial data available.")

    # Include months with no transactions
    full_index = pd.date_range(
        start=monthly_cf.index.min(),
        end=monthly_cf.index.max(),
        freq="MS"
    )

    monthly_cf = monthly_cf.reindex(
        full_index, fill_value=0.0
    )

    monthly_cf["Balance"] = (
        monthly_cf["credit"] - monthly_cf["debit"]
    )

    monthly_cf["Savings_Rate_%"] = np.where(
        monthly_cf["credit"] > 0,
        monthly_cf["Balance"] / monthly_cf["credit"] * 100,
        np.nan
    )

    # --------------------------------------------------------
    # 2. DESCRIPTIVE STATISTICS
    # --------------------------------------------------------

    variables = {
        "Income": monthly_cf["credit"],
        "Expenses": monthly_cf["debit"],
        "Balance": monthly_cf["Balance"]
    }

    statistics = pd.DataFrame({
        name: {
            "Mean_EUR": series.mean(),
            "Median_EUR": series.median(),
            "Std_Dev_EUR": series.std(ddof=1),
            "Minimum_EUR": series.min(),
            "Maximum_EUR": series.max(),
            "Range_EUR": series.max() - series.min(),
            "CV_%": (
                series.std(ddof=1) / series.mean() * 100
                if series.mean() > 0 else np.nan
            )
        }
        for name, series in variables.items()
    }).T

    # --------------------------------------------------------
    # 3. CASH FLOW STABILITY INDICATORS
    # --------------------------------------------------------

    balance = monthly_cf["Balance"]
    income = monthly_cf["credit"]
    expenses = monthly_cf["debit"]

    n_months = len(monthly_cf)

    surplus_months = int((balance > 0).sum())
    deficit_months = int((balance < 0).sum())
    balanced_months = int((balance == 0).sum())

    indicators = pd.Series({
        "Months_Analyzed": n_months,
        "Surplus_Months": surplus_months,
        "Deficit_Months": deficit_months,
        "Balanced_Months": balanced_months,
        "Surplus_Frequency_%": (
            surplus_months / n_months * 100
        ),
        "Deficit_Frequency_%": (
            deficit_months / n_months * 100
        ),
        "Average_Monthly_Balance_EUR": balance.mean(),
        "Median_Monthly_Balance_EUR": balance.median(),
        "Cash_Flow_Volatility_EUR": balance.std(ddof=1),
        "Average_Savings_Rate_%": (
            balance.sum() / income.sum() * 100
            if income.sum() > 0 else np.nan
        ),
        "Income_Expense_Correlation": (
            income.corr(expenses)
            if income.std() > 0 and expenses.std() > 0
            else np.nan
        )
    })

    # --------------------------------------------------------
    # 4. DISPLAY RESULTS
    # --------------------------------------------------------

    print("=" * 75)
    print("FINANCIAL CASH FLOW ANALYSIS")
    print("=" * 75)

    print(
        f"\nPeriod: {monthly_cf.index.min():%Y-%m} "
        f"to {monthly_cf.index.max():%Y-%m}"
    )

    print("\nDESCRIPTIVE STATISTICS\n")
    print(statistics.round(2).to_string())

    print("\nCASH FLOW STABILITY INDICATORS\n")
    print(indicators.round(2).to_string())

    print("\nMONTHLY CASH FLOW\n")
    print(
        monthly_cf.round(2).to_string()
    )

    # --------------------------------------------------------
    # 5. RETURN STRUCTURED RESULTS
    # --------------------------------------------------------

    return {
        "monthly_cash_flow": monthly_cf,
        "statistics": statistics,
        "indicators": indicators
    }


# ============================================================
# EXECUTION
# ============================================================

pf1_results = financial_cash_flow_analysis(
    rolling_months=24
)
```

```python
# ============================================================
# BOX 5
#
# PERSONAL FINANCE — SAVINGS CAPACITY & ACCUMULATION
#
# Starting from observed savings capacity, we calculate
# accumulation scenarios, distinguishing monthly savings
# from initial liquidity.
#
# Objective:
# - Estimate historical monthly savings capacity
# - Compare mean and median monthly savings
# - Measure positive and negative cash flow
# - Project savings accumulation over 12, 24, 36, 60 months
# - Compare baseline and conservative scenarios
#
# Deterministic projections based on historical cash flow.
# No investment returns or Monte Carlo simulations.
# ============================================================

def savings_capacity_analysis(
    pf1_results,
    horizons=(12, 24, 36, 60),
    initial_savings=0.0
):

    # --------------------------------------------------------
    # 1. RETRIEVE PF-1 RESULTS
    # --------------------------------------------------------

    monthly_cf = pf1_results["monthly_cash_flow"].copy()

    if monthly_cf.empty:
        raise ValueError("No monthly cash flow data available.")

    balance = monthly_cf["Balance"].astype(float)
    income = monthly_cf["credit"].astype(float)

    if not np.isfinite(initial_savings) or initial_savings < 0:
        raise ValueError("initial_savings must be non-negative.")

    # --------------------------------------------------------
    # 2. HISTORICAL SAVINGS CAPACITY
    # --------------------------------------------------------

    mean_savings = balance.mean()
    median_savings = balance.median()

    positive_savings = balance.clip(lower=0)
    monthly_deficits = (-balance.clip(upper=0))

    average_positive_savings = positive_savings.mean()
    average_monthly_deficit = monthly_deficits.mean()

    net_savings = balance.sum()

    savings_rate = (
        net_savings / income.sum() * 100
        if income.sum() > 0 else np.nan
    )

    # Baseline: historical mean monthly balance.
    # Conservative: lower of mean and median.
    # Both preserve negative savings capacity where present.

    baseline_monthly = float(mean_savings)
    conservative_monthly = float(
        min(mean_savings, median_savings)
    )

    capacity = pd.Series({
        "Months_Analyzed": len(balance),
        "Mean_Monthly_Savings_EUR": mean_savings,
        "Median_Monthly_Savings_EUR": median_savings,
        "Average_Positive_Savings_EUR": average_positive_savings,
        "Average_Monthly_Deficit_EUR": average_monthly_deficit,
        "Total_Net_Savings_EUR": net_savings,
        "Historical_Savings_Rate_%": savings_rate,
        "Baseline_Monthly_EUR": baseline_monthly,
        "Conservative_Monthly_EUR": conservative_monthly
    })

    # --------------------------------------------------------
    # 3. SAVINGS ACCUMULATION BY HORIZON
    # --------------------------------------------------------

    projections = pd.DataFrame([
        {
            "Horizon_Months": int(h),
            "Baseline_EUR": (
                initial_savings + baseline_monthly * h
            ),
            "Conservative_EUR": (
                initial_savings + conservative_monthly * h
            )
        }
        for h in horizons
    ])

    if any(int(h) <= 0 or int(h) != h for h in horizons):
        raise ValueError("Horizons must be positive whole months.")

    # --------------------------------------------------------
    # 4. MONTHLY ACCUMULATION PATHS
    # --------------------------------------------------------

    max_horizon = int(max(horizons))
    months = np.arange(max_horizon + 1)

    accumulation_paths = pd.DataFrame({
        "Month": months,
        "Baseline_EUR": (
            initial_savings + baseline_monthly * months
        ),
        "Conservative_EUR": (
            initial_savings + conservative_monthly * months
        )
    })

    # --------------------------------------------------------
    # 5. DISPLAY RESULTS
    # --------------------------------------------------------

    print("=" * 75)
    print("SAVINGS CAPACITY & ACCUMULATION")
    print("=" * 75)

    print("\nHISTORICAL SAVINGS CAPACITY\n")
    print(capacity.round(2).to_string())

    print("\nSAVINGS ACCUMULATION PROJECTIONS\n")
    print(
        projections.round(2).to_string(index=False)
    )

    # --------------------------------------------------------
    # 6. RETURN STRUCTURED RESULTS
    # --------------------------------------------------------

    return {
        "savings_capacity": capacity,
        "projections": projections,
        "accumulation_paths": accumulation_paths,
        "initial_savings": initial_savings
    }


# ============================================================
# EXECUTION
# ============================================================

pf2_results = savings_capacity_analysis(
    pf1_results=pf1_results,
    horizons=(12, 24, 36, 60),
    initial_savings=0.0
)
```

```python
# ============================================================
# BOX 6
#
# PERSONAL FINANCE — FINANCIAL RESILIENCE
#
# To calculate the months of financial autonomy, we need
# to know the available liquidity and distinguish essential
# expenses from other expenses. If this information is not
# included in the dataset, it must be provided as input:
# we will not derive it from arbitrary assumptions.
#
# Objective:
# - Estimate essential monthly expenses
# - Calculate emergency fund coverage
# - Measure financial autonomy without income
# - Assess capacity to absorb financial shocks
# - Compare current liquidity with emergency fund targets
#
# Deterministic resilience analysis.
# Monte Carlo simulations will be developed in BOX PF-4.
#
# METHODOLOGICAL ASSUMPTIONS:
#
# Available liquidity is assumed to be equal to the net
# savings accumulated over the 12 months analyzed:
#
# Available Liquidity = Total Income - Total Expenses
#
# The analysis does not consider financial positions,
# debts, assets, or savings accumulated before the
# observation period, since these are not available
# in the historical dataset.
#
# Therefore, available liquidity represents the net
# savings accumulated within the observed period.
#
# Essential expenses are identified directly from
# the CSV transaction categories:
#
# - Financial services
# - Groceries
# - Healthcare
# - Housing
# - Taxes
# - Utilities
#
# Essential Expense Ratio =
# Total Essential Expenses / Total Expenses
#
# No assumed or arbitrary expense percentage is used.
# ============================================================

def financial_resilience_analysis(
    pf1_results,
    pf2_results,
    available_liquidity=None,
    essential_expense_ratio=None,
    emergency_months=(3, 6, 12),
    shock_amounts=(1000, 3000, 5000)
):

    # --------------------------------------------------------
    # 1. RETRIEVE HISTORICAL FINANCIAL DATA
    # --------------------------------------------------------

    monthly_cf = pf1_results["monthly_cash_flow"].copy()

    if monthly_cf.empty:
        raise ValueError("No monthly cash flow data available.")

    income = monthly_cf["credit"].astype(float)
    expenses = monthly_cf["debit"].astype(float)
    balance = monthly_cf["Balance"].astype(float)

    if not np.isfinite(
        monthly_cf[["credit", "debit", "Balance"]].to_numpy()
    ).all():
        raise ValueError("Financial data contains invalid values.")

    # --------------------------------------------------------
    # 2. VALIDATE USER-DEFINED PARAMETERS
    # --------------------------------------------------------

    # Available liquidity is defined as net accumulated
    # savings over the historical observation period.
    #
    # No assumptions are made about previous debts,
    # assets, or savings outside the analyzed 12 months.

    available_liquidity = float(balance.sum())

    # Essential expenses are calculated from actual
    # transaction categories in the existing dataset.

    # Essential expense categories explicitly selected from the CSV:
    # Financial services, Groceries, Healthcare, Housing, Taxes,
    # and Utilities. Only debit transactions in these categories
    # contribute to essential expenses. The exact essential expense
    # ratio is the sum of their Amount values divided by the sum
    # of Amount across all debit transactions in the 24-month period.
    # Average monthly essential expenses = average monthly total
    # expenses multiplied by this data-derived ratio.

    essential_categories = {
        "Financial services",
        "Groceries",
        "Healthcare",
        "Housing",
        "Taxes",
        "Utilities"
    }

    transaction_data = get_rolling_data(24).copy()

    transaction_data["Amount"] = pd.to_numeric(
        transaction_data["Amount"],
        errors="raise"
    )

    expense_data = transaction_data[
        transaction_data["Transaction Type"]
        .astype(str).str.strip().str.lower() == "debit"
    ].copy()

    expense_data["Category"] = (
        expense_data["Category"].astype(str).str.strip()
    )

    total_expenses = expense_data["Amount"].sum()

    essential_total = expense_data.loc[
        expense_data["Category"].isin(essential_categories),
        "Amount"
    ].sum()

    essential_expense_ratio = (
        float(essential_total / total_expenses)
        if total_expenses > 0
        else 0.0
    )

    if available_liquidity is not None:
        available_liquidity = float(available_liquidity)

        if (
            not np.isfinite(available_liquidity)
            or available_liquidity < 0
        ):
            raise ValueError(
                "available_liquidity must be non-negative."
            )

    if essential_expense_ratio is not None:
        essential_expense_ratio = float(
            essential_expense_ratio
        )

        if (
            not np.isfinite(essential_expense_ratio)
            or not 0 <= essential_expense_ratio <= 1
        ):
            raise ValueError(
                "essential_expense_ratio must be between 0 and 1."
            )

    emergency_months = tuple(emergency_months)
    shock_amounts = tuple(shock_amounts)

    if any(
        not isinstance(m, (int, np.integer)) or m <= 0
        for m in emergency_months
    ):
        raise ValueError(
            "Emergency fund horizons must be positive integers."
        )

    if any(
        not np.isfinite(s) or s < 0
        for s in shock_amounts
    ):
        raise ValueError(
            "Shock amounts must be non-negative."
        )

    # --------------------------------------------------------
    # 3. HISTORICAL FINANCIAL INDICATORS
    # --------------------------------------------------------

    average_income = income.mean()
    average_expenses = expenses.mean()
    average_balance = balance.mean()

    median_expenses = expenses.median()
    expense_volatility = expenses.std(ddof=1)

    # Essential expenses are calculated using the exact
    # proportion derived from the selected CSV categories.

    essential_expenses = (
        average_expenses * essential_expense_ratio
        if essential_expense_ratio is not None
        else np.nan
    )

    # --------------------------------------------------------
    # 4. FINANCIAL AUTONOMY
    # --------------------------------------------------------

    # Full-expense coverage:
    # Number of months current liquidity can cover average
    # monthly expenses, assuming zero future income.

    if available_liquidity is not None:
        full_expense_coverage = (
            available_liquidity / average_expenses
            if average_expenses > 0
            else np.inf
        )
    else:
        full_expense_coverage = np.nan

    # Essential-expense coverage:
    # Number of months liquidity can cover essential expenses.

    if (
        available_liquidity is not None
        and essential_expense_ratio is not None
    ):
        essential_coverage = (
            available_liquidity / essential_expenses
            if essential_expenses > 0
            else np.inf
        )
    else:
        essential_coverage = np.nan

    # --------------------------------------------------------
    # 5. EMERGENCY FUND TARGETS
    # --------------------------------------------------------

    emergency_fund = pd.DataFrame([
        {
            "Coverage_Months": months,
            "Full_Expenses_Target_EUR": (
                average_expenses * months
            ),
            "Essential_Expenses_Target_EUR": (
                essential_expenses * months
                if essential_expense_ratio is not None
                else np.nan
            ),
            "Full_Expenses_Funding_Gap_EUR": (
                max(
                    0,
                    average_expenses * months
                    - available_liquidity
                )
                if available_liquidity is not None
                else np.nan
            ),
            "Essential_Funding_Gap_EUR": (
                max(
                    0,
                    essential_expenses * months
                    - available_liquidity
                )
                if (
                    available_liquidity is not None
                    and essential_expense_ratio is not None
                )
                else np.nan
            )
        }
        for months in emergency_months
    ])

    # --------------------------------------------------------
    # 6. FINANCIAL SHOCK ABSORPTION
    # --------------------------------------------------------

    shock_analysis = pd.DataFrame([
        {
            "Shock_EUR": shock,
            "Liquidity_After_Shock_EUR": (
                available_liquidity - shock
                if available_liquidity is not None
                else np.nan
            ),
            "Unfunded_Shock_EUR": (
                max(0, shock - available_liquidity)
                if available_liquidity is not None
                else np.nan
            ),
            "Full_Expense_Coverage_After_Shock": (
                max(0, available_liquidity - shock)
                / average_expenses
                if (
                    available_liquidity is not None
                    and average_expenses > 0
                )
                else np.nan
            ),
            "Essential_Coverage_After_Shock": (
                max(0, available_liquidity - shock)
                / essential_expenses
                if (
                    available_liquidity is not None
                    and essential_expense_ratio is not None
                    and essential_expenses > 0
                )
                else np.nan
            )
        }
        for shock in shock_amounts
    ])

    # --------------------------------------------------------
    # 7. RESILIENCE INDICATORS
    # --------------------------------------------------------

    resilience = pd.Series({
        "Months_Analyzed": len(monthly_cf),
        "Average_Monthly_Income_EUR": average_income,
        "Average_Monthly_Expenses_EUR": average_expenses,
        "Median_Monthly_Expenses_EUR": median_expenses,
        "Expense_Volatility_EUR": expense_volatility,
        "Average_Monthly_Balance_EUR": average_balance,
        "Available_Liquidity_EUR": (
            available_liquidity
            if available_liquidity is not None
            else np.nan
        ),
        "Essential_Expense_Ratio_%": (
            essential_expense_ratio * 100
            if essential_expense_ratio is not None
            else np.nan
        ),
        "Estimated_Essential_Expenses_EUR": (
            essential_expenses
        ),
        "Full_Expense_Coverage_Months": (
            full_expense_coverage
        ),
        "Essential_Expense_Coverage_Months": (
            essential_coverage
        )
    })

    # --------------------------------------------------------
    # 8. DISPLAY RESULTS
    # --------------------------------------------------------

    print("=" * 75)
    print("FINANCIAL RESILIENCE ANALYSIS")
    print("=" * 75)

    print("\nFINANCIAL RESILIENCE INDICATORS\n")
    print(resilience.round(2).to_string())

    print("\nEMERGENCY FUND TARGETS\n")
    print(
        emergency_fund.round(2).to_string(index=False)
    )

    print("\nFINANCIAL SHOCK ANALYSIS\n")
    print(
        shock_analysis.round(2).to_string(index=False)
    )

    # --------------------------------------------------------
    # 9. VISUALIZATION
    # --------------------------------------------------------

    if available_liquidity is not None:

        plt.figure(figsize=(9, 5))

        plt.bar(
            emergency_fund["Coverage_Months"].astype(str),
            emergency_fund["Full_Expenses_Target_EUR"],
            label="Full Expenses Target"
        )

        if essential_expense_ratio is not None:
            plt.plot(
                emergency_fund["Coverage_Months"].astype(str),
                emergency_fund["Essential_Expenses_Target_EUR"],
                marker="o",
                label="Essential Expenses Target"
            )

        plt.axhline(
            available_liquidity,
            color="gray",
            linestyle="--",
            label="Available Liquidity"
        )

        plt.title("Emergency Fund Coverage")
        plt.xlabel("Coverage Horizon (Months)")
        plt.ylabel("EUR")
        plt.legend()
        plt.grid(axis="y", alpha=0.3)
        plt.tight_layout()
        plt.show()

    # --------------------------------------------------------
    # 10. RETURN STRUCTURED RESULTS
    # --------------------------------------------------------

    # --------------------------------------------------------
    # 11. METRICS LEGEND
    # --------------------------------------------------------

    metrics_legend = {
        "FINANCIAL RESILIENCE INDICATORS": {
            "Months_Analyzed": (
                "Number of historical months included in the analysis.",
                "How many months of financial data am I analyzing?"
            ),
            "Average_Monthly_Income_EUR": (
                "Average monthly income over the analyzed period.",
                "How much income do I receive on average each month?"
            ),
            "Average_Monthly_Expenses_EUR": (
                "Average monthly amount of all expenses incurred.",
                "How much do I spend on average each month?"
            ),
            "Median_Monthly_Expenses_EUR": (
                "Median monthly expenses: half of the months "
                "have expenses less than or equal to this value.",
                "What is the median level of my monthly expenses?"
            ),
            "Expense_Volatility_EUR": (
                "Standard deviation of monthly expenses, in euros.",
                "How much do my expenses vary from month to month?"
            ),
            "Average_Monthly_Balance_EUR": (
                "Average monthly difference between income and expenses.",
                "What is my average monthly savings?"
            ),
            "Available_Liquidity_EUR": (
                "Net savings accumulated over the 12 months analyzed, "
                "assumed to represent available liquidity.",
                "How much net savings have I accumulated over the period?"
            ),
            "Essential_Expense_Ratio_%": (
                "Percentage of total expenses belonging "
                "to the six selected essential categories.",
                "What percentage of my expenses is essential?"
            ),
            "Estimated_Essential_Expenses_EUR": (
                "Average monthly essential expenses calculated from "
                "the selected categories in the CSV.",
                "How much do I spend on average each month on essential needs?"
            ),
            "Full_Expense_Coverage_Months": (
                "Number of months accumulated savings "
                "can cover all average expenses, without new income.",
                "For how many months can I cover all my expenses "
                "without receiving income?"
            ),
            "Essential_Expense_Coverage_Months": (
                "Number of months accumulated savings "
                "can cover essential expenses only.",
                "For how many months can I cover essential "
                "expenses only, without income?"
            )
        },

        "EMERGENCY FUND TARGETS": {
            "Coverage_Months": (
                "Number of months of financial autonomy considered: 3, 6 or 12.",
                "For how many months do I want to assess my financial autonomy?"
            ),
            "Full_Expenses_Target_EUR": (
                "Capital required to cover all average monthly "
                "expenses over the selected period.",
                "How much money do I need to cover all my expenses "
                "for 3, 6 or 12 months without income?"
            ),
            "Essential_Expenses_Target_EUR": (
                "Capital required to cover essential expenses "
                "only over the selected period.",
                "How much money do I need to cover essential "
                "expenses only for 3, 6 or 12 months?"
            ),
            "Full_Expenses_Funding_Gap_EUR": (
                "Positive difference between the capital required "
                "to cover all expenses and available liquidity. "
                "Equals zero when sufficient capital is available.",
                "How much additional savings do I need to cover "
                "all expenses over the selected period?"
            ),
            "Essential_Funding_Gap_EUR": (
                "Positive difference between the capital required "
                "to cover essential expenses and available liquidity. "
                "Equals zero when sufficient capital is available.",
                "How much additional savings do I need to cover "
                "essential expenses only?"
            )
        },

        "FINANCIAL SHOCK ANALYSIS": {
            "Shock_EUR": (
                "Amount of the simulated unexpected expense.",
                "What is the amount of the unexpected expense?"
            ),
            "Liquidity_After_Shock_EUR": (
                "Available liquidity after deducting "
                "the full shock amount. May be negative.",
                "How much liquidity remains after the unexpected expense?"
            ),
            "Unfunded_Shock_EUR": (
                "Portion of the unexpected expense that exceeds "
                "available liquidity.",
                "How much of the unexpected expense can I not cover?"
            ),
            "Full_Expense_Coverage_After_Shock": (
                "Months of remaining financial autonomy after the shock, "
                "considering all average expenses and zero income.",
                "For how many months can I cover all my "
                "expenses after the unexpected expense?"
            ),
            "Essential_Coverage_After_Shock": (
                "Months of remaining financial autonomy after the shock, "
                "considering essential expenses only and zero income.",
                "For how many months can I cover essential "
                "expenses only after the unexpected expense?"
            )
        }
    }

    for section, metrics in metrics_legend.items():

        print("\n" + "=" * 75)
        print(section + " — METRICS LEGEND")
        print("=" * 75)

        for metric, (explanation, question) in metrics.items():
            print(f"\n{metric}")
            print(f"Explanation: {explanation}")
            print(f"Question: {question}")

    return {
        "resilience_indicators": resilience,
        "emergency_fund": emergency_fund,
        "shock_analysis": shock_analysis
    }


# ============================================================
# EXECUTION
# ============================================================

pf3_results = financial_resilience_analysis(
    pf1_results=pf1_results,
    pf2_results=pf2_results,
    available_liquidity=None,
    essential_expense_ratio=None,
    emergency_months=(3, 6, 12),
    shock_amounts=(1000, 3000, 5000)
)
```

```python
# ============================================================
# BOX 7 — DETERMINISTIC FINANCIAL STRESS TESTING
# ============================================================
# PURPOSE
# Evaluate the impact of adverse financial events on household
# liquidity and financial resilience over a 24-month horizon.
# Each scenario is compared with a baseline based on observed
# monthly income and expenses.
#
# FOUR INDEPENDENT STRESS SCENARIOS:
# 1. EMPLOYMENT INCOME REDUCTION: 30% reduction for 6 months,
#    applied to average employment income, including bonuses.
# 2. EXTRAORDINARY EXPENSE: EUR 10,000 in month 1.
# 3. EXTRAORDINARY EXPENSE: EUR 20,000 in month 1.
# 4. RECURRING EXPENSE INCREASE: 15% increase for 12 months.
#
# METHODOLOGY AND ASSUMPTIONS
# - Historical baseline: average monthly income and expenses
#   calculated from the observed financial data.
# - Essential reserve: three months of average essential expenses.
# - Deterministic projections over a maximum of 24 months,
#   assuming constant baseline income and expenses.
# - Recovery: return to initial liquidity following the shock,
#   assessed separately from the no-stress baseline.
#
# OUTPUTS
# - Comparative assessment of the four stress scenarios.
# - Projected liquidity and essential reserve coverage.
# - Financial shortfalls and recovery periods.
# - Scenario-specific conclusions and comparative visualization.
# - Detailed monthly trajectories stored in pf4_results.
# ============================================================


# ----------------------- EDITABLE PARAMETERS -----------------------
# INITIAL RULE: after 12 months, accumulated net savings are frozen.
# At the end of the observation period, net savings total EUR 6,264.52.
# 50% (EUR 3,132.26) is allocated to the investment portfolio,
# while the remaining 50% (EUR 3,132.26) is retained as personal
# liquidity. From the following month, the two amounts evolve separately.

PF4_HISTORICAL_MONTHS = 12
PF4_HORIZON_MONTHS = 24
PF4_INCOME_REDUCTION_PCT = 0.30
PF4_INCOME_REDUCTION_MONTHS = 6
PF4_EXTRAORDINARY_EXPENSE_EUR = 10_000.0
PF4_EXTRAORDINARY_EXPENSE_2_EUR = 20_000.0
PF4_EXTRAORDINARY_EXPENSE_MONTH = 1
PF4_EXPENSE_INCREASE_PCT = 0.15
PF4_EXPENSE_INCREASE_MONTHS = 12
PF4_EMPLOYMENT_CATEGORY = "Employment"  # Includes salary and both bonuses.
PF4_INITIAL_LIQUIDITY = None   # None = sum of historical net cash flows.
PF4_SHOW_CHARTS = True

PF4_ESSENTIAL_CATEGORIES = {
    'Financial services', 'Groceries', 'Healthcare',
    'Housing', 'Taxes', 'Utilities'
}


def pf4_prepare_data(rolling_months=24):
    d = get_rolling_data(rolling_months).copy()

    # Normalize original CSV column names when necessary.
    if "Amount" not in d.columns and "Amount_EUR" in d.columns:
        d["Amount"] = d["Amount_EUR"]

    if "Transaction Type" not in d.columns and "Type" in d.columns:
        d["Transaction Type"] = (
            d["Type"]
            .astype(str)
            .str.strip()
            .str.lower()
            .map({"income": "credit", "expense": "debit"})
        )

    required = {"Month", "Transaction Type", "Category", "Description", "Amount"}
    missing = required - set(d.columns)
    if missing:
        raise ValueError(f"Missing columns after normalization: {sorted(missing)}")

    d["Month"] = pd.to_datetime(d["Month"]).dt.to_period("M")
    d["Transaction Type"] = d["Transaction Type"].astype(str).str.strip().str.lower()
    d["Category"] = d["Category"].astype(str).str.strip()
    d["Description"] = d["Description"].astype(str).str.strip()
    d["Amount"] = pd.to_numeric(d["Amount"], errors="raise")

    if d.empty or (d["Amount"] < 0).any():
        raise ValueError("Expected nonempty data with nonnegative Amount values.")
    if not d["Transaction Type"].isin(["credit", "debit"]).all():
        raise ValueError("Transaction Type must be credit or debit.")

    months = pd.period_range(d["Month"].min(), d["Month"].max(), freq="M")
    return d, months

def pf4_income_category_audit(rolling_months=24):
    d, months = pf4_prepare_data(rolling_months)
    income = d[d['Transaction Type'].eq('credit')]
    report = (income.groupby(['Category', 'Description'], dropna=False)['Amount']
              .agg(transactions='size', total_eur='sum', mean_transaction_eur='mean')
              .reset_index().sort_values('total_eur', ascending=False))
    print('\nPF-4 — ACTUAL INCOME CATEGORIES AND DESCRIPTIONS')
    print(report.round(2).to_string(index=False))
    return report


def pf4_run_stress_tests(
    employment_category=PF4_EMPLOYMENT_CATEGORY,
    rolling_months=PF4_HISTORICAL_MONTHS,
    horizon_months=PF4_HORIZON_MONTHS,
    income_reduction_pct=PF4_INCOME_REDUCTION_PCT,
    income_reduction_months=PF4_INCOME_REDUCTION_MONTHS,
    extraordinary_expense_eur=PF4_EXTRAORDINARY_EXPENSE_EUR,
    extraordinary_expense_2_eur=PF4_EXTRAORDINARY_EXPENSE_2_EUR,
    extraordinary_expense_month=PF4_EXTRAORDINARY_EXPENSE_MONTH,
    expense_increase_pct=PF4_EXPENSE_INCREASE_PCT,
    expense_increase_months=PF4_EXPENSE_INCREASE_MONTHS,
    initial_liquidity=PF4_INITIAL_LIQUIDITY,
    show_charts=PF4_SHOW_CHARTS,
):
    if not isinstance(horizon_months, int) or not 1 <= horizon_months <= 24:
        raise ValueError('horizon_months must be an integer from 1 to 24.')
    for label, value in [('income_reduction_months', income_reduction_months),
                         ('expense_increase_months', expense_increase_months)]:
        if not isinstance(value, int) or not 0 <= value <= horizon_months:
            raise ValueError(f'{label} must be an integer from 0 to horizon_months.')
    if not isinstance(extraordinary_expense_month, int) or not 1 <= extraordinary_expense_month <= horizon_months:
        raise ValueError('extraordinary_expense_month must be inside the horizon.')
    if not 0 <= income_reduction_pct <= 1 or expense_increase_pct < 0 or extraordinary_expense_eur < 0 or extraordinary_expense_2_eur < 0:
        raise ValueError('Invalid stress parameters: reductions must be 0..1, increases and expense >= 0.')
    if not isinstance(employment_category, str) or not employment_category.strip():
        raise ValueError('employment_category must be a nonempty category name.')

    d, months = pf4_prepare_data(rolling_months)
    credits = d[d['Transaction Type'].eq('credit')]
    debits = d[d['Transaction Type'].eq('debit')]
    selected = credits[credits['Category'].eq(employment_category.strip())]
    if selected.empty:
        raise ValueError('Employment category not found in credit transactions. Check the audit.')

    def monthly_sum(frame):
        return frame.groupby('Month')['Amount'].sum().reindex(months, fill_value=0.0)

    income_m = monthly_sum(credits)
    expense_m = monthly_sum(debits)
    employment_m = monthly_sum(selected)
    income_avg = float(income_m.mean())
    expense_avg = float(expense_m.mean())
    employment_avg = float(employment_m.mean())
    other_income_avg = income_avg - employment_avg
    if initial_liquidity is None:
        initial_liquidity = float((income_m - expense_m).sum())
    initial_liquidity = float(initial_liquidity)
    if initial_liquidity < 0:
        raise ValueError('Initial liquidity must be >= 0. Supply a verified value if necessary.')
    essential_monthly = float(debits[debits['Category'].isin(PF4_ESSENTIAL_CATEGORIES)]['Amount'].sum() / len(months))
    reserve = 3.0 * essential_monthly
    normal_surplus = income_avg - expense_avg

    definitions = {
        f'S1 — Employment income -{income_reduction_pct:.0%} / {income_reduction_months}m': 'income_reduction',
        'S2 — Extraordinary expense EUR 10,000': 'extraordinary_expense',
        'S3 — Extraordinary expense EUR 20,000': 'extraordinary_expense_2',
        'S4 — Recurring expenses +15% / 12m': 'expense_increase',
    }
    stress_end = {
        'income_reduction': income_reduction_months,
        'extraordinary_expense': extraordinary_expense_month,
        'extraordinary_expense_2': extraordinary_expense_month,
        'expense_increase': expense_increase_months,
    }
    all_paths = {}
    rows = []
    for name, kind in definitions.items():
        cash = initial_liquidity
        baseline = initial_liquidity
        records = []
        for m in range(1, horizon_months + 1):
            income = income_avg
            expenses = expense_avg
            extraordinary = 0.0
            if kind == 'income_reduction' and m <= income_reduction_months:
                income = other_income_avg + employment_avg * (1.0 - income_reduction_pct)
            elif kind in ('extraordinary_expense', 'extraordinary_expense_2') and m == extraordinary_expense_month:
                extraordinary = (extraordinary_expense_eur if kind == 'extraordinary_expense'
                                 else extraordinary_expense_2_eur)
            elif kind == 'expense_increase' and m <= expense_increase_months:
                expenses *= (1.0 + expense_increase_pct)
            net = income - expenses - extraordinary
            cash += net
            baseline += normal_surplus
            records.append({
                'Month': m, 'Income_EUR': income, 'Ordinary_Expenses_EUR': expenses,
                'Extraordinary_Expense_EUR': extraordinary,
                'Net_Cash_Flow_EUR': net, 'Stress_Liquidity_EUR': cash,
                'No_Stress_Liquidity_EUR': baseline,
                'Gap_To_Zero_EUR': max(0.0, -cash),
                'Gap_To_Reserve_EUR': max(0.0, reserve - cash),
                'Below_Reserve': cash < reserve,
                'Negative_Liquidity': cash < 0,
                'Monthly_Deficit': net < 0,
            })
        path = pd.DataFrame(records)
        all_paths[name] = path
        # Recovery must follow the end of the imposed stress. No recovery
        # is claimed if the liquidity level is still below its initial value.
        eligible = path[(path['Month'] > stress_end[kind]) &
                        (path['Stress_Liquidity_EUR'] >= initial_liquidity)]
        recovery = int(eligible.iloc[0]['Month']) if not eligible.empty else None
        rows.append({
            'Scenario': name,
            'Minimum_Liquidity_EUR': min(initial_liquidity, float(path['Stress_Liquidity_EUR'].min())),
            'Max_Gap_To_Zero_EUR': float(path['Gap_To_Zero_EUR'].max()),
            'Max_Gap_To_Reserve_EUR': float(path['Gap_To_Reserve_EUR'].max()),
            'Months_Below_Reserve': int(path['Below_Reserve'].sum()),
            'Months_Negative_Liquidity': int(path['Negative_Liquidity'].sum()),
            'Months_With_Deficit': int(path['Monthly_Deficit'].sum()),
            'Final_Liquidity_EUR': float(path['Stress_Liquidity_EUR'].iloc[-1]),
            'Final_Gap_Vs_Baseline_EUR': float(path['No_Stress_Liquidity_EUR'].iloc[-1] - path['Stress_Liquidity_EUR'].iloc[-1]),
            'Recovery_Month': recovery,
        })
    comparison = pd.DataFrame(rows).set_index('Scenario')
    inputs = {
        'Historical months': len(months), 'Average monthly income EUR': income_avg,
        'Average monthly Employment income EUR': employment_avg, 'Average other income EUR': other_income_avg,
        'Average monthly expenses EUR': expense_avg, 'Normal monthly surplus EUR': normal_surplus,
        'Initial liquidity EUR': initial_liquidity, 'Essential expenses per month EUR': essential_monthly,
        'Three-month essential reserve EUR': reserve,
    }
    # ------------------------------------------------------------------
    # READABLE REPORT: keep detailed calculations, simplify presentation.
    # ------------------------------------------------------------------
    def eur(x):
        return f"{x:,.2f} EUR"

    scenario_labels = {
        'income_reduction': f'Scenario 1 — Employment income -{income_reduction_pct:.0%} for {income_reduction_months} months',
        'extraordinary_expense': f'Scenario 2 — One-time extraordinary expense of {eur(extraordinary_expense_eur)}',
        'extraordinary_expense_2': f'Scenario 3 — One-time extraordinary expense of {eur(extraordinary_expense_2_eur)}',
        'expense_increase': f'Scenario 4 — Recurring expenses +{expense_increase_pct:.0%} for {expense_increase_months} months',
    }
    name_by_kind = {kind: name for name, kind in definitions.items()}
    display_names = {name: scenario_labels[kind] for name, kind in definitions.items()}

    print('\n' + '=' * 86)
    print('BOX PF-4 — DETERMINISTIC FINANCIAL STRESS TESTING')
    print('=' * 86)
    print('OBJECTIVE: Can ordinary expenses be paid without selling investments,')
    print('and when does liquid savings recover after an adverse event?')
    print('\nBASELINE — based on the last', len(months), 'observed months')
    print(f'Average monthly income: {eur(income_avg)}')
    print(f'Average monthly expenses: {eur(expense_avg)}')
    print(f'Average monthly net savings: {eur(normal_surplus)}')
    print(f'Initial estimated liquidity: {eur(initial_liquidity)}')
    print(f'Minimum safety reserve (3 months essential expenses): {eur(reserve)}')
    print('NOTE: initial liquidity is inferred from historical net cash flows;')
    print('it is not a verified bank balance or an investment portfolio value.')

    print('\nSCENARIO LEGEND — each scenario is simulated separately')
    print(f'S1 — Employment income reduced by {income_reduction_pct:.0%} for {income_reduction_months} months.')
    print('     Employment includes net salary, performance bonus and year-end bonus;')
    print('     the reduction applies to their combined historical monthly average.')
    print(f'S2 — One unavoidable extraordinary expense of {eur(extraordinary_expense_eur)}')
    print(f'S3 — One unavoidable extraordinary expense of {eur(extraordinary_expense_2_eur)}')
    print(f'     paid only in future month {extraordinary_expense_month}; not a recurring monthly expense.')
    print(f'     Both extraordinary-expense scenarios are simulated separately.')
    print(f'S4 — Ordinary recurring expenses increased by {expense_increase_pct:.0%}')
    print(f'     for {expense_increase_months} months, then restored to the baseline.')
    print(f'Projection horizon: {horizon_months} months. No loan, no random simulation.')

    # Display only relevant indicators. A zero is kept when it distinguishes
    # scenarios; entirely-zero funding metrics are communicated in prose.
    compact = pd.DataFrame(index=[display_names[name] for name in comparison.index])
    compact['Total impact vs baseline (EUR)'] = comparison['Final_Gap_Vs_Baseline_EUR'].to_numpy()
    compact['Minimum liquidity (EUR)'] = comparison['Minimum_Liquidity_EUR'].to_numpy()
    compact['Months below safety reserve'] = comparison['Months_Below_Reserve'].to_numpy()
    compact['Maximum reserve shortfall (EUR)'] = comparison['Max_Gap_To_Reserve_EUR'].to_numpy()
    compact['Return to initial liquidity (month)'] = [
        ('Not applicable' if float(comparison.loc[name, 'Minimum_Liquidity_EUR']) >= initial_liquidity - 1e-8
         else (str(int(comparison.loc[name, 'Recovery_Month']))
               if pd.notna(comparison.loc[name, 'Recovery_Month']) else 'Not within horizon'))
        for name in comparison.index
    ]
    print('\nKEY RESULTS — COMPARATIVE SUMMARY')
    # Split five indicators into two readable blocks: index + up to 3 metrics.
    # Scenario names remain the row index in both blocks.
    print('PART 1 OF 2')
    print(compact.iloc[:, :3].round(2).to_string())
    print('\nPART 2 OF 2')
    print(compact.iloc[:, 3:].round(2).to_string())
    print('\nMETRIC LEGEND')
    print('Total impact vs baseline: lost cumulative liquidity versus the no-stress path at month', horizon_months)
    print('Minimum liquidity: lowest cash balance, including the initial balance.')
    print('Months below safety reserve: months with cash below 3 months of essential expenses.')
    print('Maximum reserve shortfall: largest additional amount needed to preserve that reserve.')
    print('Return to initial liquidity: first month after stress ends with cash >= starting cash;')
    print('this does NOT mean the lost difference versus the baseline has been recovered.')

    print('\nSCENARIO INTERPRETATION')
    for name, kind in definitions.items():
        path = all_paths[name]
        row = comparison.loc[name]
        minimum = float(row['Minimum_Liquidity_EUR'])
        below = int(row['Months_Below_Reserve'])
        gap_reserve = float(row['Max_Gap_To_Reserve_EUR'])
        gap_zero = float(row['Max_Gap_To_Zero_EUR'])
        deficits = int(row['Months_With_Deficit'])
        recovery = row['Recovery_Month']
        final_gap = float(row['Final_Gap_Vs_Baseline_EUR'])
        print(f'\n{scenario_labels[kind]}')
        if kind == 'income_reduction':
            loss = employment_avg * income_reduction_pct
            print(f'  Employment receipts decrease by {eur(loss)} per month during the shock.')
            print(f'  Monthly net cash flow during stress: {eur(normal_surplus - loss)}.')
        elif kind in ('extraordinary_expense', 'extraordinary_expense_2'):
            amount = extraordinary_expense_eur if kind == 'extraordinary_expense' else extraordinary_expense_2_eur
            print(f'  A one-time payment of {eur(amount)} is applied in month {extraordinary_expense_month}.')
            print('  It is NOT repeated in subsequent months.')
        else:
            added = expense_avg * expense_increase_pct
            print(f'  Ordinary monthly expenses increase by {eur(added)} during the shock.')
            print(f'  Monthly net cash flow during stress: {eur(normal_surplus - added)}.')
        print(f'  Lowest estimated liquid balance: {eur(minimum)}.')
        if gap_zero > 1e-8:
            print(f'  RESULT: available cash is insufficient; at least {eur(gap_zero)}')
            print('  in additional funding would be needed at the worst point.')
            print('  This could require portfolio withdrawals or another funding source.')
        elif below > 0:
            print(f'  RESULT: expenses remain payable from estimated cash, but the')
            print(f'  essential reserve is breached for {below} month(s).')
            print(f'  Up to {eur(gap_reserve)} would be needed to keep the reserve intact.')
            print('  Portfolio withdrawal is NOT mechanically required to pay expenses,')
            print('  but may be considered if the reserve must be preserved.')
        else:
            print('  RESULT: expenses remain payable from estimated cash while')
            print('  preserving the essential safety reserve; no portfolio withdrawal')
            print('  is required under these assumptions.')
        if deficits:
            print(f'  Monthly spending exceeds monthly income in {deficits} month(s).')
        else:
            print('  Monthly income covers monthly spending throughout this scenario.')
        if minimum >= initial_liquidity - 1e-8:
            print('  Return to starting liquidity: not applicable; the starting')
            print('  liquid balance is never breached.')
        elif pd.notna(recovery):
            print(f'  Starting liquidity is restored in future month {int(recovery)}.')
        else:
            print(f'  Starting liquidity is NOT restored within {horizon_months} months.')
        if final_gap > 1e-8:
            print(f'  At month {horizon_months}, liquidity remains {eur(final_gap)} below')
            print('  the no-stress path; returning to starting cash is not full catch-up.')

    if (comparison['Max_Gap_To_Zero_EUR'] <= 1e-8).all():
        print('\nOVERALL: no scenario exhausts estimated liquid funds within the horizon.')
        print('This does not prove that an investment portfolio is unnecessary in')
        print('reality: actual cash balances and portfolio holdings are not provided.')

    if show_charts:
        # One chart replaces three repetitive charts. All lines and reference
        # thresholds are defined in the legend and explanation below.
        fig, ax = plt.subplots(figsize=(11, 5.5))
        first_path = next(iter(all_paths.values()))
        months_axis = np.r_[0, first_path['Month'].to_numpy()]
        ax.plot(months_axis,
                np.r_[initial_liquidity, first_path['No_Stress_Liquidity_EUR'].to_numpy()],
                color='black', linestyle='--', linewidth=2, label='No-stress baseline')
        for name, kind in definitions.items():
            path = all_paths[name]
            ax.plot(np.r_[0, path['Month'].to_numpy()],
                    np.r_[initial_liquidity, path['Stress_Liquidity_EUR'].to_numpy()],
                    linewidth=2, label=scenario_labels[kind])
        ax.axhline(reserve, color='gray', linestyle=':', linewidth=1.5,
                   label='Safety reserve: 3 months essential expenses')
        ax.axhline(initial_liquidity, color='gray', linestyle='-.', linewidth=1,
                   label='Initial estimated liquid balance')
        ax.axhline(0, color='gray', linewidth=0.8, alpha=0.6,
                   label='Zero cash: funding required below this line')
        ax.set(title='PF-4 — Can liquid savings absorb financial stress?',
               xlabel='Months after the historical period',
               ylabel='Projected liquid balance (EUR)')
        ax.grid(alpha=0.2)
        ax.legend(loc='best', fontsize=8)
        plt.tight_layout()
        plt.show()
        print('CHART GUIDE: Each colored line shows cash remaining after income and')
        print('expenses in that scenario. The dashed baseline is the same household')
        print('without stress. Crossing below the dotted reserve means reduced')
        print('financial safety; crossing below zero means cash alone cannot pay')
        print('expenses. Reaching the initial balance does NOT erase the gap to')
        print('the no-stress baseline. No investment returns are modeled.')
    return {'inputs': inputs, 'scenario_parameters': {
        'income_reduction_pct': income_reduction_pct,
        'income_reduction_months': income_reduction_months,
        'employment_category': employment_category,
        'extraordinary_expense_eur': extraordinary_expense_eur,
        'extraordinary_expense_2_eur': extraordinary_expense_2_eur,
        'extraordinary_expense_month': extraordinary_expense_month,
        'expense_increase_pct': expense_increase_pct,
        'expense_increase_months': expense_increase_months,
        'horizon_months': horizon_months,
    }, 'comparison': comparison, 'monthly_trajectories': all_paths}


# Run all four deterministic stress scenarios.
# Income category audit is available on demand, but is not printed
# automatically to keep the financial report concise.
# pf4_income_categories = pf4_income_category_audit(PF4_HISTORICAL_MONTHS)
pf4_results = pf4_run_stress_tests(employment_category=PF4_EMPLOYMENT_CATEGORY)
```

```python
# ============================================================
# BOX 8 — PERSONAL FINANCE ANALYTICAL TOOLS
# ============================================================
# PURPOSE
# Make the results of the personal finance analyses (PF-1 to PF-4)
# available as structured Python tools for the conversational
# AI agent, without modifying existing calculations or outputs.
#
# ANALYTICAL TOOLS
# - pf5_get_available_analyses(): list available analyses.
# - pf5_get_financial_baseline(): retrieve historical financial data
#   and baseline assumptions.
# - pf5_compare_stress_scenarios(): compare the four stress scenarios.
# - pf5_get_stress_scenario(): retrieve detailed scenario results.
# - pf5_get_monthly_projection(): retrieve monthly projected values.
# - pf5_assess_investment_protection(): assess whether available
#   liquidity can cover financial shocks without requiring
#   investment withdrawals.
# - pf5_get_analysis_result(): retrieve structured results
#   from previous personal finance analyses.
#
# OUTPUTS
# - Structured financial data and scenario results.
# - Reusable information for the agentic analytical workflow.
# - No additional charts or duplicate reports.
# ============================================================


PF5_SCENARIO_LEGEND = {
    '1': 'Employment income -30% for 6 months',
    '2': 'One-time extraordinary expense of EUR 10,000',
    '3': 'One-time extraordinary expense of EUR 20,000',
    '4': 'Recurring expenses +15% for 12 months',
}


def pf5_native(value):
    """Convert pandas/numpy scalars and containers to serializable values."""
    if isinstance(value, (np.integer,)):
        return int(value)
    if isinstance(value, (np.floating,)):
        return None if not math.isfinite(float(value)) else float(value)
    if isinstance(value, (pd.Timestamp, pd.Period)):
        return str(value)
    if isinstance(value, (np.ndarray,)):
        return [pf5_native(x) for x in value.tolist()]
    if isinstance(value, pd.DataFrame):
        return [pf5_native(r) for r in value.reset_index().to_dict('records')]
    if isinstance(value, pd.Series):
        return {str(k): pf5_native(v) for k, v in value.items()}
    if isinstance(value, dict):
        return {str(k): pf5_native(v) for k, v in value.items()}
    if isinstance(value, (list, tuple)):
        return [pf5_native(v) for v in value]
    if value is None or value is pd.NA:
        return None
    if isinstance(value, (float, np.number)) and pd.isna(value):
        return None
    return value


def pf5_require_pf4():
    if 'pf4_results' not in globals() or not isinstance(pf4_results, dict):
        raise RuntimeError('PF-4 results unavailable. Run BOX PF-4 before BOX PF-5.')
    required = {'inputs', 'scenario_parameters', 'comparison', 'monthly_trajectories'}
    missing = required - set(pf4_results)
    if missing:
        raise RuntimeError(f'PF-4 results incomplete: {sorted(missing)}')
    return pf4_results


def pf5_scenario_key(scenario):
    key = str(scenario).strip().upper()
    if key not in PF5_SCENARIO_LEGEND:
        raise ValueError('Scenario must be 1, 2, 3 or 4.')
    matches = [name for name in pf5_require_pf4()['monthly_trajectories']
               if name.split(' — ')[0].upper() == 'S' + key]
    if len(matches) != 1:
        raise RuntimeError(f'Expected one PF-4 trajectory for {key}, found {len(matches)}.')
    return key, matches[0]


def pf5_get_available_analyses():
    """List existing outputs without running or modifying upstream boxes."""
    return {
        'PF-1 available': 'pf1_results' in globals(),
        'PF-2 available': 'pf2_results' in globals(),
        'PF-3 available': 'pf3_results' in globals(),
        'PF-4 available': 'pf4_results' in globals(),
        'stress_scenarios': PF5_SCENARIO_LEGEND.copy(),
        'tools': ['pf5_get_financial_baseline', 'pf5_compare_stress_scenarios',
                  'pf5_get_stress_scenario', 'pf5_get_monthly_projection',
                  'pf5_assess_investment_protection', 'pf5_get_analysis_result'],
    }


def pf5_get_financial_baseline():
    """Return exactly the assumptions PF-4 used to generate the chart."""
    r = pf5_require_pf4()
    return {
        'baseline': pf5_native(r['inputs']),
        'scenario_parameters': pf5_native(r['scenario_parameters']),
        'note': 'Initial liquidity is estimated from historical net cash flows, not a verified bank balance.'
    }


def pf5_get_stress_scenario(scenario):
    """Return a scenario's financial consequences, without redundant zero columns."""
    code, name = pf5_scenario_key(scenario)
    r = pf5_require_pf4()
    row = r['comparison'].loc[name]
    path = r['monthly_trajectories'][name]
    initial = float(r['inputs']['Initial liquidity EUR'])
    reserve = float(r['inputs']['Three-month essential reserve EUR'])
    min_cash = float(row['Minimum_Liquidity_EUR'])
    funding = float(row['Max_Gap_To_Zero_EUR'])
    reserve_gap = float(row['Max_Gap_To_Reserve_EUR'])
    recovery = row['Recovery_Month']
    recovery = int(recovery) if pd.notna(recovery) else None
    # The PF-4 recovery indicator is only meaningful if initial cash was breached.
    if min_cash >= initial - 1e-8:
        recovery = None
        recovery_status = 'Not applicable: initial liquidity was never breached'
    elif recovery is None:
        recovery_status = 'Not recovered within projection horizon'
    else:
        recovery_status = f'Initial liquidity restored in month {recovery}'
    if funding > 1e-8:
        assessment = 'Cash is insufficient: additional funding is required to meet all expenses.'
    elif reserve_gap > 1e-8:
        assessment = 'Cash covers expenses, but the essential safety reserve is breached.'
    else:
        assessment = 'Estimated cash covers expenses without breaching the essential reserve.'
    result = {
        'scenario': code, 'description': PF5_SCENARIO_LEGEND[code],
        'total_impact_vs_no_stress_eur': float(row['Final_Gap_Vs_Baseline_EUR']),
        'minimum_projected_liquidity_eur': min_cash,
        'initial_estimated_liquidity_eur': initial,
        'minimum_safety_reserve_eur': reserve,
        'months_below_safety_reserve': int(row['Months_Below_Reserve']),
        'months_with_negative_liquidity': int(row['Months_Negative_Liquidity']),
        'months_with_monthly_deficit': int(row['Months_With_Deficit']),
        'return_to_initial_liquidity_month': recovery,
        'recovery_interpretation': recovery_status,
        'final_projected_liquidity_eur': float(row['Final_Liquidity_EUR']),
        'assessment': assessment,
    }
    if reserve_gap > 1e-8:
        result['maximum_safety_reserve_shortfall_eur'] = reserve_gap
    if funding > 1e-8:
        result['maximum_funding_required_eur'] = funding
    result['investment_note'] = ('The portfolio is not included in this calculation. '
                                 'Whether investments must be sold depends on verified cash available.')
    return pf5_native(result)


def pf5_compare_stress_scenarios():
    """Compare 1–4, keeping only decision-relevant nonzero indicators."""
    return [pf5_get_stress_scenario(code) for code in PF5_SCENARIO_LEGEND]


def pf5_get_monthly_projection(scenario, month=None):
    """Read exact numerical chart values for one month or all months (1–24)."""
    code, name = pf5_scenario_key(scenario)
    path = pf5_require_pf4()['monthly_trajectories'][name]
    if month is not None:
        if isinstance(month, bool) or not isinstance(month, (int, np.integer)):
            raise ValueError('month must be an integer between 1 and the projection horizon.')
        selected = path.loc[path['Month'] == month]
        if selected.empty:
            raise ValueError(f'Month {month} not available for {code}.')
    else:
        selected = path
    # No all-zero auxiliary flags: return actual cash flows and thresholds.
    cols = ['Month', 'Income_EUR', 'Ordinary_Expenses_EUR',
            'Extraordinary_Expense_EUR', 'Net_Cash_Flow_EUR',
            'Stress_Liquidity_EUR', 'No_Stress_Liquidity_EUR']
    records = selected[cols].to_dict('records')
    for rec in records:
        if abs(rec['Extraordinary_Expense_EUR']) < 1e-8:
            rec.pop('Extraordinary_Expense_EUR')
    return {'scenario': code, 'description': PF5_SCENARIO_LEGEND[code],
            'months': pf5_native(records),
            'safety_reserve_eur': float(pf5_require_pf4()['inputs']['Three-month essential reserve EUR'])}


def pf5_assess_investment_protection(available_cash_eur, protected_investment_eur=0.0):
    """Rebase PF-4 cash paths to VERIFIED cash outside investments.

    available_cash_eur: liquid funds that can be spent WITHOUT selling the
        investment, excluding the protected investment itself.
    protected_investment_eur: optional amount to preserve; informational only,
        NOT added to available cash and NOT used as a source of funding.
    Cash flows and shocks remain identical to PF-4; only starting cash changes.
    """
    cash = float(available_cash_eur)
    invested = float(protected_investment_eur)
    if not math.isfinite(cash) or not math.isfinite(invested) or cash < 0 or invested < 0:
        raise ValueError('Both amounts must be finite, nonnegative EUR amounts.')
    r = pf5_require_pf4()
    modeled_initial = float(r['inputs']['Initial liquidity EUR'])
    reserve = float(r['inputs']['Three-month essential reserve EUR'])
    results = []
    for code in PF5_SCENARIO_LEGEND:
        _, name = pf5_scenario_key(code)
        path = r['monthly_trajectories'][name]
        rebased = path['Stress_Liquidity_EUR'].to_numpy(dtype=float) + cash - modeled_initial
        min_cash = min(cash, float(rebased.min()))
        gap = max(0.0, -min_cash)
        reserve_gap = max(0.0, reserve - min_cash)
        # Recover verified initial cash after the shock's lowest point.
        low_month = int(path.iloc[int(np.argmin(rebased))]['Month'])
        later = path.loc[(path['Month'] > low_month) & (rebased >= cash - 1e-8), 'Month']
        recovery = None if min_cash >= cash - 1e-8 or later.empty else int(later.iloc[0])
        results.append({
            'scenario': code, 'description': PF5_SCENARIO_LEGEND[code],
            'minimum_cash_outside_investments_eur': round(min_cash, 2),
            'can_pay_without_selling_investments': gap <= 1e-8,
            'can_preserve_safety_reserve': reserve_gap <= 1e-8,
            'months_below_safety_reserve': int((rebased < reserve - 1e-8).sum()),
            'additional_funding_needed_eur': round(gap, 2),
            'cash_needed_to_preserve_reserve_eur': round(reserve_gap, 2),
            'return_to_verified_starting_cash_month': recovery,
        })
    return {
        'verified_cash_outside_investments_eur': cash,
        'protected_investment_eur': invested,
        'safety_reserve_eur': reserve,
        'scenarios': results,
        'method': 'Same deterministic PF-4 cash-flow trajectories, rebased to verified available cash.',
        'caution': ('An additional funding requirement does not automatically imply '
                    'investment liquidation: alternative funds may exist. Investment '
                    'returns, taxes, transaction costs and liquidity restrictions are not modeled.'),
    }


def pf5_get_analysis_result(box):
    """Return previously computed PF-1/PF-2/PF-3 outputs, when present."""
    key = str(box).strip().upper().replace('PF-', 'PF')
    lookup = {'PF1': 'pf1_results', 'PF2': 'pf2_results', 'PF3': 'pf3_results'}
    if key not in lookup:
        raise ValueError('Choose PF1, PF2 or PF3; use dedicated PF-4 tools for stress testing.')
    variable = lookup[key]
    if variable not in globals():
        raise RuntimeError(f'{variable} is not available. Run its BOX first.')
    return {'box': key, 'result': pf5_native(globals()[variable])}


# Structured tool registry for future agent integration (BOX PF-6).
PF5_ANALYTICAL_TOOLS = {
    'get_available_analyses': pf5_get_available_analyses,
    'get_financial_baseline': pf5_get_financial_baseline,
    'compare_stress_scenarios': pf5_compare_stress_scenarios,
    'get_stress_scenario': pf5_get_stress_scenario,
    'get_monthly_projection': pf5_get_monthly_projection,
    'assess_investment_protection': pf5_assess_investment_protection,
    'get_analysis_result': pf5_get_analysis_result,
}

print('BOX PF-5 — PERSONAL FINANCE ANALYTICAL TOOLS')
print('Read-only analytical tools registered:', ', '.join(PF5_ANALYTICAL_TOOLS))
# Verify that PF-1/PF-2/PF-3 outputs are retrievable through their registered tool.
# Verify PF-4 through its dedicated validated retrieval tool.
for box in ('PF1', 'PF2', 'PF3'):
    try:
        result = PF5_ANALYTICAL_TOOLS['get_analysis_result'](box)
        status = 'available — tool retrieval OK' if result.get('result') is not None else 'available — empty result'
    except Exception as exc:
        status = f'not retrievable — {exc}'
    print(f'{box.replace("PF", "PF-")} data: {status}')

try:
    result = PF5_ANALYTICAL_TOOLS['get_financial_baseline']()
    status = 'available — tool retrieval OK' if result.get('baseline') is not None else 'available — empty result'
except Exception as exc:
    status = f'not retrievable — {exc}'
print(f'PF-4 data: {status}')
print('No scenarios were rerun and no charts were generated.')
```

## **Preliminary Agentic Architecture — Personal Finance & Stress Testing**

This section introduces a **preliminary LangGraph workflow** to test the agentic analysis of personal finances, including budgeting, cash flows, financial resilience, and stress testing.

It serves as an **initial validation of the agentic framework**, which is subsequently extended to incorporate investment analysis and integrated financial assessments. The complete multi-agent architecture is developed and tested in the final section of the project.

```python
# ============================================================
# BOX 9
#
# Agent Creation
# AgentState and AgentDecision
#
# Single LangGraph workflow for financial analysis tools.
#
# This BOX defines only the agent state and possible actions.
# It does not create nodes, run analyses, or compile a graph:
# these operations follow in subsequent sections.
# ============================================================


class AgentState(TypedDict):
    query: str
    result: Any
    analysis_type: str
    rolling_months: int


class AgentDecision(TypedDict):
    action: Literal[
        # Financial analysis tools
        "analyze_finance",
        "analyze_deficit_expenses",
        "compare_income_expenses",

        # Additional financial analysis tools
        "analyze_cash_flow_risk",
        "simulate_savings",
        "assess_financial_resilience",
        "simulate_budget_scenarios",

        "end"
    ]
```

```python
# ============================================================
# BOX 10
#
# Creation of operational nodes
# ============================================================

# Centralized result formatting without changing tool calculations.
import math


def _pf_value(value, suffix="", digits=2):
    if value is None or isinstance(value, bool):
        return "data unavailable"
    try:
        v = float(value)
        if not math.isfinite(v):
            return "data unavailable"
        return f"{v:,.{digits}f}".replace(",", "_").replace(".", ",").replace("_", ".") + suffix
    except (ValueError, TypeError):
        return "data unavailable"


def _pf_present(value):
    if value is None:
        return False
    try:
        return math.isfinite(float(value))
    except (TypeError, ValueError):
        return False


def _pf_savings(result, query):
    cap = result.get("savings_capacity", {})
    if not isinstance(cap, dict):
        return "Savings capacity data unavailable."
    lines = ["SAVINGS CAPACITY",
             f"Historical period: {_pf_value(cap.get('Months_Analyzed'), digits=0)} months",
             f"Average monthly savings: {_pf_value(cap.get('Mean_Monthly_Savings_EUR'), ' €')}",
             f"Median monthly savings: {_pf_value(cap.get('Median_Monthly_Savings_EUR'), ' €')}",
             f"Historical savings rate: {_pf_value(cap.get('Historical_Savings_Rate_%'), '%')}",
             f"Historical net savings: {_pf_value(cap.get('Total_Net_Savings_EUR'), ' €')}"]
    projections = result.get("projections", [])
    if isinstance(projections, list) and projections:
        lines.append("\nPROJECTIONS (constant monthly savings, excluding investment returns)")
        for row in projections:
            if not isinstance(row, dict):
                continue
            horizon = row.get("Horizon_Months")
            base, conservative = row.get("Baseline_EUR"), row.get("Conservative_EUR")
            if not _pf_present(horizon):
                continue
            line = f"{int(horizon)} months: {_pf_value(base, ' €')}"
            if _pf_present(base) and _pf_present(conservative) and abs(float(base)-float(conservative)) > 0.005:
                line += f" (conservative: {_pf_value(conservative, ' €')})"
            lines.append(line)
    lines.append("\nAssessment: savings capacity is estimated from historical cash flows and is not a guaranteed amount available.")
    return "\n".join(lines)


def _pf_resilience(result, query):
    indicators = result.get("resilience_indicators", {})
    if not isinstance(indicators, dict):
        return "Financial resilience indicators unavailable."
    labels = [
        ("Months_Analyzed", "Historical period", " months", 0),
        ("Average_Monthly_Income_EUR", "Average monthly income", " €", 2),
        ("Average_Monthly_Expenses_EUR", "Average monthly expenses", " €", 2),
        ("Average_Monthly_Balance_EUR", "Average monthly balance", " €", 2),
        ("Available_Liquidity_EUR", "Available liquidity", " €", 2),
        ("Estimated_Essential_Expenses_EUR", "Average essential expenses", " €", 2),
        ("Full_Expense_Coverage_Months", "Full-expense financial autonomy", " months", 1),
        ("Essential_Expense_Coverage_Months", "Essential-expense financial autonomy", " months", 1),
    ]
    lines = ["FINANCIAL RESILIENCE"]
    for key, label, unit, digits in labels:
        if key in indicators:
            lines.append(f"{label}: {_pf_value(indicators[key], unit, digits)}")
    fund = result.get("emergency_fund", [])
    if isinstance(fund, list) and fund:
        lines.append("\nEMERGENCY FUND TARGETS")
        for row in fund:
            if not isinstance(row, dict):
                continue
            months = row.get("Coverage_Months")
            if not _pf_present(months):
                continue
            line = f"{int(months)} months: {_pf_value(row.get('Full_Expenses_Target_EUR'), ' €')} for all expenses"
            if _pf_present(row.get("Essential_Expenses_Target_EUR")):
                line += f"; {_pf_value(row['Essential_Expenses_Target_EUR'], ' €')} for essential expenses"
            lines.append(line)
    if not _pf_present(indicators.get("Available_Liquidity_EUR")):
        lines.append("\nAssessment: without verified available liquidity, the actual number of months covered cannot be established.")
    else:
        lines.append("\nAssessment: estimated financial autonomy depends on the stated liquidity and the assumption of no additional income.")
    return "\n".join(lines)


def _pf_scenarios(result, query):
    # S1–S4 remain internal identifiers: only the labels are translated.
    import re
    scenarios = result if isinstance(result, list) else [result]
    horizon = int(globals().get("PF4_HORIZON_MONTHS", 24))
    lines = [f"FINANCIAL STRESS TEST — Projection horizon: {horizon} months"]
    descriptions = {
        "1": "Employment income reduced by 30% for 6 months",
        "2": "One-time unexpected expense of EUR 10,000 (in the first month)",
        "3": "One-time unexpected expense of EUR 20,000 (in the first month)",
        "4": "Recurring expenses increased by 15% for 12 months",
    }
    valid = []
    for row in scenarios:
        if not isinstance(row, dict) or str(row.get("scenario", "")).upper() not in descriptions:
            continue
        code = str(row["scenario"]).upper()
        number = int(code)
        valid.append(row)
        lines.extend([
            "",
            f"SCENARIO {number}: {descriptions[code]}",
            f"Reduction in accumulated liquidity versus the no-stress baseline at the end of {horizon} months: {_pf_value(row.get('total_impact_vs_no_stress_eur'), ' €')}",
        ])
        status = str(row.get("recovery_interpretation") or "")
        match = re.search(r"restored in month\s+(\d+)", status, flags=re.IGNORECASE)
        if match:
            recovery = f"in simulation month {int(match.group(1))}"
        elif "Not recovered" in status:
            recovery = f"not achieved within the {horizon}-month projection"
        elif "Not applicable" in status:
            recovery = "not applicable"
        else:
            recovery = "data unavailable"
        lines.append(f"Return to initial liquidity after the shock: {recovery}")
        funding = row.get("maximum_funding_required_eur")
        if _pf_present(funding) and float(funding) > 0:
            lines.append("Available liquidity is insufficient: additional funding is required to cover expenses.")
            lines.append(f"Additional funding required: {_pf_value(funding, ' €')}")
    if not valid:
        return "Scenario results unavailable in the expected format."
    if len(valid) > 1:
        ranked = [r for r in valid if _pf_present(r.get("minimum_projected_liquidity_eur"))]
        if ranked:
            worst = min(ranked, key=lambda r: float(r["minimum_projected_liquidity_eur"]))
            lines.append(f"\nScenario with the greatest financial stress: Scenario {str(worst['scenario'])[-1]}.")
    return "\n".join(lines)


def format_financial_response(result, analysis_type, query):
    if result is None or result == "":
        return "No results available."
    if analysis_type == "simulate_savings" and isinstance(result, dict):
        return _pf_savings(result, query)
    if analysis_type == "assess_financial_resilience" and isinstance(result, dict):
        return _pf_resilience(result, query)
    if analysis_type == "simulate_budget_scenarios" and isinstance(result, (dict, list)):
        return _pf_scenarios(result, query)
    if isinstance(result, str):
        return result
    return "The tool returned unrecognized structured data: raw output is not displayed."


class FinanceToolNode(Runnable):
    def invoke(self, state: AgentState, config=None, **kwargs):
        output = analyze_finance.invoke({
            "query": state["query"],
            "rolling_months": state.get("rolling_months", 24)
        })
        return {"query": state["query"], "result": output, "analysis_type": "analyze_finance"}


class DeficitToolNode(Runnable):
    def invoke(self, state: AgentState, config=None, **kwargs):
        output = analyze_deficit_expenses.invoke({
            "query": state["query"],
            "rolling_months": state.get("rolling_months", 24)
        })
        return {"query": state["query"], "result": output, "analysis_type": "analyze_deficit_expenses"}


class CompareIncomeExpensesToolNode(Runnable):
    def invoke(self, state: AgentState, config=None, **kwargs):
        output = compare_income_expenses.invoke({
            "query": state["query"],
            "rolling_months": state.get("rolling_months", 24)
        })
        return {"query": state["query"], "result": output, "analysis_type": "compare_income_expenses"}


# Final node
def end_node(state: AgentState):
    return {"result": format_financial_response(state.get("result"), state.get("analysis_type", ""), state.get("query", ""))}

# Nodes for additional financial analyses.
# Results are retrieved through previously registered tools,
# without repeating calculations or creating additional graphs.

def _finance_select_result(result, query, area):
    """Select relevant indicators without recalculating or altering results."""
    if not isinstance(result, dict):
        return result
    q = _finance_normalize(query)
    if area == "cash":
        # Original result keys remain unchanged.
        selected = {k: result[k] for k in ("indicators", "statistics") if k in result}
        if _finance_any(q, "monthly", "month-by-month", "historical", "history", "trend", "detailed", "details") and "monthly_cash_flow" in result:
            selected["monthly_cash_flow"] = result["monthly_cash_flow"]
        return selected or result
    if area in ("savings", "resilience"):
        # Results may contain heterogeneous structures: do not discard unverified data.
        return result
    return result


def _finance_scenario_request(query):
    """Identify a single stress scenario without changing the original simulation."""
    q = _finance_normalize(query)
    # Comparative questions require all scenarios, not just the last shock mentioned.
    if (_finance_any(q, "which scenario", "compare scenarios", "scenario comparison", "compare shocks", "worst-case scenario", "most severe shock", "most at risk", "greatest risk", "most critical")
        or (_finance_any(q, "between", "among", "compare", "comparison") and
            sum(bool(_finance_any(q, *terms)) for terms in (
                ("income reduction", "income shock", "salary reduction", "reduced income"),
                ("unexpected expense", "extraordinary expense", "one-off expense", "unplanned cost"),
                ("expense increase", "recurring expenses", "higher recurring costs", "expense inflation"),
            )) >= 2)):
        return None
    if _finance_any(q, "s1", "scenario 1", "income reduction", "income shock", "salary reduction", "reduced income", "income decline", "bonus loss", "missing bonus", "performance bonus", "performance award", "employment income"):
        return "1"
    if _finance_any(q, "s3", "scenario 3"):
        return "3"
    if _finance_any(q, "s2", "scenario 2"):
        return "2"
    if _finance_any(q, "s4", "scenario 4", "recurring expense increase", "recurring expenses", "expense inflation", "expense increase", "higher recurring costs"):
        return "4"
    # Monetary amounts and percentages are dynamic inputs, not semantic triggers.
    # The current PF-4 tools expose four predefined scenarios only.
    if _finance_any(q, "unexpected expense", "extraordinary expense", "one-off expense", "unplanned cost", "emergency cost"):
        return None
    return None


class CashFlowRiskToolNode(Runnable):
    def invoke(self, state: AgentState, config=None, **kwargs):
        output = PF5_ANALYTICAL_TOOLS["get_analysis_result"]("PF1")
        selected = _finance_select_result(output["result"], state["query"], "cash")
        indicators = selected.get("indicators", {}) if isinstance(selected, dict) else {}
        statistics = selected.get("statistics", []) if isinstance(selected, dict) else []

        if not indicators:
            return {"query": state["query"], "result": selected, "analysis_type": "analyze_cash_flow_risk"}

        def number(key, decimals=1, suffix=""):
            value = indicators.get(key)
            if not isinstance(value, (int, float)):
                return "data unavailable"
            formatted = f"{value:,.{decimals}f}".replace(",", "_").replace(".", ",").replace("_", ".")
            return formatted + suffix

        income_cv = expenses_cv = None
        for item in statistics:
            if isinstance(item, dict):
                if item.get("index") == "Income":
                    income_cv = item.get("CV_%")
                elif item.get("index") == "Expenses":
                    expenses_cv = item.get("CV_%")

        months = indicators.get("Months_Analyzed")
        months_text = str(int(months)) if isinstance(months, (int, float)) else "analyzed period"
        lines = [
            f"**Income and expense regularity — last {months_text} months**",
            f"- Months with positive balance: {number('Surplus_Months', 0)} ({number('Surplus_Frequency_%', 1, '%')}).",
            f"- Months with negative balance: {number('Deficit_Months', 0)} ({number('Deficit_Frequency_%', 1, '%')}).",
            f"- Average monthly balance: {number('Average_Monthly_Balance_EUR', 2, ' €')}.",
            f"- Monthly cash flow volatility (standard deviation): {number('Cash_Flow_Volatility_EUR', 2, ' €')}.",
            f"- Average savings rate: {number('Average_Savings_Rate_%', 1, '%')}.",
        ]
        if isinstance(income_cv, (int, float)) and isinstance(expenses_cv, (int, float)):
            lines.append(
                f"- Relative variability: income {income_cv:.1f}%, expenses {expenses_cv:.1f}%.".replace(".", ",")
            )
        deficit_months = indicators.get("Deficit_Months")
        average_balance = indicators.get("Average_Monthly_Balance_EUR")
        if isinstance(deficit_months, (int, float)) and isinstance(average_balance, (int, float)):
            if average_balance > 0 and deficit_months > 0:
                lines.append(
                    "**Assessment:** the average balance is positive, but some months have a deficit. "
                    "This indicates potential temporary liquidity pressure, not a confirmed liquidity shortfall."
                )
            elif average_balance > 0 and deficit_months == 0:
                lines.append("**Assessment:** historical balances are positive in all analyzed months.")
            else:
                lines.append("**Assessment:** the average balance is not positive; deficit months should be examined.")
        lines.append(
            "Actual available liquidity cannot be determined from historical cash flows alone: "
            "the verified available cash balance is needed."
        )
        return {"query": state["query"], "result": "\n\n".join(lines), "analysis_type": "analyze_cash_flow_risk"}


class SavingsSimulationToolNode(Runnable):
    def invoke(self, state: AgentState, config=None, **kwargs):
        output = PF5_ANALYTICAL_TOOLS["get_analysis_result"]("PF2")
        return {"query": state["query"], "result": _finance_select_result(output["result"], state["query"], "savings"), "analysis_type": "simulate_savings"}


class FinancialResilienceToolNode(Runnable):
    def invoke(self, state: AgentState, config=None, **kwargs):
        output = PF5_ANALYTICAL_TOOLS["get_analysis_result"]("PF3")
        return {"query": state["query"], "result": _finance_select_result(output["result"], state["query"], "resilience"), "analysis_type": "assess_financial_resilience"}


class BudgetScenariosToolNode(Runnable):
    def invoke(self, state: AgentState, config=None, **kwargs):
        # Full comparison: four independent scenarios, S1–S4.
        # Monthly data remain available through the registered tools.
        scenario = _finance_scenario_request(state["query"])
        if scenario is not None:
            output = PF5_ANALYTICAL_TOOLS["get_stress_scenario"](scenario)
        else:
            output = PF5_ANALYTICAL_TOOLS["compare_stress_scenarios"]()
        return {"query": state["query"], "result": output, "analysis_type": "simulate_budget_scenarios"}
```

```python
# ============================================================
# BOX 11
#
# Decision-making agent
# ============================================================
# Selects one of the seven available actions in the graph.
# Specific requests take priority over general ones.
# This code performs no calculations and creates no additional graphs.
# ============================================================

def agent_node(state: AgentState) -> AgentDecision:
    query = _finance_normalize(state["query"])

    # Comparisons between financial shocks always require all S1–S4 scenarios.
    if (_finance_any(query, "scenario comparison", "compare shocks", "compare scenarios",
                     "which scenario", "which scenarios", "worst-case scenario",
                     "most severe shock", "most at risk", "greatest risk")
        or (_finance_any(query, "between", "compare") and
            sum(bool(_finance_has(query, *terms)) for terms in (
                ("income reduction", "income shock", "salary reduction"),
                ("unexpected expense", "extraordinary expense", "one-off expense"),
                ("expense increase", "recurring expenses", "higher recurring costs"),
            )) >= 2)):
        return {"action": "simulate_budget_scenarios", "analysis_type": "simulate_budget_scenarios"}

    # Hypothetical financial shocks and recovery scenarios. Amounts and durations
    # are parameters, not semantic triggers.
    if (
        _finance_any(query, "stress test", "financial shock", "income shock",
                     "income reduction", "income decline", "reduced income",
                     "salary reduction", "salary cut", "bonus loss", "missing bonus",
                     "performance bonus", "performance award", "unexpected expense",
                     "extraordinary expense", "one-off expense", "unplanned cost",
                     "emergency cost", "recurring expense increase", "expense increase",
                     "expense inflation", "higher recurring costs", "liquidity recovery",
                     "recovery time", "recovery period", "time to recover",
                     "post-shock recovery", "liquidity restoration", "return to baseline",
                     "financial setback", "liquidity to recover",
                     "s1", "s2", "s3", "s4")
        or (_finance_has(query, "reduce", "reduction", "decrease", "decline", "loss", "fall") and
            _finance_has(query, "income", "salary", "employment", "bonus", "earnings"))
        or (_finance_has(query, "increase", "rise", "higher") and
            _finance_has(query, "expenses", "costs", "spending"))
        or (_finance_has(query, "investment", "portfolio") and
            _finance_has(query, "sell", "withdraw", "protect", "preserve"))
    ):
        return {"action": "simulate_budget_scenarios", "analysis_type": "simulate_budget_scenarios"}

    # Historical deficit drivers must not be confused with identifying deficit months.
    if (_finance_any(query, "expenses in deficit months", "deficit drivers",
                     "largest expenses during deficits", "deficit analysis",
                     "spending drivers", "negative balance causes")
        or (_finance_has(query, "deficit", "negative balance") and
            _finance_has(query, "main expenses", "largest expenses", "drivers", "causes", "analyze"))):
        return {"action": "analyze_deficit_expenses", "analysis_type": "analyze_deficit_expenses"}

    # Historical comparisons and ratios, including the savings rate.
    if (_finance_any(query, "compare months", "monthly comparison", "compare two months",
                     "month-over-month comparison", "annual comparison", "yearly comparison",
                     "overall comparison", "general comparison", "income versus expenses",
                     "income vs expenses", "expense-to-income ratio", "expenses to income",
                     "spending-to-income ratio", "expense income percentage",
                     "expenses divided by income", "spending ratio", "savings rate",
                     "saving rate", "savings percentage", "savings-to-income ratio",
                     "percentage of income saved", "monthly savings rate",
                     "annual income versus expenses", "yearly income vs expenses",
                     "monthly income versus expenses", "monthly income vs expenses",
                     "total income versus expenses", "aggregate income vs expenses")
        or (_finance_has(query, "compare", "comparison", "versus") and
            _finance_has(query, "income", "expenses", "months", "years"))
        or (_finance_has(query, "compare", "comparison") and
            len(re.findall(r"\b\d{4} \d{2}\b", query)) >= 2)):
        return {"action": "compare_income_expenses", "analysis_type": "compare_income_expenses"}

    # Financial autonomy and adequacy of essential-expense reserves.
    if (_finance_any(query, "financial autonomy", "months of autonomy", "liquidity runway",
                     "financial runway", "months of expense coverage", "liquidity duration",
                     "survival months", "emergency fund", "emergency reserve", "safety reserve",
                     "essential expenses", "essential expense coverage", "emergency savings adequacy",
                     "reserve sufficiency", "difficult period")
        or (_finance_has(query, "months") and
            _finance_has(query, "autonomy", "survive", "cover", "runway"))):
        return {"action": "assess_financial_resilience", "analysis_type": "assess_financial_resilience"}

    # Savings capacity and accumulation projections, not historical average savings.
    if (_finance_any(query, "savings capacity", "saving capacity", "how much can i save",
                     "available savings", "monthly savings potential", "disposable surplus",
                     "current savings ability", "future savings", "savings projection",
                     "savings accumulation", "projected savings", "savings growth",
                     "accumulation over time", "future savings balance", "savings simulation",
                     "saving simulation", "savings capacity simulation", "ability to save",
                     "simulated savings", "savings accumulation model",
                     "savings projection results", "set money aside")
        or (_finance_has(query, "savings", "save", "accumulation") and
            _finance_has(query, "future", "project", "simulate", "grow", "capacity"))):
        return {"action": "simulate_savings", "analysis_type": "simulate_savings"}

    # Cash-flow stability and liquidity risk, distinct from hypothetical shocks.
    if (_finance_any(query, "income regularity", "expense regularity", "cash flow regularity",
                     "liquidity risk", "liquidity problems", "cash flow variability",
                     "income stability", "expense stability", "critical months", "critical cash flow",
                     "cash flow stress", "liquidity pressure", "cash flow shortfall",
                     "financially vulnerable months", "cash flow stability", "cash flow volatility",
                     "cash flow monitoring", "financial stability indicators",
                     "income expense fluctuations", "cash flow risk factors",
                     "financial stability", "financial resilience")
        or (_finance_has(query, "regular", "stable", "stability", "irregular", "variability", "volatile") and
            _finance_has(query, "income", "expenses", "cash flow", "cash", "spending"))
        or (_finance_has(query, "risk", "problems", "pressure") and
            _finance_has(query, "liquidity", "cash flow"))
        or (_finance_has(query, "critical") and
            _finance_has(query, "cash", "flow", "months"))):
        return {"action": "analyze_cash_flow_risk", "analysis_type": "analyze_cash_flow_risk"}

    # General descriptive historical analysis and statistics.
    if (_finance_any(query, "monthly totals", "monthly total", "monthly summary",
                     "monthly income", "monthly expenses", "monthly balance",
                     "monthly net cash flow", "monthly aggregation", "top 3", "top three",
                     "largest income items", "largest expenses", "highest income",
                     "highest expenses", "income and expense ranking", "average monthly savings",
                     "mean monthly savings", "average savings", "monthly surplus average",
                     "average net cash flow", "average amount saved", "expense categories",
                     "spending categories", "expenditure breakdown", "expenses by category",
                     "spending distribution", "category expenses", "income categories",
                     "income sources", "income breakdown", "income distribution",
                     "income by category", "revenue categories", "deficit months",
                     "months in deficit", "negative monthly balance", "expenses exceed income",
                     "monthly shortfall", "annual summary", "yearly summary", "annual totals",
                     "yearly income", "yearly expenses", "annual net savings",
                     "annual financial overview", "overall summary", "general summary",
                     "financial overview", "total income", "total expenses", "total savings",
                     "overall cash flow")
        or _finance_has(query, "totals", "summary", "statistics", "categories", "distribution",
                        "deficit", "balance", "income", "expenses", "savings")):
        return {"action": "analyze_finance", "analysis_type": "analyze_finance"}

    return {"action": "end", "analysis_type": "end"}
```

```python
# ============================================================
# BOX 12
#
# Building the unified LangGraph graph
#
# Connects the decision-making agent to the seven operational nodes.
# The graph is built and compiled only once.
# ============================================================

graph = StateGraph(AgentState)

graph.add_node("agent", agent_node)
graph.add_node("analyze_finance", FinanceToolNode())
graph.add_node("analyze_deficit_expenses", DeficitToolNode())
graph.add_node("compare_income_expenses", CompareIncomeExpensesToolNode())
graph.add_node("analyze_cash_flow_risk", CashFlowRiskToolNode())
graph.add_node("simulate_savings", SavingsSimulationToolNode())
graph.add_node("assess_financial_resilience", FinancialResilienceToolNode())
graph.add_node("simulate_budget_scenarios", BudgetScenariosToolNode())
graph.add_node("end", end_node)

graph.set_entry_point("agent")

graph.add_conditional_edges("agent", lambda x: x["action"], {
    "analyze_finance": "analyze_finance",
    "analyze_deficit_expenses": "analyze_deficit_expenses",
    "compare_income_expenses": "compare_income_expenses",
    "analyze_cash_flow_risk": "analyze_cash_flow_risk",
    "simulate_savings": "simulate_savings",
    "assess_financial_resilience": "assess_financial_resilience",
    "simulate_budget_scenarios": "simulate_budget_scenarios",
    "end": "end"
})

for node_name in (
    "analyze_finance",
    "analyze_deficit_expenses",
    "compare_income_expenses",
    "analyze_cash_flow_risk",
    "simulate_savings",
    "assess_financial_resilience",
    "simulate_budget_scenarios",
):
    graph.add_edge(node_name, "end")

app = graph.compile()

# Display the graph structure.
from IPython.display import Image, display

try:
    display(Image(app.get_graph().draw_mermaid_png()))
except Exception:
    # Fallback: print the graph in Mermaid format.
    print(app.get_graph().draw_mermaid())
```

```python
# ============================================================
# BOX 13
# GRADIO — PERSONAL FINANCE INTELLIGENCE
# ============================================================

# Questions for the three financial analysis tools.
ANALYSIS_QUESTIONS = [
    "Show me the monthly totals",
    "Show me the top 3 income and expense items",
    "What is my average monthly savings?",
    "Show me the expense categories",
    "Show me the income categories",
    "Which months have a deficit?",
    "Show me the annual summary",
    "Show me the overall summary",
    "Analyze the main expenses in deficit months",
    "Compare 2026-01 and 2026-02",
    "Show me the annual comparison of income and expenses",
    "Show me the monthly comparison of income and expenses",
    "Show me the expense-to-income ratio",
    "Show me the savings rate",
    "Show me the overall comparison of income and expenses",
]

# Questions framed as real-life financial decisions.
# Each question contains terms recognized by the decision-making agent.
# Numerical assumptions are those of the simulations already available.
DECISION_QUESTIONS = [
    "Are my income and expenses regular enough, or am I at risk of liquidity problems?",
    "Are there months when my cash flows become particularly critical?",
    "How stable is my cash flow, and what should I monitor?",
    "Given my current income and expenses, how much can I save?",
    "If I continue saving at my current pace, how might my future savings grow?",
    "What does the savings simulation reveal about my ability to set money aside?",
    "If I faced a difficult period, how many months of financial autonomy would I have?",
    "Is my emergency fund sufficient to cover essential expenses?",
    "How resilient are my finances if my income decreases?",
    "If I did not receive my usual bonus or performance award and my employment income fell by 30% for six months, what would happen to my liquidity?",
    "If I suddenly had to pay an unexpected expense of €10,000, could I afford it?",
    "And if an unexpected expense of €20,000 arose, how severely would it affect my savings?",
    "If my recurring expenses increased by 15% for one year, how much would my future liquidity decrease?",
    "Between an income reduction, an unexpected expense, and an increase in expenses, which scenario would put my safety reserve most at risk?",
    "After a financial setback, how long would it take for my liquidity to recover?",
    "Under the simulated financial shocks, could I cover my expenses without selling investments?",
]

# ============================================================
# 2. PREPARE FINANCIAL DATA FOR CHARTS
# ============================================================

def prepare_financial_data():

    data = get_rolling_data(24)

    data["Month"] = pd.to_datetime(data["Month"])

    monthly = (
        data.groupby(["Month", "Transaction Type"])["Amount"]
        .sum()
        .unstack(fill_value=0)
        .sort_index()
    )

    for col in ["credit", "debit"]:
        if col not in monthly.columns:
            monthly[col] = 0.0

    monthly["Balance"] = monthly["credit"] - monthly["debit"]

    return data, monthly


# ============================================================
# 3. CHART 1 — MONTHLY INCOME, EXPENSES AND BALANCE
# ============================================================

def create_monthly_chart(monthly):

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=monthly.index,
            y=monthly["credit"],
            mode="lines+markers",
            name="Income",
            line=dict(color="#2196F3", width=3)
        )
    )

    fig.add_trace(
        go.Scatter(
            x=monthly.index,
            y=monthly["debit"],
            mode="lines+markers",
            name="Expenses",
            line=dict(color="#FF9800", width=3)
        )
    )

    fig.add_trace(
        go.Scatter(
            x=monthly.index,
            y=monthly["Balance"],
            mode="lines+markers",
            name="Balance",
            line=dict(color="#4CAF50", width=3)
        )
    )

    fig.update_layout(
        title="Monthly Income, Expenses and Balance",
        xaxis_title="Month",
        yaxis_title="EUR",
        hovermode="x unified",
        template="plotly_white",
        height=420,
        margin=dict(l=40, r=30, t=60, b=40)
    )

    return fig


# ============================================================
# 4. CHART 2 — EXPENSE DISTRIBUTION BY CATEGORY
# ============================================================

def create_expense_chart(data):

    expenses = data[
        data["Transaction Type"] == "debit"
    ]

    category_totals = (
        expenses.groupby("Category")["Amount"]
        .sum()
        .sort_values(ascending=False)
        .reset_index()
    )

    fig = px.bar(
        category_totals,
        x="Amount",
        y="Category",
        orientation="h",
        title="Expense Distribution by Category",
        labels={
            "Amount": "Total Expenses (EUR)",
            "Category": "Expense Category"
        },
        color="Amount",
        color_continuous_scale="Blues"
    )

    fig.update_layout(
        template="plotly_white",
        height=450,
        yaxis=dict(autorange="reversed"),
        showlegend=False,
        coloraxis_showscale=False,
        margin=dict(l=40, r=30, t=60, b=40)
    )

    return fig


# ============================================================
# 5. CHART 3 — MONTHLY SURPLUS AND DEFICIT
# ============================================================

def create_balance_chart(monthly):

    colors = [
        "#4CAF50" if value >= 0 else "#F44336"
        for value in monthly["Balance"]
    ]

    fig = go.Figure()

    fig.add_trace(
        go.Bar(
            x=monthly.index,
            y=monthly["Balance"],
            marker_color=colors,
            name="Monthly Balance",
            hovertemplate=(
                "Month: %{x|%Y-%m}<br>"
                "Balance: €%{y:,.2f}<extra></extra>"
            )
        )
    )

    fig.add_hline(
        y=0,
        line_color="gray",
        line_width=1
    )

    fig.update_layout(
        title="Monthly Surplus and Deficit",
        xaxis_title="Month",
        yaxis_title="Balance (EUR)",
        template="plotly_white",
        height=420,
        margin=dict(l=40, r=30, t=60, b=40)
    )

    return fig


# ============================================================
# 6. EXECUTE LANGGRAPH QUERY
# ============================================================

def run_financial_analysis(question, rolling_months):

    if not question or not question.strip():
        return "Enter a question."

    try:

        result = app.invoke({
            "query": question.strip(),
            "result": "",
            "rolling_months": int(rolling_months)
        })

        answer = result.get(
            "result",
            "No results available."
        )

        if not answer or not str(answer).strip():
            return "No results available."

        return answer if isinstance(answer, str) else format_financial_response(
            answer, result.get("analysis_type", ""), question.strip()
        )

    except Exception as e:

        return f"Error during analysis: {str(e)}"


# ============================================================
# 7. GENERATE ALL THREE CHARTS
# ============================================================

def generate_charts():

    data, monthly = prepare_financial_data()

    return (
        create_monthly_chart(monthly),
        create_expense_chart(data),
        create_balance_chart(monthly)
    )


# ============================================================
# 8. GRADIO INTERFACE
# ============================================================

CUSTOM_CSS = """
.gradio-container { max-width: 1320px !important; margin: auto !important; }
#finance-header { padding: 16px 2px 10px; }
#finance-header h1 { font-size: 2rem; letter-spacing: -0.035em; margin-bottom: 4px; }
#finance-header p { color: #64748b; }
.question-example {
    position: relative !important;
    text-align: left !important;
    justify-content: flex-start !important;
    white-space: normal !important;
    min-height: 51px !important;
    padding: 11px 35px 11px 14px !important;
    line-height: 1.35 !important;
    font-weight: 450 !important;
}
.question-example::after {
    content: "+";
    position: absolute;
    right: 14px;
    top: 50%;
    transform: translateY(-50%);
    font-size: 20px;
    pointer-events: none;
    opacity: .65;
}
#analysis-panel, #decision-panel { min-width: 0 !important; }
#answer-panel textarea { font-family: inherit !important; line-height: 1.55 !important; }
"""

with gr.Blocks(title="Personal Finance Intelligence", css=CUSTOM_CSS) as demo:
    gr.Markdown(
        """# Personal Finance Intelligence

        Budget analysis and insights for financial decision-making · 24-month historical dataset
        """,
        elem_id="finance-header"
    )

    with gr.Row():
        rolling_selector = gr.Radio(
            choices=[("3 months", 3), ("6 months", 6), ("12 months", 12), ("2 years", 24)],
            value=12,
            label="Analysis period",
            interactive=True,
        )

    question_input = gr.Textbox(
        label="Your question",
        placeholder="Enter a question or select one of the examples in the two columns…",
        lines=2,
    )

    with gr.Row():
        analyze_button = gr.Button("Analyze question", variant="primary", scale=3)
        clear_button = gr.Button("Clear", variant="secondary", scale=1)

    gr.Markdown("### Answer")
    answer_output = gr.Textbox(
        label="Agent response",
        placeholder="The LangGraph agent's response will appear here.",
        lines=10,
        max_lines=28,
        interactive=False,
        elem_id="answer-panel",
    )

    analyze_button.click(
        fn=run_financial_analysis,
        inputs=[question_input, rolling_selector],
        outputs=answer_output,
    )
    question_input.submit(
        fn=run_financial_analysis,
        inputs=[question_input, rolling_selector],
        outputs=answer_output,
    )
    clear_button.click(fn=lambda: ("", ""), inputs=[], outputs=[question_input, answer_output])

    gr.Markdown("### Choose a question")
    gr.Markdown("Choose a situation that interests you: the question will appear in the field above. You can edit it before clicking **Analyze question**.")

    with gr.Row(equal_height=False):
        with gr.Column(scale=1, min_width=310, elem_id="analysis-panel"):
            gr.Markdown("#### Financial Analysis")
            gr.Markdown("Income, expenses, deficits, categories, and historical comparisons.")
            for example in ANALYSIS_QUESTIONS:
                button = gr.Button(example, variant="secondary", elem_classes=["question-example"])
                button.click(fn=lambda q=example: q, inputs=[], outputs=question_input)

        with gr.Column(scale=1, min_width=310, elem_id="decision-panel"):
            gr.Markdown("#### Financial Decision Intelligence")
            gr.Markdown("Practical questions about savings, unexpected events, financial security, and future decisions.")
            for example in DECISION_QUESTIONS:
                button = gr.Button(example, variant="secondary", elem_classes=["question-example"])
                button.click(fn=lambda q=example: q, inputs=[], outputs=question_input)

    gr.Markdown(
        """---
        **Methodological note.** The Expense-to-Income Ratio measures the share of income absorbed by expenses.
        Simulations use predefined assumptions: a 30% reduction in employment income for six months is not necessarily equivalent to losing an annual bonus. Initial liquidity estimated
        from historical cash flows is not the same as a verified bank balance. The selected period
        is passed to the agent; previously computed analyses retain their own calculation horizons.
        """
    )

# ============================================================
# 9. LAUNCH GRADIO
# ============================================================

demo.launch(share=True, debug=False, quiet=True, show_error=False)
```

**Agent for investment portfolio construction and optimization**

This phase of the project aims to model and optimize an investment portfolio through:

**1. Financial Asset Analysis (STOXX Europe 50 Equities)**

- An initial universe of 50 stocks belonging to the **STOXX Europe 50** is analyzed to select the **Top 25 equities** through a ranking based on five years of historical price data.

- The **distribution of historical price series** is analyzed, including the presence of **kurtosis**.

- **Future returns** are simulated through **bootstrap resampling** of historical daily returns, incorporating the empirical return distribution and the extreme events reflected in kurtosis. **Portfolio value projections** are then generated across different investment horizons and market scenarios.

Following the asset analysis and selection, the next step is to construct an efficient portfolio.

**2. Identification of an Efficient Portfolio**

- **Expected returns** and **annualized volatility** are calculated.

- The **covariance matrix** between assets is estimated, which is essential for capturing shared risk relationships.

- A **Monte Carlo simulation** of 5,000 portfolios with randomly assigned weights is performed, calculating **Expected Return**, **Volatility**, and **Sharpe Ratio** for each portfolio.

- Finally, the **maximum Sharpe ratio portfolio** is identified, representing the portfolio with the **best risk-adjusted return**, balancing expected performance and volatility.

```python
# ============================================================
# BOX 15 — STOXX EUROPE 50: HISTORICAL PRICE DATA COLLECTION
# ============================================================
# Download five years of daily closing prices for 50 stocks,
# convert all prices to EUR using historical daily exchange rates,
# and consolidate the results into a single dataset.
# ============================================================

# 1. DATE SETTINGS

end_time = datetime.now()
start_time = end_time - pd.DateOffset(years=5)

start_date = start_time.strftime('%Y-%m-%d')
end_date = (end_time + timedelta(days=1)).strftime('%Y-%m-%d')

DATA_DIR = "stoxx_europe_50_data"
os.makedirs(DATA_DIR, exist_ok=True)

# 2. STOXX EUROPE 50 CONSTITUENTS

# The 50 tickers and their metadata are defined in previous boxes.
assert len(tickers) == 50
assert set(tickers) == set(stock_metadata.index)
if stock_metadata["Currency"].isna().any():
    missing = stock_metadata.index[stock_metadata["Currency"].isna()].tolist()
    raise ValueError(f"Currency unavailable for: {missing}. Rerun BOX 14.")

# 3. HISTORICAL PRICE DOWNLOAD AND EUR CONVERSION

# Unadjusted closing prices are used; returns exclude dividends.
dataframes = {}
download_errors = {}
PRICE_COLUMNS = ["Close"]
logging.getLogger("yfinance").setLevel(logging.CRITICAL)


def normalize_prices(data):
    data = data.copy()
    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)
    data = data[PRICE_COLUMNS].copy()
    data.index = pd.to_datetime(data.index)
    if data.index.tz is not None:
        data.index = data.index.tz_localize(None)
    data.index = data.index.normalize()
    data.index.name = "Date"
    for col in PRICE_COLUMNS:
        data[col] = pd.to_numeric(data[col], errors="coerce")
    data.sort_index(inplace=True)
    data = data[~data.index.duplicated(keep="last")]
    return data


# UK prices quoted in pence (GBp/GBX) are divided by 100.
# Currency information is obtained from stock_metadata.
def currency_to_base(currency):
    c = str(currency).strip()
    if c in ("GBp", "GBX", "GBx"):
        return "GBP", 0.01
    return c.upper(), 1.0


base_currencies = {
    currency_to_base(c)[0]
    for c in stock_metadata["Currency"]
    if currency_to_base(c)[0] != "EUR"
}

# Historical daily FX rates convert local closing prices to EUR.
# Missing FX observations are forward-filled for up to 3 days.
fx_series = {}
fx_errors = {}

for currency in sorted(base_currencies):
    fx_ticker = f"{currency}EUR=X"
    try:
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            fx_data = yf.download(
                fx_ticker,
                start=(start_time - timedelta(days=10)).strftime('%Y-%m-%d'),
                end=end_date,
                interval="1d",
                auto_adjust=False,
                repair=False,
                progress=False,
                threads=False
            )
        if fx_data.empty:
            raise ValueError("Historical FX rate unavailable")
        fx_close = normalize_prices(fx_data)["Close"]
        if fx_close.empty or (fx_close <= 0).any():
            raise ValueError("Empty or invalid FX series")
        fx_series[currency] = fx_close
    except Exception as e:
        fx_errors[currency] = str(e)

if fx_errors:
    raise RuntimeError(f"EUR FX download failed: {fx_errors}")

for ticker, name in tickers.items():
    try:
        quote_currency = stock_metadata.loc[ticker, "Currency"]
        base_currency, unit_factor = currency_to_base(quote_currency)

        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            data = yf.download(
                ticker,
                start=start_date,
                end=end_date,
                interval="1d",
                auto_adjust=False,
                repair=False,
                progress=False,
                threads=False
            )

        if data.empty:
            raise ValueError("No data available")
        df_close = normalize_prices(data)
        if df_close.empty:
            raise ValueError("Empty dataset")

        if base_currency == "EUR":
            df_close["Close"] = df_close["Close"] * unit_factor
        else:
            fx = fx_series[base_currency]
            calendar = fx.index.union(df_close.index).sort_values()
            aligned_fx = fx.reindex(calendar).ffill(limit=3).reindex(df_close.index)
            if aligned_fx.isna().any():
                dates = aligned_fx.index[aligned_fx.isna()].strftime('%Y-%m-%d').tolist()
                raise ValueError(f"Missing {base_currency}/EUR exchange rate for {dates[:5]}")
            df_close["Close"] = df_close["Close"] * unit_factor * aligned_fx

        df_close = df_close.round(4)
        filename = os.path.join(DATA_DIR, f"{ticker.replace('.', '_')}_close_EUR.csv")
        df_close.to_csv(filename)
        dataframes[ticker] = df_close
    except Exception as e:
        download_errors[ticker] = str(e)

if not dataframes:
    raise RuntimeError(f"No EUR price series available: {download_errors}")

# 4. CONSOLIDATED PRICE DATASET

# One column per ticker; missing values are handled in BOX 16.
close_df2s = []

for ticker, df_close in dataframes.items():

    close_df2 = df_close[["Close"]].copy()
    close_df2.columns = [ticker]

    close_df2s.append(close_df2)

df2 = pd.concat(close_df2s, axis=1)

df2 = df2.sort_index(axis=1)
df2 = df2.sort_index()

# 5. FINAL DISPLAY

print("\n" + "=" * 90)
print("CLOSING PRICES IN EUR — STOXX EUROPE 50")
print("=" * 90)

print(f"Requested period: 5 years from {start_date} (through the latest available trading session)")
print(f"Available tickers: {df2.shape[1]}/50")
print(f"Total trading sessions: {df2.shape[0]}")
print("Currency of all displayed prices: EUR")

TICKERS_PER_BLOCK = 15

for i in range(0, df2.shape[1], TICKERS_PER_BLOCK):

    block = df2.iloc[
        :5,
        i:i + TICKERS_PER_BLOCK
    ]

    print(f"\nTICKERS {i + 1}-{min(i + TICKERS_PER_BLOCK, df2.shape[1])}")
    print(block.to_string())

# 6. FINAL CHECK

if download_errors:
    print(f"\nUnavailable stocks: {len(download_errors)}")
    print(", ".join(download_errors.keys()))
```

## **Investment Assessment & Portfolio Construction**

```python
# ============================================================
# BOX 18
#
# Equity Ranking & Top 25 Selection — STOXX Europe 50
#
# Ranking:
# 50% annualized return (higher = better)
# 50% annualized volatility (lower = better)
#
# The combined score starts at 1 and increases progressively.
# A lower combined score indicates a better overall ranking.
# This is because rank 1 is assigned to the highest annualized
# return and rank 1 to the lowest annualized volatility.
# Therefore, a stock ranking first in both categories receives
# a combined score of 1, while less favorable stocks receive
# progressively higher scores.
#
# Output: 25 selected stocks for BOX 19
# ============================================================


# ============================================================
# 1. DAILY RETURNS
# ============================================================

# EUR-denominated prices already prepared in BOXES 15 and 16.
daily_returns = df2.pct_change(fill_method=None)

TRADING_DAYS = 252

# ============================================================
# 2. ANNUALIZED RETURN AND VOLATILITY
# ============================================================

annual_returns = daily_returns.mean() * TRADING_DAYS

annual_volatility = (
    daily_returns.std() * np.sqrt(TRADING_DAYS)
)

# ============================================================
# 3. METRICS TABLE
# ============================================================

ranking_df = pd.DataFrame({
    "Annualized Return": annual_returns,
    "Annualized Volatility": annual_volatility
})

# Add metadata already retrieved in BOX 14.
ranking_df = ranking_df.join(
    stock_metadata[["Company", "Country", "Sector"]],
    how="left"
)

# Verify that all 50 stocks are available.
if len(ranking_df) != 50:
    raise ValueError(
        f"Expected 50 stocks, found {len(ranking_df)}."
    )

if ranking_df[
    ["Annualized Return", "Annualized Volatility"]
].isna().any().any():
    raise ValueError(
        "Missing metrics: check data from BOXES 15-16."
    )

# ============================================================
# 4. COMBINED RANKING 50% / 50%
# ============================================================

# Position 1 = best result.
ranking_df["Return Rank"] = ranking_df[
    "Annualized Return"
].rank(ascending=False, method="average")

ranking_df["Volatility Rank"] = ranking_df[
    "Annualized Volatility"
].rank(ascending=True, method="average")

# Lower score = better.
ranking_df["Combined Score"] = (
    0.5 * ranking_df["Return Rank"]
    + 0.5 * ranking_df["Volatility Rank"]
)

ranking_df = ranking_df.sort_values(
    ["Combined Score", "Annualized Return"],
    ascending=[True, False]
)

ranking_df["Ranking Position"] = (
    ranking_df["Combined Score"].rank(method="min", ascending=True).astype(int)
)

# ============================================================
# 5. TOP 25 SELECTION
# ============================================================

top_25 = ranking_df.head(25).copy()

# Variable for BOX 19.
selected_tickers = top_25.index.tolist()

# ============================================================
# 6. RESULTS PRESENTATION
# ============================================================

display_columns = [
    "Ranking Position",
    "Company",
    "Country",
    "Sector",
    "Annualized Return",
    "Annualized Volatility",
    "Combined Score"
]

def format_ranking(df):
    result = df[display_columns].copy()

    result["Annualized Return"] = (
        result["Annualized Return"] * 100
    )
    result["Annualized Volatility"] = (
        result["Annualized Volatility"] * 100
    )

    return result.to_string(
        formatters={
            "Annualized Return": lambda x: f"{x:.2f}%",
            "Annualized Volatility": lambda x: f"{x:.2f}%",
            "Combined Score": lambda x: f"{x:.2f}"
        }
    )

print("\n" + "=" * 100)
print("Ranking designed to select 25 stocks from an initial universe of 50 STOXX Europe 50 constituents, based on historical price series over the past 5 years.")
print("=" * 100)
print("Methodology: 50% annualized return ranking (higher = better) + 50% annualized volatility ranking (lower = better).")
print("Combined Score = 0.5 × Return Rank + 0.5 × Volatility Rank; lower score = better ranking position.")
print("The five years represent the historical period; 252 trading days are the annualization convention.")
print("The selection of 25 stocks precedes portfolio and portfolio-weight optimization.")
print("\nFULL RANKING — the top 25 stocks are selected (ties in Combined Score):")
print(format_ranking(ranking_df))
```

```python
# ============================================================
# BOX 19
#
# Monte Carlo Portfolio Optimization — STOXX Europe 50
#
# 5,000 simulations on the 25 stocks selected in BOX 18
# Maximum 4 stocks per sector in each portfolio
# Selection of the portfolio with the highest Sharpe Ratio
# ============================================================


# ============================================================
# 1. DATA FOR THE 25 SELECTED STOCKS
# ============================================================

portfolio_returns = daily_returns[selected_tickers].dropna()

TRADING_DAYS = 252
RISK_FREE_RATE = 0.0
NUM_PORTFOLIOS = 5000
MAX_ASSETS_PER_SECTOR = 4
RANDOM_SEED = 42

if portfolio_returns.empty:
    raise ValueError(
        "No returns available for the 25 stocks."
    )

num_assets = len(selected_tickers)

mean_returns = portfolio_returns.mean() * TRADING_DAYS
cov_matrix = portfolio_returns.cov() * TRADING_DAYS

# Sectors of the stocks selected in BOX 18.
asset_sectors = top_25.loc[selected_tickers, "Sector"]

if asset_sectors.isna().any():
    raise ValueError(
        "Sector missing for one or more selected stocks."
    )

# Indices of stocks grouped by sector.
sector_indices = {
    sector: np.where(asset_sectors.values == sector)[0]
    for sector in asset_sectors.unique()
}

# ============================================================
# 2. FUNCTION FOR PORTFOLIO METRICS
# ============================================================

def portfolio_metrics(weights):

    expected_return = np.dot(
        weights,
        mean_returns.values
    )

    volatility = np.sqrt(
        weights.T @ cov_matrix.values @ weights
    )

    sharpe = (
        (expected_return - RISK_FREE_RATE) / volatility
        if volatility > 0 else np.nan
    )

    return expected_return, volatility, sharpe

# ============================================================
# 3. MONTE CARLO — 5,000 SIMULATIONS
# ============================================================

rng = np.random.default_rng(RANDOM_SEED)

mc_weights = np.zeros((NUM_PORTFOLIOS, num_assets))
mc_results = np.zeros((NUM_PORTFOLIOS, 3))

for i in range(NUM_PORTFOLIOS):

    # Randomly select up to 4 stocks per sector.
    selected_indices = []

    for sector, indices in sector_indices.items():

        n_select = min(
            MAX_ASSETS_PER_SECTOR,
            len(indices)
        )

        chosen = rng.choice(
            indices,
            size=n_select,
            replace=False
        )

        selected_indices.extend(chosen)

    # Assign weights only to the selected stocks.
    weights = np.zeros(num_assets)

    random_weights = rng.dirichlet(
        np.ones(len(selected_indices))
    )

    weights[selected_indices] = random_weights

    mc_weights[i] = weights
    mc_results[i] = portfolio_metrics(weights)

mc_df = pd.DataFrame(
    mc_results,
    columns=[
        "Annualized Return",
        "Annualized Volatility",
        "Sharpe Ratio"
    ]
)

# ============================================================
# 4. SELECTION OF THE MAXIMUM SHARPE PORTFOLIO
# ============================================================

best_idx = mc_df["Sharpe Ratio"].idxmax()

optimal_weights = mc_weights[best_idx].copy()

optimal_return, optimal_volatility, optimal_sharpe = (
    portfolio_metrics(optimal_weights)
)

portfolio_weights = pd.Series(
    optimal_weights,
    index=selected_tickers,
    name="Portfolio Weight"
).sort_values(ascending=False)

# ============================================================
# 5. SECTOR CONSTRAINT VALIDATION
# ============================================================

active_tickers = portfolio_weights[
    portfolio_weights > 0
].index

portfolio_sector_counts = (
    asset_sectors.loc[active_tickers]
    .value_counts()
)

if (portfolio_sector_counts > MAX_ASSETS_PER_SECTOR).any():
    raise ValueError(
        "The portfolio exceeds the limit of 4 stocks per sector."
    )

# ============================================================
# 6. RESULTS
# ============================================================

print("\n" + "=" * 90)
print("MONTE CARLO PORTFOLIO OPTIMIZATION — STOXX EUROPE 50")
print("=" * 90)

print(f"\nSimulated portfolios: {NUM_PORTFOLIOS}")
print(f"Available stocks: {num_assets}")
print(f"Maximum stocks per sector: {MAX_ASSETS_PER_SECTOR}")

print("\nOPTIMAL PORTFOLIO — MAXIMUM SHARPE RATIO")

print(f"Annualized Expected Return: {optimal_return:.2%}")
print(f"Annualized Volatility:      {optimal_volatility:.2%}")
print(f"Sharpe Ratio:               {optimal_sharpe:.4f}")

print("\nPORTFOLIO WEIGHTS")

print(
    (portfolio_weights[portfolio_weights > 0] * 100).to_string(
        float_format=lambda x: f"{x:.2f}%"
    )
)

print(f"\nTotal portfolio weight: {portfolio_weights.sum():.2%}")
print(f"Stocks included: {len(active_tickers)}")

print("\nSECTOR DISTRIBUTION — NUMBER OF STOCKS")
print(portfolio_sector_counts.to_string())

# ============================================================
# 7. MONTE CARLO CHART — RISK / RETURN
# ============================================================

plt.figure(figsize=(11, 7))

scatter = plt.scatter(
    mc_df["Annualized Volatility"] * 100,
    mc_df["Annualized Return"] * 100,
    c=mc_df["Sharpe Ratio"],
    cmap="viridis",
    s=10,
    alpha=0.6
)

plt.colorbar(scatter, label="Sharpe Ratio")

plt.scatter(
    optimal_volatility * 100,
    optimal_return * 100,
    marker="*",
    s=300,
    color="red",
    edgecolor="black",
    label="Optimal Portfolio — Maximum Sharpe"
)

plt.xlabel("Annualized Volatility (%)")
plt.ylabel("Annualized Expected Return (%)")
plt.title("Monte Carlo Portfolio Optimization — STOXX Europe 50")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Return Estimation Methodology and Scenario Analysis — Historical Bootstrap

**Preliminary Analysis of Return Distributions**

The Jarque-Bera normality test and *excess kurtosis* analysis assess the statistical characteristics of daily returns for the individual assets included in the portfolio.

A p-value below 0.05 leads to the rejection of the normality hypothesis. Positive excess kurtosis indicates heavier tails than a Gaussian distribution, associated with a higher frequency of extreme returns.

**The observed high excess kurtosis and non-normal return distributions of the individual assets provide the statistical rationale for applying historical bootstrap simulations to the optimized portfolio as a whole in subsequent boxes.**

**Why Don't We Rely Solely on Annualized Expected Returns?**

Annualized expected return provides an average estimate of asset performance but does not adequately describe the variability of possible outcomes, particularly in the presence of non-normal distributions and extreme events.

To assess the financial consequences of an investment, it is necessary to consider not only the average expected return but also potential loss and gain scenarios over different time horizons.

**The Bootstrap Method**

Bootstrap is a non-parametric statistical simulation technique that generates new samples by randomly resampling historical observations with replacement.

In the subsequent portfolio analysis, **daily returns are jointly resampled across the assets included in the optimized portfolio**, preserving their contemporaneous relationships. These returns are combined using the optimal portfolio weights to generate **1,000 possible trajectories for the portfolio as a whole**, providing a distribution of potential investment outcomes.

Unlike a parametric model based on the normal distribution, bootstrap does not impose a specific theoretical distribution on returns and can reproduce extreme events observed in the historical sample.

However, the method remains dependent on observed data: it cannot anticipate events that have never occurred in the historical sample, and independent daily resampling does not automatically preserve the temporal dependence of returns.

**Construction of Financial Scenarios**

The subsequent simulations estimate three portfolio-level scenarios:

- **Unfavorable (5th percentile):** an outcome in the lower tail of the simulated distribution.
- **Central (50th percentile):** the median outcome of the simulations.
- **Favorable (95th percentile):** an outcome in the upper tail of the simulated distribution.

Scenarios are calculated over flexible time horizons of up to 24 months to represent cumulative portfolio gains and losses according to investment duration.

Percentage returns can subsequently be applied to the initial investment capital, determined as 50% of the positive net surplus accumulated over the previous 12 months (Income − Expenses).

**Objective**

To move from a single expected return estimate to **a comprehensive distribution of potential outcomes for the optimized portfolio**, providing AI agents with quantitative scenarios to assess investment performance and its combined impact on personal finances, liquidity, and financial sustainability.

```python
# ============================================================
# BOX 22
#
# Portfolio Bootstrap — Rolling Gains & Losses
#
# 1,000 bootstrap simulations of the optimal portfolio
# Flexible horizon: from 1 day to 24 months
# Scenarios: unfavorable P5, central P50, favorable P95
# Initial allocation: 50% of accumulated net savings is invested,
# while the remaining 50% is retained as personal liquidity.
# Both components evolve separately over the following 24 months.
# Output: percentage gains and losses for each horizon
# ============================================================

# Methodology — Initial Savings Allocation
# After the 12-month observation period, accumulated net savings
# amount to EUR 6,264.52, equally divided between investment capital
# (EUR 3,132.26) and available personal liquidity (EUR 3,132.26).

# ============================================================
# 1. PARAMETERS
# ============================================================

NUM_BOOTSTRAP_SIMULATIONS = 1000
BOOTSTRAP_MONTHS = 24
TRADING_DAYS = 252
TRADING_DAYS_PER_MONTH = 21
BOOTSTRAP_DAYS = BOOTSTRAP_MONTHS * TRADING_DAYS_PER_MONTH
BOOTSTRAP_SEED = 42

INVESTMENT_ALLOCATION = 0.50

# ============================================================
# 2. INITIAL CAPITAL — PERSONAL FINANCE BOX 1
# ============================================================

# BOX 1 already contains the monthly DataFrame:
# Income, Expense and Balance = Income - Expense.

monthly_finance = monthly.sort_index().tail(12).copy()

if len(monthly_finance) < 12:
    raise ValueError(
        "At least 12 months of financial data are required."
    )

income_12m = float(monthly_finance["Income"].sum())
expenses_12m = float(monthly_finance["Expense"].sum())

net_surplus_12m = income_12m - expenses_12m

available_surplus_12m = max(net_surplus_12m, 0.0)

initial_investment = (
    available_surplus_12m * INVESTMENT_ALLOCATION
)

remaining_liquidity = (
    available_surplus_12m - initial_investment
)

# ============================================================
# 3. HISTORICAL PORTFOLIO RETURNS
# ============================================================

bootstrap_tickers = portfolio_weights[
    portfolio_weights > 0
].index.tolist()

bootstrap_weights = portfolio_weights.loc[
    bootstrap_tickers
].to_numpy(dtype=float)

bootstrap_returns = portfolio_returns[
    bootstrap_tickers
].dropna()

if bootstrap_returns.empty:
    raise ValueError(
        "Historical returns are not available for bootstrap."
    )

if not np.isclose(bootstrap_weights.sum(), 1.0):
    raise ValueError(
        "The sum of portfolio weights must equal 1."
    )

historical_returns = bootstrap_returns.to_numpy()

# ============================================================
# 4. MULTIVARIATE BOOTSTRAP — 1,000 TRAJECTORIES
# ============================================================

rng = np.random.default_rng(BOOTSTRAP_SEED)

sample_indices = rng.integers(
    low=0,
    high=len(historical_returns),
    size=(NUM_BOOTSTRAP_SIMULATIONS, BOOTSTRAP_DAYS)
)

# Joint resampling of historical trading days:
# preserves contemporaneous correlations among stocks.
sampled_returns = historical_returns[sample_indices]

portfolio_bootstrap_returns = (
    sampled_returns @ bootstrap_weights
)

# ============================================================
# 5. CUMULATIVE GAINS AND LOSSES
# ============================================================

# Initial value normalized to 1.
# Percentage simulations are independent of the invested capital.

bootstrap_growth = np.concatenate(
    [
        np.ones((NUM_BOOTSTRAP_SIMULATIONS, 1)),
        np.cumprod(
            1 + portfolio_bootstrap_returns,
            axis=1
        )
    ],
    axis=1
)

bootstrap_paths = bootstrap_growth

# Cumulative percentage gain or loss.
bootstrap_pnl_pct = (bootstrap_growth - 1) * 100

# Gain or loss in EUR on invested capital.
bootstrap_pnl_eur = (
    bootstrap_growth - 1
) * initial_investment

# Investment value in EUR.
bootstrap_portfolio_values_eur = (
    bootstrap_growth * initial_investment
)

# ============================================================
# 6. DAILY SCENARIOS
# ============================================================

bootstrap_unfavorable = np.percentile(
    bootstrap_pnl_pct, 5, axis=0
)

bootstrap_central = np.percentile(
    bootstrap_pnl_pct, 50, axis=0
)

bootstrap_favorable = np.percentile(
    bootstrap_pnl_pct, 95, axis=0
)

bootstrap_scenarios = pd.DataFrame({
    "Day": np.arange(BOOTSTRAP_DAYS + 1),
    "Month": np.arange(BOOTSTRAP_DAYS + 1)
             / TRADING_DAYS_PER_MONTH,
    "Unfavorable P5 (%)": bootstrap_unfavorable,
    "Central P50 (%)": bootstrap_central,
    "Favorable P95 (%)": bootstrap_favorable
})

# ============================================================
# 7. FUNCTION FOR ANY TIME HORIZON
# ============================================================

def get_portfolio_scenarios(months=None, days=None):

    if (months is None) == (days is None):
        raise ValueError(
            "Specify either months or days, not both."
        )

    if months is not None:
        horizon_days = round(
            float(months) * TRADING_DAYS_PER_MONTH
        )
    else:
        horizon_days = int(days)

    if not 1 <= horizon_days <= BOOTSTRAP_DAYS:
        raise ValueError(
            f"Allowed horizon: 1-{BOOTSTRAP_DAYS} "
            "trading days."
        )

    returns_at_horizon = bootstrap_pnl_pct[:, horizon_days]

    p5, p50, p95 = np.percentile(
        returns_at_horizon,
        [5, 50, 95]
    )

    return {
        "Months": horizon_days / TRADING_DAYS_PER_MONTH,
        "Trading Days": horizon_days,
        "Unfavorable P5 (%)": float(p5),
        "Central P50 (%)": float(p50),
        "Favorable P95 (%)": float(p95),
        "Unfavorable P5 (EUR)": float(
            initial_investment * p5 / 100
        ),
        "Central P50 (EUR)": float(
            initial_investment * p50 / 100
        ),
        "Favorable P95 (EUR)": float(
            initial_investment * p95 / 100
        )
    }

# ============================================================
# 8. ROLLING TABLE — GAINS AND LOSSES
# ============================================================

rolling_horizons = [1, 3, 6, 9, 12, 18, 24]

bootstrap_rolling_summary = pd.DataFrame([
    get_portfolio_scenarios(months=m)
    for m in rolling_horizons
])

print("\n" + "=" * 100)
print("PORTFOLIO BOOTSTRAP — ROLLING GAINS & LOSSES")
print("=" * 100)

print(f"\nBootstrap simulations: {NUM_BOOTSTRAP_SIMULATIONS}")
print(f"Maximum horizon: {BOOTSTRAP_MONTHS} months")

print("\nCUMULATIVE GAINS / LOSSES BY HORIZON")

print(f"\nIncome — last 12 months:          EUR {income_12m:,.2f}")
print(f"Expenses — last 12 months:        EUR {expenses_12m:,.2f}")
print(f"Net surplus — last 12 months:     EUR {net_surplus_12m:,.2f}")
print(f"Initial investment — 50%:        EUR {initial_investment:,.2f}")
print(f"Remaining surplus liquidity:     EUR {remaining_liquidity:,.2f}")
print("INITIAL BASIS: net savings frozen after 12 months; 50% invested and 50% retained as personal liquidity. The two amounts evolve separately.")

print("\nGAINS / LOSSES BY INVESTMENT HORIZON")

print(
    bootstrap_rolling_summary[
        [
            "Months",
            "Unfavorable P5 (%)",
            "Central P50 (%)",
            "Favorable P95 (%)"
        ]
    ].to_string(
        index=False,
        formatters={
            "Months": lambda x: f"{x:.0f}",
            "Unfavorable P5 (%)": lambda x: f"{x:+.2f}%",
            "Central P50 (%)": lambda x: f"{x:+.2f}%",
            "Favorable P95 (%)": lambda x: f"{x:+.2f}%"
        }
    )
)

# ============================================================
# 9. ROLLING CHART — PERCENTAGE GAINS AND LOSSES
# ============================================================

plt.figure(figsize=(12, 7))

months_axis = (
    np.arange(BOOTSTRAP_DAYS + 1)
    / TRADING_DAYS_PER_MONTH
)

plt.fill_between(
    months_axis,
    bootstrap_unfavorable,
    bootstrap_favorable,
    alpha=0.15,
    label="P5–P95 Range"
)

plt.plot(
    months_axis,
    bootstrap_unfavorable,
    label="Unfavorable — P5",
    linewidth=2
)

plt.plot(
    months_axis,
    bootstrap_central,
    label="Central — P50",
    linewidth=2.5
)

plt.plot(
    months_axis,
    bootstrap_favorable,
    label="Favorable — P95",
    linewidth=2
)

plt.axhline(
    0,
    color="black",
    linestyle="--",
    linewidth=1,
    label="Break-even"
)

plt.xlabel("Months")
plt.ylabel("Cumulative Gain / Loss (%)")
plt.title(
    "Portfolio Bootstrap — Rolling Gains & Losses\n"
    f"Initial Investment: EUR {initial_investment:,.2f} "
    "(50% of 12-Month Net Surplus)"
)

plt.xlim(0, BOOTSTRAP_MONTHS)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# 10. OUTPUT FOR BOX 22 AND THE AGENTS
# ============================================================

# income_12m, expenses_12m:
# Income and expenses over the last 12 months from BOX 1.

# net_surplus_12m:
# Net surplus over the last 12 months.

# initial_investment:
# Invested capital, equal to 50% of the positive surplus.

# remaining_liquidity:
# Share of the positive surplus retained as liquidity.

# bootstrap_pnl_pct:
# Percentage gains and losses across the 1,000 trajectories.

# bootstrap_pnl_eur:
# Gains and losses in EUR across the 1,000 trajectories.

# bootstrap_portfolio_values_eur:
# Investment value in EUR for each trajectory.

# bootstrap_scenarios:
# P5, P50 and P95 scenarios for each trading day.

# bootstrap_rolling_summary:
# Results at 1, 3, 6, 9, 12, 18 and 24 months.

# get_portfolio_scenarios(months=12):
# Percentage and monetary scenarios at 12 months.

# get_portfolio_scenarios(days=100):
# Percentage and monetary scenarios at 100 trading days.

print("\nBootstrap rolling scenarios ready for BOX 22.")
```

## **Quantitative Tools for the Agentic System**

In the preceding boxes, we illustrated the calculation methodologies and analyzed their results. At this stage, we develop the Python tools that will autonomously perform these calculations and analyses within the agentic system.

```python
# ============================================================
# BOX 23
#
# Investment Analysis Tools — STOXX Europe 50
#
# Autonomous analysis and selection engine:
# - Calculation of daily returns
# - Calculation of average annualized return
# - Calculation of annualized volatility
# - Combined ranking: 50% return, 50% volatility
# - Selection of the top 25 stocks
#
# Input: daily prices in EUR and metadata
# Output: full ranking, Top 25, and selected tickers
# ============================================================


# ============================================================
# 1. MOTORE DI CALCOLO DEL RANKING
# ============================================================

def calculate_equity_ranking(
    prices,
    metadata,
    top_n=25,
    trading_days=252
):
    # Input validation.
    if not isinstance(prices, pd.DataFrame):
        raise TypeError("prices must be a DataFrame.")

    if not isinstance(metadata, pd.DataFrame):
        raise TypeError("metadata must be a DataFrame.")

    if prices.empty:
        raise ValueError("Price dataset is empty.")

    if prices.columns.has_duplicates:
        raise ValueError("Duplicate tickers in price data.")

    if metadata.index.has_duplicates:
        raise ValueError("Duplicate tickers in metadata.")

    if len(prices.columns) != 50:
        raise ValueError(
            f"Expected 50 stocks, found {len(prices.columns)}."
        )

    if not isinstance(top_n, int) or not 1 <= top_n <= 50:
        raise ValueError("top_n must be between 1 and 50.")

    if trading_days <= 0:
        raise ValueError("trading_days must be positive.")

    metadata_columns = ["Company", "Country", "Sector"]

    if not all(c in metadata.columns for c in metadata_columns):
        raise ValueError(
            "Incomplete metadata: Company, Country, and Sector required."
        )

    if not set(prices.columns).issubset(set(metadata.index)):
        raise ValueError(
            "Missing metadata for one or more stocks."
        )

    # Independent copy of price data.
    price_data = prices.copy()

    price_data.index = pd.to_datetime(price_data.index)
    price_data = price_data.sort_index()

    if price_data.index.has_duplicates:
        raise ValueError("Duplicate dates in price dataset.")

    price_data = price_data.apply(
        pd.to_numeric,
        errors="coerce"
    )

    if (price_data <= 0).any().any():
        raise ValueError(
            "Available prices must be positive."
        )

    # ========================================================
    # 2. DAILY RETURNS
    # ========================================================

    daily_returns = price_data.pct_change(
        fill_method=None
    )

    # ========================================================
    # 3. ANNUALIZED METRICS
    # ========================================================

    annual_returns = (
        daily_returns.mean() * trading_days
    )

    annual_volatility = (
        daily_returns.std() * np.sqrt(trading_days)
    )

    ranking_df = pd.DataFrame({
        "Annualized Return": annual_returns,
        "Annualized Volatility": annual_volatility
    })

    ranking_df = ranking_df.join(
        metadata[metadata_columns],
        how="left"
    )

    if ranking_df[
        ["Annualized Return", "Annualized Volatility"]
    ].isna().any().any():
        raise ValueError(
            "Returns or volatility unavailable "
            "for all stocks."
        )

    if ranking_df[metadata_columns].isna().any().any():
        raise ValueError(
            "Incomplete metadata for one or more stocks."
        )

    # ========================================================
    # 4. RANKING COMBINATO 50% / 50%
    # ========================================================

    ranking_df["Return Rank"] = ranking_df[
        "Annualized Return"
    ].rank(
        ascending=False,
        method="average"
    )

    ranking_df["Volatility Rank"] = ranking_df[
        "Annualized Volatility"
    ].rank(
        ascending=True,
        method="average"
    )

    ranking_df["Combined Score"] = (
        0.5 * ranking_df["Return Rank"]
        + 0.5 * ranking_df["Volatility Rank"]
    )

    ranking_df = ranking_df.sort_values(
        ["Combined Score", "Annualized Return"],
        ascending=[True, False]
    )

    ranking_df["Ranking Position"] = (
        ranking_df["Combined Score"].rank(method="min", ascending=True).astype(int)
    )

    # ========================================================
    # 5. TOP 25 SELECTION
    # ========================================================

    top_25 = ranking_df.head(top_n).copy()

    selected_tickers = top_25.index.tolist()

    # ========================================================
    # 6. STRUCTURED OUTPUT
    # ========================================================

    return {
        "universe": "STOXX Europe 50",
        "total_stocks": len(ranking_df),
        "selected_stocks": len(selected_tickers),
        "trading_days": trading_days,
        "return_weight": 0.50,
        "volatility_weight": 0.50,
        "daily_returns": daily_returns,
        "ranking_df": ranking_df,
        "top_25": top_25,
        "selected_tickers": selected_tickers
    }

# ============================================================
# 7. TOOL — FULL RANKING
# ============================================================

def get_equity_ranking(
    prices,
    metadata,
    limit=None
):
    result = calculate_equity_ranking(
        prices=prices,
        metadata=metadata
    )

    ranking = result["ranking_df"]

    if limit is not None:
        if not isinstance(limit, int) or not 1 <= limit <= 50:
            raise ValueError("limit must be between 1 and 50.")

        ranking = ranking.head(limit)

    return ranking.copy()

# ============================================================
# 8. TOOL — TOP 25 SELECTION
# ============================================================

def get_top_25_stocks(
    prices,
    metadata
):
    result = calculate_equity_ranking(
        prices=prices,
        metadata=metadata
    )

    return result["top_25"].copy()

# ============================================================
# 9. TOOL — SINGLE STOCK ANALYSIS
# ============================================================

def get_stock_analysis(
    ticker,
    prices,
    metadata
):
    result = calculate_equity_ranking(
        prices=prices,
        metadata=metadata
    )

    ticker = str(ticker).strip().upper()

    matching_tickers = {
        str(t).upper(): t
        for t in result["ranking_df"].index
    }

    if ticker not in matching_tickers:
        return {
            "ticker": ticker,
            "found": False,
            "message": "Ticker not found in the analyzed universe."
        }

    original_ticker = matching_tickers[ticker]

    stock = result["ranking_df"].loc[
        original_ticker
    ].to_dict()

    return {
        "ticker": original_ticker,
        "found": True,
        "selected_top_25": (
            original_ticker in result["selected_tickers"]
        ),
        "metrics": stock
    }

# ============================================================
# 10. TOOL REGISTRY
# ============================================================

investment_analysis_tools = {
    "calculate_equity_ranking": calculate_equity_ranking,
    "get_equity_ranking": get_equity_ranking,
    "get_top_25_stocks": get_top_25_stocks,
    "get_stock_analysis": get_stock_analysis
}

# ============================================================
# 11. TEST WITH CURRENT NOTEBOOK DATA
# ============================================================

# df2 and stock_metadata are only test inputs.
# The engine does not use precomputed results from BOX 18.

investment_analysis_result = calculate_equity_ranking(
    prices=df2,
    metadata=stock_metadata
)

test_ranking = investment_analysis_result["ranking_df"]
test_top_25 = investment_analysis_result["top_25"]
test_tickers = investment_analysis_result["selected_tickers"]

assert len(test_ranking) == 50
assert len(test_top_25) == 25
assert len(test_tickers) == 25

print("\n" + "=" * 90)
print("INVESTMENT ANALYSIS TOOL — CALCULATION TEST")
print("=" * 90)

print(f"\nStocks analyzed: {len(test_ranking)}")
print(f"Stocks selected: {len(test_top_25)}")
print("Ranking: 50% annualized return / 50% volatility")

print("\nTOP 25 — CALCULATED BY THE TOOL")

print(
    test_top_25[
        [
            "Ranking Position",
            "Company",
            "Country",
            "Sector",
            "Annualized Return",
            "Annualized Volatility",
            "Combined Score"
        ]
    ].to_string(
        formatters={
            "Annualized Return": lambda x: f"{x:.2%}",
            "Annualized Volatility": lambda x: f"{x:.2%}",
            "Combined Score": lambda x: f"{x:.2f}"
        }
    )
)

print("\nINVESTMENT ANALYSIS CALCULATION TOOL READY.")
```

```python
# ============================================================
# BOX 24
#
# Portfolio Optimization Tool — STOXX Europe 50
#
# Autonomous Monte Carlo optimization engine:
# - 5,000 simulated portfolios using the Top 25 stocks
# - Maximum 4 stocks per sector in each portfolio
# - Calculation of return, volatility, and Sharpe Ratio
# - Selection of the highest Sharpe Ratio among simulated portfolios
# - Calculation of portfolio composition and percentage weights
#
# Input: Investment Analysis engine results (BOX 23)
# Output: optimal portfolio and Monte Carlo simulations
#
# No additional numerical optimization
# ============================================================


# ============================================================
# 1. AUTONOMOUS OPTIMIZATION ENGINE
# ============================================================

def calculate_optimal_portfolio(
    investment_result,
    num_portfolios=5000,
    max_assets_per_sector=4,
    risk_free_rate=0.0,
    trading_days=252,
    random_seed=42
):
    # ========================================================
    # 2. INPUT VALIDATION
    # ========================================================

    if not isinstance(investment_result, dict):
        raise TypeError(
            "investment_result must be the output "
            "of the Investment Analysis engine."
        )

    required_keys = [
        "daily_returns",
        "top_25",
        "selected_tickers"
    ]

    missing_keys = [
        key for key in required_keys
        if key not in investment_result
    ]

    if missing_keys:
        raise ValueError(
            "Missing inputs: " + ", ".join(missing_keys)
        )

    if not isinstance(num_portfolios, int) or num_portfolios < 1:
        raise ValueError(
            "num_portfolios must be a positive integer."
        )

    if (
        not isinstance(max_assets_per_sector, int)
        or max_assets_per_sector < 1
    ):
        raise ValueError(
            "max_assets_per_sector must be positive."
        )

    if trading_days <= 0:
        raise ValueError("trading_days must be positive.")

    daily_returns = investment_result["daily_returns"]
    top_25 = investment_result["top_25"]
    selected_tickers = list(
        investment_result["selected_tickers"]
    )

    if not isinstance(daily_returns, pd.DataFrame):
        raise TypeError("daily_returns must be a DataFrame.")

    if not isinstance(top_25, pd.DataFrame):
        raise TypeError("top_25 must be a DataFrame.")

    if not selected_tickers:
        raise ValueError("No stocks selected.")

    if len(selected_tickers) != len(set(selected_tickers)):
        raise ValueError("Duplicate selected tickers.")

    if not set(selected_tickers).issubset(daily_returns.columns):
        raise ValueError(
            "Missing returns for one or more stocks."
        )

    if not set(selected_tickers).issubset(top_25.index):
        raise ValueError(
            "Missing metadata for one or more stocks."
        )

    # ========================================================
    # 3. SELECTED STOCK DATA AND METRICS
    # ========================================================

    portfolio_returns = daily_returns[
        selected_tickers
    ].dropna()

    if len(portfolio_returns) < 2:
        raise ValueError(
            "Insufficient observations for covariance calculation."
        )

    num_assets = len(selected_tickers)

    mean_returns = (
        portfolio_returns.mean() * trading_days
    )

    cov_matrix = (
        portfolio_returns.cov() * trading_days
    )

    asset_sectors = top_25.loc[
        selected_tickers, "Sector"
    ]

    if asset_sectors.isna().any():
        raise ValueError(
            "Missing sector for one or more stocks."
        )

    if not np.isfinite(mean_returns.to_numpy()).all():
        raise ValueError("Invalid mean returns.")

    if not np.isfinite(cov_matrix.to_numpy()).all():
        raise ValueError("Invalid covariance matrix.")

    # ========================================================
    # 4. STOCK INDICES BY SECTOR
    # ========================================================

    sector_indices = {
        sector: np.where(
            asset_sectors.values == sector
        )[0]
        for sector in asset_sectors.unique()
    }

    # ========================================================
    # 5. METRICS CALCULATION FUNCTION
    # ========================================================

    def portfolio_metrics(weights):

        expected_return = np.dot(
            weights,
            mean_returns.values
        )

        volatility = np.sqrt(
            weights.T @ cov_matrix.values @ weights
        )

        sharpe = (
            (expected_return - risk_free_rate) / volatility
            if volatility > 0 else np.nan
        )

        return expected_return, volatility, sharpe

    # ========================================================
    # 6. MONTE CARLO — 5,000 SIMULATIONS
    # ========================================================

    rng = np.random.default_rng(random_seed)

    mc_weights = np.zeros(
        (num_portfolios, num_assets)
    )

    mc_results = np.zeros(
        (num_portfolios, 3)
    )

    for i in range(num_portfolios):

        selected_indices = []

        # Random selection of up to 4 stocks per sector.
        for sector, indices in sector_indices.items():

            n_select = min(
                max_assets_per_sector,
                len(indices)
            )

            chosen = rng.choice(
                indices,
                size=n_select,
                replace=False
            )

            selected_indices.extend(chosen)

        # Random weights assigned only to selected stocks.
        weights = np.zeros(num_assets)

        random_weights = rng.dirichlet(
            np.ones(len(selected_indices))
        )

        weights[selected_indices] = random_weights

        mc_weights[i] = weights

        mc_results[i] = portfolio_metrics(weights)

    mc_df = pd.DataFrame(
        mc_results,
        columns=[
            "Annualized Return",
            "Annualized Volatility",
            "Sharpe Ratio"
        ]
    )

    if not np.isfinite(
        mc_df["Sharpe Ratio"].to_numpy()
    ).any():
        raise ValueError(
            "No portfolio with a valid Sharpe Ratio."
        )

    # ========================================================
    # 7. MAXIMUM SHARPE RATIO SELECTION
    # ========================================================

    best_idx = mc_df["Sharpe Ratio"].idxmax()

    optimal_weights = mc_weights[
        best_idx
    ].copy()

    optimal_return, optimal_volatility, optimal_sharpe = (
        portfolio_metrics(optimal_weights)
    )

    portfolio_weights = pd.Series(
        optimal_weights,
        index=selected_tickers,
        name="Portfolio Weight"
    ).sort_values(ascending=False)

    # ========================================================
    # 8. PORTFOLIO COMPOSITION
    # ========================================================

    portfolio_composition = pd.DataFrame({
        "Ticker": portfolio_weights.index,
        "Weight": portfolio_weights.values
    })

    portfolio_composition["Company"] = (
        portfolio_composition["Ticker"]
        .map(top_25["Company"])
    )

    portfolio_composition["Country"] = (
        portfolio_composition["Ticker"]
        .map(top_25["Country"])
    )

    portfolio_composition["Sector"] = (
        portfolio_composition["Ticker"]
        .map(top_25["Sector"])
    )

    portfolio_composition = portfolio_composition[
        portfolio_composition["Weight"] > 0
    ].copy()

    portfolio_composition = (
        portfolio_composition
        .sort_values("Weight", ascending=False)
        .reset_index(drop=True)
    )

    portfolio_composition["Weight (%)"] = (
        portfolio_composition["Weight"] * 100
    )

    # ========================================================
    # 9. SECTOR CONSTRAINT VALIDATION
    # ========================================================

    portfolio_sector_counts = (
        portfolio_composition["Sector"]
        .value_counts()
    )

    if (
        portfolio_sector_counts > max_assets_per_sector
    ).any():
        raise ValueError(
            "The portfolio exceeds the sector limit."
        )

    if not np.isclose(
        portfolio_weights.sum(), 1.0
    ):
        raise ValueError(
            "Portfolio weights do not sum to 1."
        )

    # ========================================================
    # 10. SECTOR DISTRIBUTION
    # ========================================================

    sector_distribution = (
        portfolio_composition
        .groupby("Sector", as_index=False)
        .agg(
            Stock_Count=("Ticker", "count"),
            Portfolio_Weight_Pct=("Weight (%)", "sum")
        )
        .sort_values(
            "Portfolio_Weight_Pct",
            ascending=False
        )
    )

    # ========================================================
    # 11. STRUCTURED OUTPUT
    # ========================================================

    return {
        "universe": "STOXX Europe 50",
        "num_portfolios": num_portfolios,
        "available_stocks": num_assets,
        "portfolio_stocks": len(portfolio_composition),
        "max_assets_per_sector": max_assets_per_sector,
        "risk_free_rate": risk_free_rate,
        "trading_days": trading_days,
        "random_seed": random_seed,
        "optimal_return": float(optimal_return),
        "optimal_volatility": float(optimal_volatility),
        "optimal_sharpe": float(optimal_sharpe),
        "optimal_weights": optimal_weights,
        "portfolio_weights": portfolio_weights,
        "portfolio_composition": portfolio_composition,
        "portfolio_sector_counts": portfolio_sector_counts,
        "sector_distribution": sector_distribution,
        "portfolio_returns": portfolio_returns,
        "mean_returns": mean_returns,
        "cov_matrix": cov_matrix,
        "mc_df": mc_df,
        "mc_weights": mc_weights
    }

# ============================================================
# 12. TOOL — DIRECT CALCULATION FROM PRICES AND METADATA
# ============================================================

def optimize_portfolio_from_prices(
    prices,
    metadata,
    num_portfolios=5000,
    max_assets_per_sector=4,
    risk_free_rate=0.0,
    trading_days=252,
    random_seed=42
):
    # Calls the BOX 23 engine.
    # Calculates the ranking and Top 25 from input prices.
    investment_result = calculate_equity_ranking(
        prices=prices,
        metadata=metadata,
        top_n=25,
        trading_days=trading_days
    )

    # Performs the Monte Carlo optimization.
    return calculate_optimal_portfolio(
        investment_result=investment_result,
        num_portfolios=num_portfolios,
        max_assets_per_sector=max_assets_per_sector,
        risk_free_rate=risk_free_rate,
        trading_days=trading_days,
        random_seed=random_seed
    )

# ============================================================
# 13. TOOL REGISTRY
# ============================================================

portfolio_optimization_tools = {
    "calculate_optimal_portfolio": calculate_optimal_portfolio,
    "optimize_portfolio_from_prices": (
        optimize_portfolio_from_prices
    )
}

# ============================================================
# 14. AUTONOMOUS CALCULATION TEST
# ============================================================

# The test starts directly from prices and metadata.
# It does not use precomputed results from BOXES 18, 19, and 20.

portfolio_optimization_result = (
    optimize_portfolio_from_prices(
        prices=df2,
        metadata=stock_metadata,
        num_portfolios=5000,
        max_assets_per_sector=4,
        risk_free_rate=0.0,
        trading_days=252,
        random_seed=42
    )
)

test_composition = portfolio_optimization_result[
    "portfolio_composition"
]

test_sectors = portfolio_optimization_result[
    "portfolio_sector_counts"
]

assert np.isclose(
    test_composition["Weight"].sum(),
    1.0
)

assert (test_sectors <= 4).all()

assert len(
    portfolio_optimization_result["mc_df"]
) == 5000

assert np.isclose(
    portfolio_optimization_result["optimal_sharpe"],
    portfolio_optimization_result["mc_df"][
        "Sharpe Ratio"
    ].max()
)

print("\n" + "=" * 90)
print("PORTFOLIO OPTIMIZATION TOOL — CALCULATION TEST")
print("=" * 90)

print("\nMONTE CARLO OPTIMIZATION")

print(
    f"Simulated portfolios: "
    f"{portfolio_optimization_result['num_portfolios']}"
)

print(
    f"Available stocks: "
    f"{portfolio_optimization_result['available_stocks']}"
)

print(
    f"Stocks in optimal portfolio: "
    f"{portfolio_optimization_result['portfolio_stocks']}"
)

print(
    f"Maximum stocks per sector: "
    f"{portfolio_optimization_result['max_assets_per_sector']}"
)

print("\nOPTIMAL PORTFOLIO — MAXIMUM SHARPE RATIO")

print(
    f"Annualized Expected Return: "
    f"{portfolio_optimization_result['optimal_return']:.2%}"
)

print(
    f"Annualized Volatility: "
    f"{portfolio_optimization_result['optimal_volatility']:.2%}"
)

print(
    f"Sharpe Ratio: "
    f"{portfolio_optimization_result['optimal_sharpe']:.4f}"
)

print("\nPORTFOLIO COMPOSITION")

print(
    test_composition[
        [
            "Ticker",
            "Company",
            "Country",
            "Sector",
            "Weight (%)"
        ]
    ].to_string(
        index=False,
        formatters={
            "Weight (%)": lambda x: f"{x:.2f}%"
        }
    )
)

print("\nSECTOR DISTRIBUTION")

print(
    portfolio_optimization_result[
        "sector_distribution"
    ].to_string(
        index=False,
        formatters={
            "Portfolio_Weight_Pct": lambda x: f"{x:.2f}%"
        }
    )
)

print("\nPORTFOLIO OPTIMIZATION CALCULATION TOOL READY.")
```

```python
# ============================================================
# BOX 25
#
# Portfolio Scenario Tool — STOXX Europe 50
#
# Autonomous financial scenario simulation engine:
# - Multivariate bootstrap of portfolio returns
# - 1,000 simulated paths
# - Unfavorable P5, central P50, favorable P95 scenarios
# - Flexible horizon from 1 day to 24 months
# - Percentage and monetary gains and losses
# - Initial capital: 50% of the net surplus over the last 12 months
# STARTING RULE: accumulated net savings are frozen after 12 months.
# 50% funds the investment portfolio, while the remaining 50% stays as
# personal liquidity; from the following month, the two amounts evolve separately.
#
# Input: optimized portfolio and Personal Finance transactions
# Output: structured scenarios for the agentic system
#
# No reuse of precomputed simulations
# ============================================================


# ============================================================
# 1. PARAMETERS AND SUPPORT FUNCTIONS
# ============================================================

DEFAULT_SIMULATIONS = 1000
MAX_HORIZON_MONTHS = 24
TRADING_DAYS_PER_MONTH = 21
DEFAULT_SEED = 42
DEFAULT_ALLOCATION = 0.50

DEFAULT_HORIZONS = (1, 3, 6, 9, 12, 18, 24)


def _validate_horizon(months=None, days=None):

    if (months is None) == (days is None):
        raise ValueError(
            "Specify either months or days, not both."
        )

    if months is not None:

        if (
            isinstance(months, bool)
            or not isinstance(months, (int, float))
            or not np.isfinite(months)
        ):
            raise ValueError("months must be numeric.")

        horizon_days = round(
            months * TRADING_DAYS_PER_MONTH
        )

    else:

        if (
            isinstance(days, bool)
            or not isinstance(days, (int, np.integer))
        ):
            raise ValueError(
                "days must be an integer number of trading days."
            )

        horizon_days = int(days)

    max_days = (
        MAX_HORIZON_MONTHS * TRADING_DAYS_PER_MONTH
    )

    if not 1 <= horizon_days <= max_days:
        raise ValueError(
            f"Allowed horizon: 1-{max_days} trading days."
        )

    return horizon_days


# ============================================================
# 2. TOOL — AUTONOMOUS INVESTABLE CAPITAL CALCULATION
# ============================================================

def calculate_investable_capital(
    transactions,
    allocation=DEFAULT_ALLOCATION
):

    if not isinstance(transactions, pd.DataFrame):
        raise TypeError(
            "transactions must be a DataFrame."
        )

    required_columns = [
        "Month",
        "Type",
        "Amount_EUR"
    ]

    if not all(
        col in transactions.columns
        for col in required_columns
    ):
        raise ValueError(
            "Incomplete transactions: required columns are "
            "Month, Type e Amount_EUR."
        )

    if not 0 <= allocation <= 1:
        raise ValueError(
            "allocation must be between 0 and 1."
        )

    data = transactions[
        required_columns
    ].copy()

    data["Month"] = pd.to_datetime(
        data["Month"]
    ).dt.to_period("M")

    data["Amount_EUR"] = pd.to_numeric(
        data["Amount_EUR"],
        errors="raise"
    )

    if not np.isfinite(
        data["Amount_EUR"].to_numpy()
    ).all():
        raise ValueError(
            "Invalid financial amounts."
        )

    if (data["Amount_EUR"] < 0).any():
        raise ValueError(
            "Transaction amounts must be "
            "positive; Income and Expense are distinguished by Type."
        )

    if not data["Type"].isin(
        ["Income", "Expense"]
    ).all():
        raise ValueError(
            "Type must contain Income or Expense."
        )

    monthly_data = data.pivot_table(
        index="Month",
        columns="Type",
        values="Amount_EUR",
        aggfunc="sum",
        fill_value=0
    ).sort_index()

    for column in ["Income", "Expense"]:
        if column not in monthly_data.columns:
            monthly_data[column] = 0.0

    last_month = monthly_data.index.max()

    expected_months = pd.period_range(
        end=last_month,
        periods=12,
        freq="M"
    )

    if not expected_months.isin(
        monthly_data.index
    ).all():
        raise ValueError(
            "12 consecutive months of "
            "Personal Finance data are not available."
        )

    last_12_months = monthly_data.loc[
        expected_months
    ]

    income_12m = float(
        last_12_months["Income"].sum()
    )

    expenses_12m = float(
        last_12_months["Expense"].sum()
    )

    net_surplus_12m = (
        income_12m - expenses_12m
    )

    positive_surplus = max(
        net_surplus_12m,
        0.0
    )

    initial_investment = (
        positive_surplus * allocation
    )

    remaining_liquidity = (
        positive_surplus - initial_investment
    )

    return {
        "period_start": str(expected_months[0]),
        "period_end": str(expected_months[-1]),
        "income_12m_eur": income_12m,
        "expenses_12m_eur": expenses_12m,
        "net_surplus_12m_eur": net_surplus_12m,
        "positive_surplus_eur": positive_surplus,
        "allocation_pct": allocation * 100,
        "initial_investment_eur": initial_investment,
        "remaining_liquidity_eur": remaining_liquidity
    }


# ============================================================
# 3. AUTONOMOUS BOOTSTRAP ENGINE
# ============================================================

def calculate_portfolio_scenarios(
    portfolio_result,
    transactions,
    months=24,
    num_simulations=DEFAULT_SIMULATIONS,
    random_seed=DEFAULT_SEED,
    allocation=DEFAULT_ALLOCATION,
    days=None
):

    # ========================================================
    # 4. INPUT VALIDATION
    # ========================================================

    if not isinstance(portfolio_result, dict):
        raise TypeError(
            "portfolio_result must be the output "
            "of the Portfolio Optimization engine."
        )

    required_keys = [
        "portfolio_returns",
        "portfolio_weights"
    ]

    missing_keys = [
        key for key in required_keys
        if key not in portfolio_result
    ]

    if missing_keys:
        raise ValueError(
            "Missing inputs: " + ", ".join(missing_keys)
        )

    if (
        not isinstance(num_simulations, int)
        or num_simulations < 1
    ):
        raise ValueError(
            "num_simulations must be a positive integer."
        )

    if days is not None:
        horizon_days = _validate_horizon(days=days)
    else:
        horizon_days = _validate_horizon(months=months)

    # ========================================================
    # 5. INITIAL CAPITAL
    # ========================================================

    capital = calculate_investable_capital(
        transactions=transactions,
        allocation=allocation
    )

    initial_investment = capital[
        "initial_investment_eur"
    ]

    # ========================================================
    # 6. PORTFOLIO RETURNS AND WEIGHTS
    # ========================================================

    portfolio_returns = portfolio_result[
        "portfolio_returns"
    ]

    portfolio_weights = portfolio_result[
        "portfolio_weights"
    ]

    if not isinstance(portfolio_returns, pd.DataFrame):
        raise TypeError(
            "portfolio_returns must be a DataFrame."
        )

    if not isinstance(portfolio_weights, pd.Series):
        raise TypeError(
            "portfolio_weights must be a Series."
        )

    active_weights = portfolio_weights[
        portfolio_weights > 0
    ]

    tickers = active_weights.index.tolist()

    if not tickers:
        raise ValueError(
            "The portfolio contains no active stocks."
        )

    if not set(tickers).issubset(
        portfolio_returns.columns
    ):
        raise ValueError(
            "Missing returns for some stocks."
        )

    weights = active_weights.to_numpy(
        dtype=float
    )

    if not np.isclose(weights.sum(), 1.0):
        raise ValueError(
            "The sum of weights must equal 1."
        )

    historical_returns = (
        portfolio_returns[tickers]
        .dropna()
        .to_numpy(dtype=float)
    )

    if len(historical_returns) < 2:
        raise ValueError(
            "Insufficient historical observations."
        )

    if not np.isfinite(
        historical_returns
    ).all():
        raise ValueError(
            "Invalid historical returns."
        )

    # ========================================================
    # 7. MULTIVARIATE BOOTSTRAP
    # ========================================================

    rng = np.random.default_rng(
        random_seed
    )

    sample_indices = rng.integers(
        low=0,
        high=len(historical_returns),
        size=(num_simulations, horizon_days)
    )

    # Trading days are resampled jointly.
    # Contemporaneous correlations are preserved
    # in the empirical sample distribution.

    sampled_returns = historical_returns[
        sample_indices
    ]

    simulated_portfolio_returns = (
        sampled_returns @ weights
    )

    # ========================================================
    # 8. INVESTMENT PATHS
    # ========================================================

    bootstrap_growth = np.concatenate(
        [
            np.ones((num_simulations, 1)),
            np.cumprod(
                1 + simulated_portfolio_returns,
                axis=1
            )
        ],
        axis=1
    )

    # Percentage gains and losses.
    bootstrap_pnl_pct = (
        bootstrap_growth - 1
    ) * 100

    # Gains and losses in EUR.
    bootstrap_pnl_eur = (
        bootstrap_growth - 1
    ) * initial_investment

    # Value of invested capital.
    bootstrap_values_eur = (
        bootstrap_growth * initial_investment
    )

    # ========================================================
    # 9. P5, P50, P95 SCENARIOS
    # ========================================================

    percentiles = np.percentile(
        bootstrap_growth,
        [5, 50, 95],
        axis=0
    )

    scenario_names = [
        "Unfavorable P5",
        "Central P50",
        "Favorable P95"
    ]

    scenario_paths = pd.DataFrame({
        "Day": np.arange(horizon_days + 1),
        "Month": (
            np.arange(horizon_days + 1)
            / TRADING_DAYS_PER_MONTH
        ),
        "Unfavorable P5 (%)": (
            percentiles[0] - 1
        ) * 100,
        "Central P50 (%)": (
            percentiles[1] - 1
        ) * 100,
        "Favorable P95 (%)": (
            percentiles[2] - 1
        ) * 100,
        "Unfavorable P5 (EUR)": (
            percentiles[0] - 1
        ) * initial_investment,
        "Central P50 (EUR)": (
            percentiles[1] - 1
        ) * initial_investment,
        "Favorable P95 (EUR)": (
            percentiles[2] - 1
        ) * initial_investment,
        "Unfavorable P5 Value (EUR)": (
            percentiles[0] * initial_investment
        ),
        "Central P50 Value (EUR)": (
            percentiles[1] * initial_investment
        ),
        "Favorable P95 Value (EUR)": (
            percentiles[2] * initial_investment
        )
    })

    # ========================================================
    # 10. RESULTS AT THE REQUESTED HORIZON
    # ========================================================

    final_scenarios = []

    for i, name in enumerate(scenario_names):

        growth = float(
            percentiles[i, -1]
        )

        pnl_pct = (growth - 1) * 100

        pnl_eur = (
            initial_investment * (growth - 1)
        )

        final_value = (
            initial_investment * growth
        )

        final_scenarios.append({
            "scenario": name,
            "gain_loss_pct": pnl_pct,
            "gain_loss_eur": pnl_eur,
            "final_value_eur": final_value
        })

    # ========================================================
    # 11. STRUCTURED OUTPUT
    # ========================================================

    return {
        "universe": "STOXX Europe 50",
        "horizon_months": (
            horizon_days / TRADING_DAYS_PER_MONTH
        ),
        "horizon_trading_days": horizon_days,
        "num_simulations": num_simulations,
        "random_seed": random_seed,
        "portfolio_stocks": len(tickers),
        "portfolio_weights_used": active_weights.copy(),
        "capital": capital,
        "scenarios": final_scenarios,
        "scenario_paths": scenario_paths,
        "bootstrap_growth": bootstrap_growth,
        "bootstrap_pnl_pct": bootstrap_pnl_pct,
        "bootstrap_pnl_eur": bootstrap_pnl_eur,
        "bootstrap_values_eur": bootstrap_values_eur
    }


# ============================================================
# 12. TOOL — SCENARIOS AT ANY HORIZON
# ============================================================

def get_portfolio_scenarios(
    prices,
    metadata,
    transactions,
    months=12,
    days=None,
    num_simulations=DEFAULT_SIMULATIONS,
    random_seed=DEFAULT_SEED
):

    # Independently calculates the ranking and Top 25.
    investment_result = calculate_equity_ranking(
        prices=prices,
        metadata=metadata
    )

    # Independently calculates the optimal portfolio.
    portfolio_result = calculate_optimal_portfolio(
        investment_result=investment_result,
        num_portfolios=5000,
        max_assets_per_sector=4,
        random_seed=42
    )

    # Independently calculates the bootstrap scenarios.
    return calculate_portfolio_scenarios(
        portfolio_result=portfolio_result,
        transactions=transactions,
        months=months,
        days=days,
        num_simulations=num_simulations,
        random_seed=random_seed,
        allocation=DEFAULT_ALLOCATION
    )


# ============================================================
# 13. TOOL — MULTIPLE ROLLING SCENARIOS
# ============================================================

def get_rolling_portfolio_scenarios(
    portfolio_result,
    transactions,
    horizons=DEFAULT_HORIZONS,
    num_simulations=DEFAULT_SIMULATIONS,
    random_seed=DEFAULT_SEED
):

    if not horizons:
        raise ValueError(
            "Specify at least one time horizon."
        )

    horizons = list(horizons)

    for month in horizons:
        _validate_horizon(months=month)

    # A single simulation up to the maximum horizon.
    # Intermediate results are extracted from the same
    # paths, without repeating the bootstrap.

    max_months = max(horizons)

    result = calculate_portfolio_scenarios(
        portfolio_result=portfolio_result,
        transactions=transactions,
        months=max_months,
        num_simulations=num_simulations,
        random_seed=random_seed
    )

    paths = result["bootstrap_growth"]

    initial_investment = result[
        "capital"
    ]["initial_investment_eur"]

    rows = []

    for month in horizons:

        day = _validate_horizon(
            months=month
        )

        p5, p50, p95 = np.percentile(
            paths[:, day],
            [5, 50, 95]
        )

        rows.append({
            "Months": month,
            "Unfavorable P5 (%)": (p5 - 1) * 100,
            "Central P50 (%)": (p50 - 1) * 100,
            "Favorable P95 (%)": (p95 - 1) * 100,
            "Unfavorable P5 (EUR)": (
                (p5 - 1) * initial_investment
            ),
            "Central P50 (EUR)": (
                (p50 - 1) * initial_investment
            ),
            "Favorable P95 (EUR)": (
                (p95 - 1) * initial_investment
            ),
            "Unfavorable P5 Value (EUR)": (
                p5 * initial_investment
            ),
            "Central P50 Value (EUR)": (
                p50 * initial_investment
            ),
            "Favorable P95 Value (EUR)": (
                p95 * initial_investment
            )
        })

    rolling_summary = pd.DataFrame(rows)

    return {
        "capital": result["capital"],
        "portfolio_weights_used": result["portfolio_weights_used"].copy(),
        "horizons": horizons,
        "rolling_summary": rolling_summary,
        "scenario_paths": result["scenario_paths"],
        "num_simulations": num_simulations
    }


# ============================================================
# 14. TOOL REGISTRY
# ============================================================

portfolio_scenario_tools = {
    "calculate_investable_capital": (
        calculate_investable_capital
    ),
    "calculate_portfolio_scenarios": (
        calculate_portfolio_scenarios
    ),
    "get_portfolio_scenarios": (
        get_portfolio_scenarios
    ),
    "get_rolling_portfolio_scenarios": (
        get_rolling_portfolio_scenarios
    )
}

# ============================================================
# 15. TEST WITH CURRENT NOTEBOOK DATA
# ============================================================

# df: Personal Finance transactions from BOX 1.
# portfolio_optimization_result: output of the BOX 24 engine.
#
# The data are test inputs, not precomputed scenarios.
# The tool actually performs the bootstrap simulation.

scenario_test = get_rolling_portfolio_scenarios(
    portfolio_result=portfolio_optimization_result,
    transactions=df,
    horizons=(1, 3, 5, 6, 9, 12, 18, 24),
    num_simulations=1000,
    random_seed=42
)

capital_test = scenario_test["capital"]
rolling_test = scenario_test["rolling_summary"]

assert len(rolling_test) == 8

assert np.isclose(
    capital_test["initial_investment_eur"],
    max(capital_test["net_surplus_12m_eur"], 0) * 0.50
)

assert np.isclose(
    capital_test["initial_investment_eur"]
    + capital_test["remaining_liquidity_eur"],
    capital_test["positive_surplus_eur"]
)

print("\n" + "=" * 110)
print("PORTFOLIO SCENARIO TOOL — CALCULATION TEST")
print("=" * 110)

print("\nBOOTSTRAP CONFIGURATION")

print(
    f"Simulations: "
    f"{scenario_test['num_simulations']}"
)

print(
    f"Maximum horizon: "
    f"{max(scenario_test['horizons'])} months"
)

print("\nCUMULATIVE GAINS / LOSSES BY HORIZON")

print(
    f"\nIncome — last 12 months: "
    f"EUR {capital_test['income_12m_eur']:,.2f}"
)

print(
    f"Expenses — last 12 months: "
    f"EUR {capital_test['expenses_12m_eur']:,.2f}"
)

print(
    f"Net surplus — last 12 months: "
    f"EUR {capital_test['net_surplus_12m_eur']:,.2f}"
)

print(
    f"Initial investment — 50%: "
    f"EUR {capital_test['initial_investment_eur']:,.2f}"
)

print(
    f"Remaining surplus liquidity: "
    f"EUR {capital_test['remaining_liquidity_eur']:,.2f}"
)

print("STARTING BASIS: net savings frozen after 12 months; 50% invested and 50% retained as personal liquidity. The two portions evolve separately.")
print("\nROLLING SCENARIOS — GAINS / LOSSES")

print(
    rolling_test[
        [
            "Months",
            "Unfavorable P5 (%)",
            "Central P50 (%)",
            "Favorable P95 (%)",
            "Unfavorable P5 (EUR)",
            "Central P50 (EUR)",
            "Favorable P95 (EUR)"
        ]
    ].to_string(
        index=False,
        formatters={
            "Months": lambda x: f"{x:g}",
            "Unfavorable P5 (%)": lambda x: f"{x:+.2f}%",
            "Central P50 (%)": lambda x: f"{x:+.2f}%",
            "Favorable P95 (%)": lambda x: f"{x:+.2f}%",
            "Unfavorable P5 (EUR)": lambda x: f"{x:+,.2f}",
            "Central P50 (EUR)": lambda x: f"{x:+,.2f}",
            "Favorable P95 (EUR)": lambda x: f"{x:+,.2f}"
        }
    )
)

print("\nPORTFOLIO SCENARIO CALCULATION TOOL READY.")
```

## **Integrated Multi-Agent Architecture — Financial & Investment Analysis**

This section introduces the complete agentic architecture, building on the Personal Finance workflow already tested in its core components, and extending it to incorporate investment analysis, portfolio risk assessment, and integrated financial scenarios.

The LangGraph-based multi-agent system coordinates specialized agents and analytical tools to evaluate personal finances, investment outcomes, and their combined impact on financial resilience and sustainability.

```python
# ============================================================
# BOX 26 — AG-1: UNIFIED AGENT STATE
# Financial Decision Intelligence
#
# Unified state for the final LangGraph multi-agent architecture.
# Integrates personal finance, financial stress testing,
# portfolio optimization, and investment risk analysis.
# Provides a shared foundation for agent orchestration
# and integrated financial assessments.
# ============================================================

# Methodology — Initial Savings Allocation
# Accumulated net savings after 12 months: EUR 6,264.52
# Investment portfolio allocation (50%): EUR 3,132.26
# Available personal liquidity (50%):   EUR 3,132.26
# Both components evolve separately over the following 24 months.

FinancialQueryType = Literal[
    "personal_finance", "investment", "integrated", "unknown"
]

# Semantic contract for the 41-question English benchmark:
# 15 Financial Analysis, 15 Financial Decision-Making, 11 integrated questions.
# These are state-level slots, not a keyword router or fabricated query results.
# The orchestrator will populate them from the user's actual wording.
SEMANTIC_BENCHMARK_COVERAGE = {
    "financial_analysis": 15,
    "financial_decision_making": 15,
    "integrated_investment": 11,
}
SEMANTIC_QUERY_TYPES = (
    "personal_finance", "investment", "integrated", "unknown"
)
SEMANTIC_SCENARIOS = (
    "Unfavorable P5", "Central P50", "Favorable P95"
)


class UnifiedAgentState(TypedDict, total=False):
    # Query and orchestration
    query: str
    query_type: FinancialQueryType
    analysis_type: str
    rolling_months: int
    requested_parameters: dict
    # Semantic interpretation is performed downstream, not at initialization.
    semantic_intent: Optional[str]
    semantic_subintents: list
    semantic_entities: dict
    semantic_confidence: Optional[float]
    semantic_ambiguities: list
    # Explicit request parameters: never infer absent values as user requests.
    requested_horizon_months: Optional[float]
    requested_horizon_days: Optional[int]
    requested_personal_scenario: Optional[str]
    requested_market_scenario: Optional[str]
    requested_market_percentile: Optional[int]
    requested_income_reduction_pct: Optional[float]
    requested_extra_expense_eur: Optional[float]
    requested_expense_change_pct: Optional[float]
    requested_liquidity_target_eur: Optional[float]
    requested_output_metrics: list
    requires_integrated_assessment: bool
    semantic_validation_status: str
    income_reduction_months: Optional[int]  # requested duration: 1 to 24; no default
    execution_plan: list
    active_agent: Optional[str]

    # Personal Finance
    budget_data: Any
    personal_finance_results: dict
    income_12m_eur: Optional[float]
    expenses_12m_eur: Optional[float]
    net_surplus_12m_eur: Optional[float]

    # Liquidity and investment allocation
    initial_liquidity_eur: Optional[float]
    investment_allocation_pct: float
    initial_investment_eur: Optional[float]
    remaining_liquidity_eur: Optional[float]

    # Investment
    investment_analysis_result: dict
    portfolio_optimization_result: dict
    portfolio_composition: Any
    portfolio_weights: Any
    portfolio_sharpe: Optional[float]

    # Bootstrap
    portfolio_scenario_result: dict
    portfolio_scenarios: Any
    portfolio_scenario_paths: Any

    # Personal Finance scenarios
    personal_finance_baseline: dict
    personal_finance_stress_scenarios: dict

    # Integrated assessment
    integrated_scenarios: dict
    integrated_assessment: dict

    # Results and conversation
    result: Any
    agent_results: dict
    conversation_history: list
    errors: list
    status: str

    # Data provenance and validation
    data_sources: dict
    result_parameters: dict
    missing_data: list
    validation_warnings: list


# ============================================================
# 1. STATE INITIALIZATION
# ============================================================

def create_unified_agent_state(
    query="",
    initial_liquidity_eur=None,
    allocation=0.50
) -> UnifiedAgentState:

    if not isinstance(query, str):
        raise TypeError("query must be a string.")

    if not 0 <= allocation <= 1:
        raise ValueError("Allocation must be between 0 and 1.")

    return UnifiedAgentState(
        query=query,
        query_type="unknown",
        analysis_type="",
        rolling_months=24,
        requested_parameters={},
        semantic_intent=None,
        semantic_subintents=[],
        semantic_entities={},
        semantic_confidence=None,
        semantic_ambiguities=[],
        requested_horizon_months=None,
        requested_horizon_days=None,
        requested_personal_scenario=None,
        requested_market_scenario=None,
        requested_market_percentile=None,
        requested_income_reduction_pct=None,
        requested_extra_expense_eur=None,
        requested_expense_change_pct=None,
        requested_liquidity_target_eur=None,
        requested_output_metrics=[],
        requires_integrated_assessment=False,
        semantic_validation_status="not_interpreted",
        income_reduction_months=None,
        execution_plan=[],
        active_agent=None,

        budget_data=None,
        personal_finance_results={},
        income_12m_eur=None,
        expenses_12m_eur=None,
        net_surplus_12m_eur=None,

        initial_liquidity_eur=initial_liquidity_eur,
        investment_allocation_pct=allocation * 100,
        initial_investment_eur=None,
        remaining_liquidity_eur=None,

        investment_analysis_result={},
        portfolio_optimization_result={},
        portfolio_composition=None,
        portfolio_weights=None,
        portfolio_sharpe=None,

        portfolio_scenario_result={},
        portfolio_scenarios=None,
        portfolio_scenario_paths=None,

        personal_finance_baseline={},
        personal_finance_stress_scenarios={},

        integrated_scenarios={},
        integrated_assessment={},

        result=None,
        agent_results={},
        conversation_history=[],
        errors=[],
        status="initialized",

        data_sources={},
        result_parameters={},
        missing_data=[],
        validation_warnings=[]
    )


# ============================================================
# 2. SAFE STATE UPDATE
# ============================================================

def update_unified_agent_state(
    state: UnifiedAgentState,
    updates: dict
) -> UnifiedAgentState:

    unknown = set(updates) - set(
        UnifiedAgentState.__annotations__
    )

    if unknown:
        raise KeyError(f"Unknown state fields: {unknown}")

    updated = dict(state)
    updated.update(updates)
    return updated


# ============================================================
# 2A. SEMANTIC REQUEST CONTRACT — VALIDATION ONLY
# ============================================================

def validate_semantic_request(state: UnifiedAgentState) -> list:
    """Validate extracted intent/parameters without guessing or routing."""
    issues = []
    query_type = state.get("query_type", "unknown")
    if query_type not in SEMANTIC_QUERY_TYPES:
        issues.append("Unsupported query_type.")

    if state.get("requires_integrated_assessment") and query_type != "integrated":
        issues.append("Integrated assessment requires integrated query_type.")

    duration = state.get("income_reduction_months")
    if duration is not None and (
        isinstance(duration, bool)
        or not isinstance(duration, (int, np.integer))
        or not 1 <= duration <= 24
    ):
        issues.append("Income reduction duration must be 1–24 months.")

    horizon_months = state.get("requested_horizon_months")
    if horizon_months is not None and (
        isinstance(horizon_months, bool)
        or not isinstance(horizon_months, (int, float, np.integer, np.floating))
        or not np.isfinite(horizon_months)
        or not 0 < horizon_months <= 24
    ):
        issues.append("Requested horizon must be within 24 months.")

    horizon_days = state.get("requested_horizon_days")
    if horizon_days is not None and (
        isinstance(horizon_days, bool)
        or not isinstance(horizon_days, (int, np.integer))
        or not 1 <= horizon_days <= 504
    ):
        issues.append("Requested horizon must be 1–504 trading days.")

    percentile = state.get("requested_market_percentile")
    if percentile is not None and percentile not in (5, 50, 95):
        issues.append("Market percentile must be P5, P50 or P95.")

    market_scenario = state.get("requested_market_scenario")
    if market_scenario is not None and market_scenario not in SEMANTIC_SCENARIOS:
        issues.append("Unknown market scenario.")
    if percentile is not None and market_scenario is not None:
        expected = {5: "Unfavorable P5", 50: "Central P50", 95: "Favorable P95"}
        if market_scenario != expected.get(percentile):
            issues.append("Market scenario and percentile disagree.")

    for key in ("requested_income_reduction_pct", "requested_expense_change_pct"):
        value = state.get(key)
        if value is not None and (
            isinstance(value, bool)
            or not isinstance(value, (int, float, np.integer, np.floating))
            or not np.isfinite(value)
        ):
            issues.append(f"Invalid numeric parameter: {key}.")

    for key in ("requested_extra_expense_eur", "requested_liquidity_target_eur"):
        value = state.get(key)
        if value is not None and (
            isinstance(value, bool)
            or not isinstance(value, (int, float, np.integer, np.floating))
            or not np.isfinite(value)
            or value < 0
        ):
            issues.append(f"Invalid nonnegative amount: {key}.")

    return issues


# ============================================================
# 3. FINANCIAL POSITION
# ============================================================

def get_financial_position(
    state: UnifiedAgentState,
    portfolio_value_eur=None
):

    liquidity = state.get("remaining_liquidity_eur")
    invested = state.get("initial_investment_eur")

    # Do not silently substitute zero for missing data.
    if portfolio_value_eur is None:
        portfolio_value_eur = invested

    total = None
    if liquidity is not None and portfolio_value_eur is not None:
        total = float(liquidity) + float(portfolio_value_eur)

    return {
        "available_liquidity_eur": liquidity,
        "invested_capital_eur": invested,
        "portfolio_value_eur": portfolio_value_eur,
        "total_financial_wealth_eur": total
    }


# ============================================================
# 4. READ EXISTING PERSONAL FINANCE RESULTS
# ============================================================

def _box26_read_personal_finance(state, namespace):

    pf4 = namespace.get("pf4_results")

    if not isinstance(pf4, dict):
        state["missing_data"].append("pf4_results")
        return

    state["personal_finance_results"] = pf4
    state["data_sources"]["personal_finance"] = "pf4_results"

    # Preserve original PF-4 assumptions and outputs.
    inputs = pf4.get("inputs", {})
    if isinstance(inputs, dict):
        liquidity = inputs.get("Initial liquidity EUR")

        if liquidity is not None:
            state["initial_liquidity_eur"] = float(liquidity)

    baseline_tool = namespace.get("pf5_get_financial_baseline")

    if callable(baseline_tool):
        try:
            state["personal_finance_baseline"] = baseline_tool()
            state["data_sources"]["pf_baseline"] = (
                "pf5_get_financial_baseline"
            )
        except Exception as exc:
            state["validation_warnings"].append(
                f"PF baseline retrieval: {exc}"
            )

    # Retrieve the four original scenarios without
    # inserting predefined financial results.
    scenario_tool = namespace.get("pf5_get_stress_scenario")

    if callable(scenario_tool):
        scenarios = {}

        for scenario in (1, 2, 3, 4):
            try:
                scenarios[f"scenario_{scenario}"] = (
                    scenario_tool(scenario)
                )
            except Exception as exc:
                state["validation_warnings"].append(
                    f"PF scenario {scenario}: {exc}"
                )

        state["personal_finance_stress_scenarios"] = scenarios

        if len(scenarios) == 4:
            state["data_sources"]["pf_stress"] = (
                "pf5_get_stress_scenario"
            )


# ============================================================
# 5. READ EXISTING INVESTMENT RESULTS
# ============================================================

def _box26_read_investment(state, namespace):

    ranking = namespace.get("investment_analysis_result")

    if isinstance(ranking, dict):
        state["investment_analysis_result"] = ranking
        state["data_sources"]["ranking"] = (
            "investment_analysis_result"
        )
    else:
        state["missing_data"].append(
            "investment_analysis_result"
        )

    portfolio = namespace.get("portfolio_optimization_result")

    if isinstance(portfolio, dict):
        state["portfolio_optimization_result"] = portfolio

        state["portfolio_composition"] = portfolio.get(
            "portfolio_composition"
        )

        state["portfolio_weights"] = portfolio.get(
            "portfolio_weights"
        )

        state["portfolio_sharpe"] = portfolio.get(
            "optimal_sharpe"
        )

        state["data_sources"]["optimization"] = (
            "portfolio_optimization_result"
        )
    else:
        state["missing_data"].append(
            "portfolio_optimization_result"
        )


# ============================================================
# 6. CALCULATE INVESTABLE CAPITAL USING BOX 25
# ============================================================

def _box26_read_capital(state, namespace):

    transactions = namespace.get("df")
    capital_tool = namespace.get("calculate_investable_capital")

    if transactions is None or not callable(capital_tool):
        state["missing_data"].append(
            "calculate_investable_capital / df"
        )
        return

    allocation = state["investment_allocation_pct"] / 100.0

    try:
        capital = capital_tool(
            transactions=transactions,
            allocation=allocation
        )

        state["budget_data"] = transactions

        for field in (
            "income_12m_eur",
            "expenses_12m_eur",
            "net_surplus_12m_eur",
            "initial_investment_eur"
        ):
            state[field] = float(capital[field])


      # Note: investment portfolio allocation and available liquidity
      # evolve separately over the following 24 months.
        state["result_parameters"]["surplus_allocation"] = capital

        initial_cash = state.get("initial_liquidity_eur")
        invested = state["initial_investment_eur"]

        if initial_cash is not None:
            state["remaining_liquidity_eur"] = (
                float(initial_cash) - invested
            )

            if state["remaining_liquidity_eur"] < 0:
                state["validation_warnings"].append(
                    "Investment exceeds estimated initial liquidity."
                )

        state["data_sources"]["capital"] = (
            "calculate_investable_capital"
        )

    except Exception as exc:
        state["validation_warnings"].append(
            f"Capital calculation: {exc}"
        )


# ============================================================
# 7. READ EXISTING BOOTSTRAP ONLY WHEN COMPATIBLE
# ============================================================

def _box26_read_bootstrap(state, namespace):

    # Box 25 already produces scenario_test in the notebook.
    # Reuse only if horizon, allocation and invested capital
    # match the current state.

    result = namespace.get("scenario_test")

    if not isinstance(result, dict):
        state["missing_data"].append(
            "compatible_portfolio_scenarios"
        )
        return

    capital = result.get("capital", {})

    if not isinstance(capital, dict):
        return

    months = result.get("horizons", [])
    invested = state.get("initial_investment_eur")
    stored_invested = capital.get("initial_investment_eur")

    paths = result.get("scenario_paths")
    required_pnl = [
        "Unfavorable P5 (EUR)",
        "Central P50 (EUR)",
        "Favorable P95 (EUR)"
    ]
    paths_valid = (
        isinstance(paths, pd.DataFrame)
        and "Month" in paths.columns
        and all(column in paths.columns for column in required_pnl)
        and not paths.empty
        and paths[required_pnl].notna().all().all()
        and paths["Month"].min() <= 0
        and paths["Month"].max() >= 24
    )

    # Compare the exact weight snapshot stored by Box 25 with
    # the current optimized portfolio, ticker by ticker.
    portfolio_result = state.get("portfolio_optimization_result", {})
    weights = portfolio_result.get("portfolio_weights")
    returns = portfolio_result.get("portfolio_returns")
    stored_weights = result.get("portfolio_weights_used")
    weights_match = (
        isinstance(stored_weights, pd.Series)
        and isinstance(weights, pd.Series)
        and stored_weights.index.is_unique
        and weights.index.is_unique
        and set(stored_weights.index) == set(weights[weights > 0].index)
        and np.allclose(
            stored_weights.sort_index().to_numpy(dtype=float),
            weights[weights > 0].sort_index().to_numpy(dtype=float),
            rtol=1e-10, atol=1e-12
        )
    )
    portfolio_valid = (
        isinstance(weights, pd.Series)
        and isinstance(returns, pd.DataFrame)
        and not weights.empty
        and set(weights[weights > 0].index).issubset(returns.columns)
        and np.isclose(float(weights.sum()), 1.0)
    )

    compatible = (
        24 in months
        and result.get("num_simulations") == 1000
        and invested is not None
        and stored_invested is not None
        and math.isclose(
            float(invested), float(stored_invested),
            rel_tol=1e-9, abs_tol=0.01
        )
        and paths_valid
        and portfolio_valid
        and weights_match
    )

    if not compatible:
        state["missing_data"].append("compatible_portfolio_scenarios")
        return

    state["portfolio_scenario_result"] = result
    state["portfolio_scenario_paths"] = paths
    state["result_parameters"]["bootstrap"] = {
        "horizon_months": 24,
        "num_simulations": 1000,
        "capital_eur": float(stored_invested),
        "source": "scenario_test",
        "complete_monthly_paths_verified": True,
        "portfolio_weights_provenance_verified": True,
        "portfolio_weights_used": stored_weights.copy()
    }

    state["data_sources"]["bootstrap"] = "scenario_test"


# ============================================================
# 8. INITIALIZE FROM THE EXISTING NOTEBOOK
# ============================================================

def initialize_unified_state_from_notebook(
    query="",
    namespace=None,
    allocation=0.50
) -> UnifiedAgentState:

    if namespace is None:
        namespace = globals()

    state = create_unified_agent_state(
        query=query,
        allocation=allocation
    )

    _box26_read_personal_finance(state, namespace)
    _box26_read_investment(state, namespace)
    _box26_read_capital(state, namespace)
    _box26_read_bootstrap(state, namespace)

    state["status"] = "data_loaded"

    return state


# ============================================================
# 9. VALIDATION
# ============================================================

unified_state = initialize_unified_state_from_notebook()

assert unified_state["rolling_months"] == 24
assert unified_state["investment_allocation_pct"] == 50.0

assert len(unified_state["personal_finance_stress_scenarios"]) <= 4

position = get_financial_position(unified_state)

print("\n" + "=" * 90)
print("BOX 26 — UNIFIED AGENT STATE")
print("=" * 90)

print("\nSTATE")
print("Fields:", len(unified_state))
print("Status:", unified_state["status"])
print("Horizon:", unified_state["rolling_months"], "months")
print("Allocation:", unified_state["investment_allocation_pct"], "%")

print("STARTING POINT: net savings frozen after 12 months; 50% invested and 50% retained as personal liquidity. The two portions evolve separately.")
print("\nFINANCIAL POSITION")
for key, value in position.items():
    print(f"{key}: {value}")

print("\nDATA SOURCES")
for key, value in unified_state["data_sources"].items():
    print(f"{key}: {value}")

print("\nPERSONAL FINANCE SCENARIOS")
print(
    "Retrieved:",
    len(unified_state["personal_finance_stress_scenarios"]),
    "/ 4"
)

print("\nMISSING OR UNVERIFIED DATA")
for item in unified_state["missing_data"]:
    print("-", item)

print("\nVALIDATION WARNINGS")
for item in unified_state["validation_warnings"]:
    print("-", item)

print("\nBOX 26 INITIALIZATION COMPLETED.")


# ============================================================
# 10. EXTENDED VALIDATION TESTS — BOX 26
# ============================================================

assert set(unified_state["personal_finance_stress_scenarios"]) == {
    "scenario_1", "scenario_2", "scenario_3", "scenario_4"
}, "The four numbered Personal Finance scenarios must be available."
assert all(
    unified_state["personal_finance_stress_scenarios"][f"scenario_{i}"] is not None
    for i in range(1, 5)
), "A Personal Finance scenario has no result."

capital_data = unified_state["result_parameters"].get("surplus_allocation", {})
for key in ("income_12m_eur", "expenses_12m_eur", "net_surplus_12m_eur", "initial_investment_eur"):
    assert unified_state.get(key) is not None and math.isfinite(float(unified_state[key])), key
assert math.isclose(
    unified_state["income_12m_eur"] - unified_state["expenses_12m_eur"],
    unified_state["net_surplus_12m_eur"], abs_tol=0.02
), "Income minus expenses does not reconcile with the annual surplus."
assert math.isclose(
    unified_state["initial_investment_eur"],
    max(unified_state["net_surplus_12m_eur"], 0) * 0.50,
    abs_tol=0.02
), "Investable capital must equal 50% of positive annual surplus."
assert math.isclose(
    unified_state["remaining_liquidity_eur"] + unified_state["initial_investment_eur"],
    unified_state["initial_liquidity_eur"], abs_tol=0.02
), "The investment transfer must not alter initial total wealth."

rolling_paths = unified_state.get("portfolio_scenario_paths")
assert isinstance(rolling_paths, pd.DataFrame), "Rolling bootstrap paths are missing."
assert "Month" in rolling_paths.columns
assert set(range(25)).issubset(set(rolling_paths["Month"].astype(int))), (
    "Rolling bootstrap must cover every month from 0 through 24."
)
rolling_pnl_columns = ["Unfavorable P5 (EUR)", "Central P50 (EUR)", "Favorable P95 (EUR)"]
assert all(col in rolling_paths.columns for col in rolling_pnl_columns)
assert np.isfinite(rolling_paths[rolling_pnl_columns].to_numpy(dtype=float)).all()
assert len(rolling_paths["Month"].unique()) == len(rolling_paths), (
    "Rolling bootstrap contains duplicate months."
)
assert np.all(
    rolling_paths["Unfavorable P5 (EUR)"].to_numpy() <=
    rolling_paths["Central P50 (EUR)"].to_numpy() + 1e-8
)
assert np.all(
    rolling_paths["Central P50 (EUR)"].to_numpy() <=
    rolling_paths["Favorable P95 (EUR)"].to_numpy() + 1e-8
)
assert all(
    abs(float(rolling_paths.loc[rolling_paths["Month"] == 0, col].iloc[0])) <= 0.02
    for col in rolling_pnl_columns
), "Bootstrap gains/losses must be zero at month 0."

# Verify the real bootstrap weight snapshot against optimized weights.
bootstrap_meta = unified_state["result_parameters"].get("bootstrap", {})
assert bootstrap_meta.get("complete_monthly_paths_verified") is True
assert bootstrap_meta.get("portfolio_weights_provenance_verified") is True
stored = bootstrap_meta["portfolio_weights_used"]
optimized = unified_state["portfolio_optimization_result"]["portfolio_weights"]
optimized = optimized[optimized > 0]
assert set(stored.index) == set(optimized.index)
assert np.allclose(stored.sort_index(), optimized.sort_index(), rtol=1e-10, atol=1e-12)
print("BOX 26 — EXTENDED VALIDATION TESTS PASSED")
print("BOX 26 — BOOTSTRAP WEIGHTS MATCH OPTIMIZED PORTFOLIO")


# ============================================================
# 11. SEMANTIC STATE CONTRACT TESTS — 41-QUESTION BENCHMARK
# ============================================================

assert sum(SEMANTIC_BENCHMARK_COVERAGE.values()) == 41
assert validate_semantic_request(unified_state) == []
assert unified_state["semantic_validation_status"] == "not_interpreted"
assert unified_state["requested_market_scenario"] is None
assert unified_state["income_reduction_months"] is None

# Confirm the state can carry indirect, integrated requests without
# performing keyword matching or inventing a missing shock duration.
semantic_probe = update_unified_agent_state(unified_state, {
    "query": "How would a financial setback affect liquidity recovery and investments?",
    "query_type": "integrated",
    "semantic_intent": "financial_resilience",
    "semantic_subintents": ["liquidity_recovery", "investment_impact"],
    "requires_integrated_assessment": True,
    "requested_market_scenario": "Unfavorable P5",
    "requested_market_percentile": 5,
    "semantic_validation_status": "interpreted_pending_execution",
})
assert validate_semantic_request(semantic_probe) == []
assert semantic_probe["income_reduction_months"] is None
assert semantic_probe["requested_horizon_months"] is None
assert unified_state["query_type"] == "unknown"
print("BOX 26 — SEMANTIC STATE CONTRACT TESTS PASSED")
```

```python
# ============================================================
# BOX 28 — AG-2: FINANCIAL ORCHESTRATOR
# Financial Decision Intelligence
#
# - Intent recognition and query classification
# - Existing tool registry
# - Parameter-aware result reuse
# - Personal Finance / Investment / Integrated routing
# - 15-scenario workflow planning
# - LangGraph-compatible orchestration node
# ============================================================


# ============================================================
# 1. VERIFY BOX 26
# ============================================================

required_box26 = [
    "UnifiedAgentState",
    "create_unified_agent_state",
    "initialize_unified_state_from_notebook"
]

for name in required_box26:
    if name not in globals():
        raise RuntimeError(
            f"Missing {name}. Execute corrected Box 26 first."
        )


# ============================================================
# 2. EXISTING TOOL REGISTRY
# ============================================================

def build_financial_tool_registry(namespace=None):

    if namespace is None:
        namespace = globals()

    pf_registry = namespace.get("PF5_ANALYTICAL_TOOLS", {})

    if not isinstance(pf_registry, dict):
        pf_registry = {}

    investment_registry = {}

    investment_functions = {
        "get_equity_ranking": "get_equity_ranking",
        "get_top_25_stocks": "get_top_25_stocks",
        "get_stock_analysis": "get_stock_analysis",
        "calculate_optimal_portfolio": "calculate_optimal_portfolio",
        "calculate_investable_capital": "calculate_investable_capital",
        "calculate_portfolio_scenarios": "calculate_portfolio_scenarios",
        "get_portfolio_scenarios": "get_portfolio_scenarios",
        "get_rolling_portfolio_scenarios": "get_rolling_portfolio_scenarios"
    }

    for key, function_name in investment_functions.items():
        fn = namespace.get(function_name)
        if callable(fn):
            investment_registry[key] = fn

    return {
        "personal_finance": {
            key: value
            for key, value in pf_registry.items()
            if callable(value)
        },
        "investment": investment_registry
    }


FINANCIAL_TOOL_REGISTRY = build_financial_tool_registry()


# ============================================================
# 3. INTENT RECOGNITION
# ============================================================

PERSONAL_PATTERNS = [
    r"\bbudget\b",
    r"\btotali mensili\b",
    r"\bcategorie di entrata\b",
    r"\bmesi in deficit\b",
    r"\briepilogo annuale\b",
    r"\briepilogo generale\b",
    r"\brisparmio medio mensile\b",
    r"\bcategorie di spesa\b",
    r"\bprincipali spese\b",
    r"\bconfronto annuale\b",
    r"\bconfronto mensile\b",
    r"\brapporto tra spese ed entrate\b",
    r"\btasso di risparmio\b",
    r"\bconfronto generale\b",
    r"\bconfronta\b",
    r"\bflussi? di cassa\b",
    r"\bautonomia finanziaria\b",
    r"\bflusso di cassa\b",
    r"\bcash[\s-]?flow\b",
    r"\bincome\b",
    r"\bexpenses?\b",
    r"\bsavings?\b",
    r"\bliquidity\b",
    r"\bsurplus\b",
    r"\bfinancial resilience\b",
    r"\bfinancial position\b",
    r"\bfinancial stress\b",
    r"\bemergency\b",
    r"\bsalary\b",
    r"\bspending\b",
    r"\breddito\b",
    r"\bentrate\b",
    r"\buscite\b",
    r"\bspese\b",
    r"\bspesa\b",
    r"\brisparmi[oa]?\b",
    r"\bliquidit[aà]\b",
    r"\bfinanza personale\b",
    r"\bposizione finanziaria\b",
    r"\bpatrimonio\b",
    r"\bfabbisogno\b"
    r"\btop\s*3\b",
    r"\bregolari\b",
    r"\bproblemi di liquidit[aà]\b",
    r"\bflussi di cassa\b",
    r"\bcritici\b",
    r"\bstabile\b",
    r"\btenere sotto controllo\b",
    r"\bquanto posso risparmiare\b",
    r"\brisparmio futuro\b",
    r"\bsimulazione del risparmio\b",
    r"\bmettere soldi da parte\b",
    r"\bfondo di emergenza\b",
    r"\bspese essenziali\b",
    r"\bresilienza finanziaria\b",
    r"\bentrate si riducono\b",
    r"\briduzione del reddito\b",
    r"\breddito da lavoro\b",
    r"\bbonus\b",
    r"\bpremio di produttivit[aà]\b",
    r"\bspesa straordinaria\b",
    r"\bspesa imprevista\b",
    r"\bspese ricorrenti\b",
    r"\bliquidit[aà] futura\b",
    r"\briserve? di sicurezza\b",
    r"\btempi di recupero\b",
    r"\bimprevist[io]\b",
    r"\bsenza vendere investimenti\b",
]

INVESTMENT_PATTERNS = [
    r"\binvest\w*\b",
    r"\bportfolio\b",
    r"\bstocks?\b",
    r"\bequit(?:y|ies)\b",
    r"\bsharpe\b",
    r"\bvolatility\b",
    r"\bbootstrap\b",
    r"\bmonte carlo\b",
    r"\btop[\s-]?25\b",
    r"\basset allocation\b",
    r"\bmarket performance\b",
    r"\bmarket risk\b",
    r"\bportafoglio\b",
    r"\btitoli\b",
    r"\bazioni\b",
    r"\brendimento\b",
    r"\bvolatilit[aà]\b",
    r"\bmercati?\b",
    r"\bmercato\b",
    r"\brendimenti? contenuti\b",
    r"\brendimenti? moderati\b"
]

INTEGRATED_PATTERNS = [
    r"\binvest\w*.*\b(?:surplus|liquidity|savings|cash|budget)\b",
    r"\b(?:surplus|liquidity|savings|cash|budget).*"
    r"\binvest\w*\b",
    r"\b(?:portfolio|portafoglio).*"
    r"\b(?:liquidity|cash|savings|liquidit[aà]|risparmio)\b",
    r"\b(?:liquidity|cash|savings|liquidit[aà]|risparmio).*"
    r"\b(?:portfolio|portafoglio)\b",
    r"\bfinancial position.*\bmarket\b",
    r"\bpatrimonio.*\binvestiment\w*\b",
    r"\b(?:due|entrambe le) componenti\b",
    r"\b(?:two|both)\s+(?:financial\s+)?components\b",
    r"\b(?:liquidit[aà]|risparmio).*\binvestiment\w*\b",
    r"\binvestiment\w*.*\b(?:liquidit[aà]|risparmio)\b"
]

COMPARISON_PATTERNS = [
    r"\bcompare\b",
    r"\bcomparison\b",
    r"\bversus\b",
    r"\bvs\.?\b",
    r"\bdifference\b",
    r"\bbetter\b",
    r"\bworse\b",
    r"\bconfront\w*\b",
    r"\bdifferenz\w*\b",
    r"\brispetto a\b",
    r"\bmiglior\w*\b",
    r"\bpeggior\w*\b"
]

STRESS_PATTERNS = [
    r"\bstress\b",
    r"\bshock\b",
    r"\badverse\b",
    r"\bunexpected expense\b",
    r"\bincome reduction\b",
    r"\bextraordinary expense\b",
    r"\bspesa straordinaria\b",
    r"\bspesa imprevista\b",
    r"\briduzione del reddito\b"
]


def _matches(text, patterns):
    return any(
        re.search(pattern, text, flags=re.IGNORECASE)
        for pattern in patterns
    )


def classify_financial_query(query):

    if not isinstance(query, str):
        raise TypeError("Query must be a string.")

    text = " ".join(query.lower().split())

    if not text:
        return "unknown"

    # Reuse the complete semantic decision logic of the original
    # Personal Finance orchestrator (Box 11), without changing its tools.
    original_personal_decision = agent_node({"query": query})
    personal = (
        _matches(text, PERSONAL_PATTERNS)
        or original_personal_decision.get("action") != "end"
    )
    investment = _matches(text, INVESTMENT_PATTERNS)
    integrated = _matches(text, INTEGRATED_PATTERNS)

    if integrated or (personal and investment):
        return "integrated"

    if personal:
        return "personal_finance"

    if investment:
        return "investment"

    return "unknown"


# ============================================================
# 4. QUERY REQUIREMENTS
# ============================================================

def identify_query_requirements(query):

    text = " ".join(query.lower().split())
    query_type = classify_financial_query(query)

    comparison = _matches(text, COMPARISON_PATTERNS)
    stress = _matches(text, STRESS_PATTERNS)
    duration_match = re.search(r"(?:per|durante)\s+(un|uno|una|tre|sei|dodici|\d{1,2})\s+mes[ei]\b", text)
    duration_values = {"un": 1, "uno": 1, "una": 1, "1": 1,
                       "3": 3, "tre": 3, "6": 6, "sei": 6,
                       "12": 12, "dodici": 12}
    income_reduction_months = (duration_values.get(duration_match.group(1), int(duration_match.group(1)) if duration_match.group(1).isdigit() else None) if duration_match else None)
    if income_reduction_months is not None and not 1 <= income_reduction_months <= 24:
        raise ValueError("Durata shock ammessa: da 1 a 24 mesi.")

    return {
        "query_type": query_type,
        "comparison_requested": comparison,
        "stress_analysis_requested": stress,
        "income_reduction_months": income_reduction_months,
        "integrated_analysis_requested": (
            query_type == "integrated"
        ),
        "horizon_months": 24,
        "personal_scenario_count": 5,
        "investment_scenario_count": 3,
        "integrated_combination_count": (
            15 if query_type == "integrated" else 0
        )
    }


# ============================================================
# 5. RESULT AVAILABILITY AND PARAMETER COMPATIBILITY
# ============================================================

def check_available_financial_results(state):

    pf = bool(state.get("personal_finance_results"))

    pf_scenarios = (
        bool(state.get("personal_finance_baseline"))
        and all(
            f"scenario_{i}" in state.get(
                "personal_finance_stress_scenarios", {}
            )
            for i in range(1, 5)
        )
    )

    # The stored ranking and optimization are the current notebook results.
    # Reuse is restricted to requests with no conflicting parameters.
    requested = state.get("requested_parameters", {})
    default_portfolio_request = not any(
        key in requested for key in (
            "tickers", "weights", "portfolio_weights", "capital_eur",
            "num_portfolios", "max_assets_per_sector", "ranking_weights"
        )
    )
    ranking = (
        bool(state.get("investment_analysis_result"))
        and default_portfolio_request
    )
    portfolio = (
        bool(state.get("portfolio_optimization_result"))
        and default_portfolio_request
    )

    # Use the bootstrap already transferred to Unified State by Box 26.
    # Only check compatibility with the current request, not the
    # internal simulation settings of the completed bootstrap.
    bootstrap_result = state.get("portfolio_scenario_result")
    bootstrap_paths = state.get("portfolio_scenario_paths")
    bootstrap_parameters = state.get("result_parameters", {}).get("bootstrap", {})
    optimized_weights = state.get("portfolio_optimization_result", {}).get("portfolio_weights")
    stored_weights = bootstrap_parameters.get("portfolio_weights_used")

    def _same_weights(left, right):
        if not isinstance(left, pd.Series) or not isinstance(right, pd.Series):
            return False
        left = left[left > 0].sort_index()
        right = right[right > 0].sort_index()
        return (
            left.index.is_unique and right.index.is_unique
            and left.index.equals(right.index)
            and np.allclose(left.to_numpy(dtype=float), right.to_numpy(dtype=float),
                            rtol=1e-10, atol=1e-12)
        )

    # The stored bootstrap must refer to the optimized portfolio.
    portfolio_match = _same_weights(stored_weights, optimized_weights)
    requested_weights = requested.get("portfolio_weights", requested.get("weights"))
    if requested_weights is not None:
        portfolio_match = portfolio_match and _same_weights(
            stored_weights, requested_weights
        )
    requested_assets = requested.get("tickers", requested.get("assets"))
    if requested_assets is not None:
        portfolio_match = (
            portfolio_match
            and isinstance(stored_weights, pd.Series)
            and set(requested_assets) == set(stored_weights[stored_weights > 0].index)
        )

    invested = state.get("initial_investment_eur")
    stored_capital = bootstrap_parameters.get("capital_eur")
    requested_capital = requested.get("capital_eur", invested)
    capital_match = (
        invested is not None and stored_capital is not None
        and requested_capital is not None
        and np.isclose(float(invested), float(stored_capital), rtol=1e-9, atol=0.01)
        and np.isclose(float(requested_capital), float(stored_capital), rtol=1e-9, atol=0.01)
    )

    requested_months = requested.get("horizon_months", state.get("rolling_months", 24))
    months_match = (
        isinstance(bootstrap_paths, pd.DataFrame)
        and "Month" in bootstrap_paths.columns
        and isinstance(requested_months, (int, np.integer))
        and requested_months >= 0
        and requested_months in set(bootstrap_paths["Month"].dropna().tolist())
    )

    bootstrap = (
        isinstance(bootstrap_result, dict) and bool(bootstrap_result)
        and portfolio_match and capital_match and months_match
    )

    integrated = (
        len(state.get("integrated_scenarios", {})) == 15
        and state.get("result_parameters", {}).get(
            "integrated_horizon_months"
        ) == 24
    )

    return {
        "personal_finance": pf,
        "personal_finance_scenarios": pf_scenarios,
        "investment_ranking": ranking,
        "portfolio_optimization": portfolio,
        "portfolio_bootstrap": bootstrap,
        "integrated_assessment": integrated
    }


# ============================================================
# 6. WORKFLOW PLANNING
# ============================================================

def build_financial_execution_plan(
    query_type,
    available,
    requirements
):

    plan = []

    def add_step(agent, purpose, reuse=False):
        plan.append({
            "agent": agent,
            "purpose": purpose,
            "action": "reuse" if reuse else "execute"
        })

    if query_type in ("personal_finance", "integrated"):

        add_step(
            "personal_finance_agent",
            "personal_finance_analysis",
            available["personal_finance"]
            and available["personal_finance_scenarios"]
        )

    if query_type in ("investment", "integrated"):

        add_step(
            "investment_risk_agent",
            "portfolio_risk_and_scenarios",
            available["investment_ranking"]
            and available["portfolio_optimization"]
            and available["portfolio_bootstrap"]
        )

    if query_type == "integrated":

        add_step(
            "financial_assessment_agent",
            "integrated_15_scenario_assessment",
            available["integrated_assessment"]
        )

    if query_type != "unknown":

        add_step(
            "response_generation_agent",
            "financial_response_generation"
        )

    return plan


# ============================================================
# 7. MISSING DATA
# ============================================================

def identify_missing_financial_data(
    query_type,
    available
):

    required = []

    if query_type in ("personal_finance", "integrated"):
        required += [
            "personal_finance",
            "personal_finance_scenarios"
        ]

    if query_type in ("investment", "integrated"):
        required += [
            "investment_ranking",
            "portfolio_optimization",
            "portfolio_bootstrap"
        ]

    if query_type == "integrated":
        required.append("integrated_assessment")

    return [
        item for item in required
        if not available.get(item, False)
    ]


# ============================================================
# 8. MAIN FINANCIAL ORCHESTRATOR
# ============================================================

def financial_orchestrator(
    state: UnifiedAgentState
) -> UnifiedAgentState:

    updated = dict(state)

    query = updated.get("query", "")
    if isinstance(query, str) and query.strip().casefold().rstrip("?.! ") == "con gli imprevisti simulati, potrei far fronte alle spese senza vendere investimenti":
        updated.setdefault("requested_parameters", {})["horizon_months"] = 24

    requirements = identify_query_requirements(query)
    query_type = requirements["query_type"]

    available = check_available_financial_results(updated)

    missing = identify_missing_financial_data(
        query_type, available
    )

    plan = build_financial_execution_plan(
        query_type,
        available,
        requirements
    )

    updated["query_type"] = query_type
    updated["income_reduction_months"] = requirements["income_reduction_months"]
    updated["rolling_months"] = 24

    updated["analysis_type"] = (
        "comparative"
        if requirements["comparison_requested"]
        else "standard"
    )

    updated["requested_parameters"] = {
        **updated.get("requested_parameters", {}),
        **requirements,
        "available_results": available,
        "missing_data": missing
    }

    updated["execution_plan"] = plan
    updated["missing_data"] = missing

    updated["active_agent"] = (
        plan[0]["agent"] if plan else None
    )

    updated["status"] = (
        "clarification_required"
        if query_type == "unknown"
        else "planned"
    )

    return updated


# ============================================================
# 9. LANGGRAPH-COMPATIBLE NODE
# ============================================================

def financial_orchestrator_node(
    state: UnifiedAgentState
) -> dict:

    result = financial_orchestrator(state)

    return {
        "query_type": result["query_type"],
        "analysis_type": result["analysis_type"],
        "rolling_months": result["rolling_months"],
        "requested_parameters": result["requested_parameters"],
        "income_reduction_months": result["requested_parameters"].get("income_reduction_months"),
        "execution_plan": result["execution_plan"],
        "active_agent": result["active_agent"],
        "missing_data": result["missing_data"],
        "status": result["status"]
    }


# ============================================================
# 10. ROUTING
# ============================================================

def route_financial_request(state):

    routes = {
        "personal_finance": "personal_finance_agent",
        "investment": "investment_risk_agent",
        "integrated": "integrated_workflow",
        "unknown": "clarification_required"
    }

    return routes.get(
        state.get("query_type", "unknown"),
        "clarification_required"
    )


# ============================================================
# 11. CLASSIFICATION TESTS
# ============================================================

TEST_QUERIES = [
    ("What is my monthly cash flow?", "personal_finance"),
    ("How would a 10000 euro unexpected expense affect my savings?",
     "personal_finance"),
    ("Which stocks are included in the Top 25 ranking?",
     "investment"),
    ("What is the Sharpe Ratio of my portfolio?",
     "investment"),
    ("How would investing 50% of my surplus affect my liquidity?",
     "integrated"),
    ("Compare my savings and portfolio performance under adverse scenarios.",
     "integrated"),
    ("How would a 20000 euro unexpected expense affect my liquidity and investment portfolio?",
     "integrated"),
    ("What happens to my financial position if I invest half of my available surplus?",
     "integrated"),
    ("Confronta la mia liquidità e il portafoglio in uno scenario di stress.",
     "integrated"),
    ("Tell me a joke.", "unknown")
]

print("\n" + "=" * 90)
print("BOX 28 — FINANCIAL ORCHESTRATOR")
print("=" * 90)

print("\nQUERY CLASSIFICATION TESTS")

for i, (query, expected) in enumerate(TEST_QUERIES, 1):

    detected = classify_financial_query(query)

    assert detected == expected, (
        f"Test {i}: expected {expected}, got {detected}"
    )

    print(f"TEST {i}: {detected} — PASS")


# ============================================================
# 12. TOOL REGISTRY TEST
# ============================================================

print("\nEXISTING TOOL REGISTRY")

for domain, tools in FINANCIAL_TOOL_REGISTRY.items():
    print(f"\n{domain}: {len(tools)} tools")

    for name in tools:
        print(" -", name)


# ============================================================
# 13. REAL STATE ORCHESTRATION TEST
# ============================================================

test_state = dict(unified_state)

test_state["query"] = (
    "Compare my liquidity and investment portfolio "
    "under financial stress scenarios."
)

orchestrated_state = financial_orchestrator(test_state)

assert orchestrated_state["query_type"] == "integrated"
assert orchestrated_state["rolling_months"] == 24
assert len(orchestrated_state["execution_plan"]) == 4

assert (
    orchestrated_state["requested_parameters"][
        "integrated_combination_count"
    ] == 15
)

print("\nINTEGRATED EXECUTION PLAN")

for i, step in enumerate(
    orchestrated_state["execution_plan"], 1
):
    print(
        f"{i}. {step['agent']} "
        f"| {step['purpose']} "
        f"| {step['action']}"
    )

print("\nDATA AVAILABILITY")

available = orchestrated_state[
    "requested_parameters"
]["available_results"]

for key, value in available.items():
    print(f"{key}: {value}")

print("\nMISSING DATA")
for item in orchestrated_state["missing_data"]:
    print("-", item)

print("\nROUTING")
print(route_financial_request(orchestrated_state))

print("\nSCENARIO ARCHITECTURE")
print("Personal Finance: 5 conditions")
print("Investment Risk: 3 scenarios")
print("Integrated combinations: 15")
print("Horizon: 24 months")

print("\n" + "=" * 90)
print("BOX 28 — CLASSIFICATION AND PLANNING TESTS PASSED")
print("READY FOR VALIDATION BEFORE BOX 28")
print("=" * 90)


# ============================================================
# 14. EXTENDED VALIDATION TESTS — BOX 28
# ============================================================

for question, expected_type, expected_route, expected_agents in [
    ("What is my monthly cash flow?", "personal_finance", "personal_finance_agent", 2),
    ("What is the Sharpe Ratio of my portfolio?", "investment", "investment_risk_agent", 2),
    ("Compare my liquidity and investment portfolio under stress.", "integrated", "integrated_workflow", 4),
]:
    route_state = dict(unified_state)
    route_state["query"] = question
    routed = financial_orchestrator(route_state)
    assert routed["query_type"] == expected_type
    assert route_financial_request(routed) == expected_route
    assert len(routed["execution_plan"]) == expected_agents
    assert routed["execution_plan"][-1]["agent"] == "response_generation_agent"

# ============================================================
# BOX 28 — BOOTSTRAP REUSE TESTS (EXISTING UNIFIED STATE)
# ============================================================

# No new bootstrap calculation and no second initialization.
# All tests use independent copies of the same Box 26 state.
bootstrap_test_state = deepcopy(unified_state)
bootstrap_test_state["query"] = (
    "Compare my liquidity and investment portfolio under stress."
)
bootstrap_test_state["requested_parameters"] = {"horizon_months": 24}

assert check_available_financial_results(bootstrap_test_state)["portfolio_bootstrap"], (
    "Bootstrap is not reusable from the existing Box 26 unified_state."
)

planned = financial_orchestrator(deepcopy(bootstrap_test_state))
investment_step = next(
    step for step in planned["execution_plan"]
    if step["agent"] == "investment_risk_agent"
)
assert investment_step["action"] == "reuse"

# Every month already covered by the rolling results can be reused.
months_available = set(
    unified_state["portfolio_scenario_paths"]["Month"].dropna().tolist()
)
for month in range(1, 25):
    if month in months_available:
        case = deepcopy(bootstrap_test_state)
        case["requested_parameters"]["horizon_months"] = month
        assert check_available_financial_results(case)["portfolio_bootstrap"], month

# An unavailable period must not reuse the existing result.
period_mismatch = deepcopy(bootstrap_test_state)
period_mismatch["requested_parameters"]["horizon_months"] = (
    max(months_available) + 1
)
assert not check_available_financial_results(period_mismatch)["portfolio_bootstrap"]

# Incompatible capital and weights must not reuse the existing result.
capital_mismatch = deepcopy(bootstrap_test_state)
capital_mismatch["requested_parameters"]["capital_eur"] = -1.0
assert not check_available_financial_results(capital_mismatch)["portfolio_bootstrap"]

weights_mismatch = deepcopy(bootstrap_test_state)
weights_mismatch["requested_parameters"]["portfolio_weights"] = (
    unified_state["result_parameters"]["bootstrap"]["portfolio_weights_used"].copy() * 0.9
)
assert not check_available_financial_results(weights_mismatch)["portfolio_bootstrap"]

print("BOX 28 — BOOTSTRAP REUSE TESTS PASSED")
print("Compatible portfolio/capital/period: REUSE")
print("Incompatible portfolio/capital/period: EXECUTE")
```

```python
# ============================================================
# BOX 29 — AG-4: INVESTMENT RISK AGENT
#
# Reuses the existing stock ranking, optimized portfolio,
# and historical bootstrap simulations to assess investment risk.
#
# Initial capital allocation:
# 50% of accumulated net savings (€3,132.26) is invested,
# while the remaining 50% (€3,132.26) is retained as personal liquidity.
# Both components evolve separately over the following 24 months.
# ============================================================


def investment_risk_agent(state: UnifiedAgentState) -> UnifiedAgentState:
    """Expose verified investment results without recomputing the portfolio."""
    updated = dict(state)
    updated["agent_results"] = dict(state.get("agent_results") or {})
    updated["errors"] = list(state.get("errors") or [])
    updated["active_agent"] = "investment_risk_agent"

    available = check_available_financial_results(state)
    ranking = state.get("investment_analysis_result") or {}
    optimization = state.get("portfolio_optimization_result") or {}
    scenarios = state.get("portfolio_scenario_result") or {}
    rolling_paths = state.get("portfolio_scenario_paths")
    capital = state.get("initial_investment_eur")

    # Do not substitute a different portfolio or silently recompute results.
    updated["agent_results"]["investment_risk_agent"] = {
        "ranking": ranking if available["investment_ranking"] else None,
        "portfolio_optimization": (
            optimization if available["portfolio_optimization"] else None
        ),
        "portfolio_weights": (
            state.get("portfolio_weights")
            if available["portfolio_optimization"] else None
        ),
        "initial_investment_eur": capital,
        "portfolio_scenarios": (
            scenarios if available["portfolio_bootstrap"] else None
        ),
        "rolling_portfolio_scenarios": (
            rolling_paths if available["portfolio_bootstrap"] else None
        ),
        "bootstrap_reused": bool(available["portfolio_bootstrap"]),
        "source": "Existing Investment results (Boxes 18–25 and 26)",
    }

    missing = [
        key for key in (
            "investment_ranking", "portfolio_optimization", "portfolio_bootstrap"
        ) if not available[key]
    ]
    if missing:
        updated["errors"].append(
            "Investment results unavailable or incompatible: " + ", ".join(missing)
        )
    updated["status"] = (
        "investment_ready" if not missing else "investment_incomplete"
    )
    return updated


def investment_risk_agent_node(state: UnifiedAgentState) -> dict:
    """LangGraph node: return only the fields updated by AG-4."""
    result = investment_risk_agent(state)
    return {
        "agent_results": result["agent_results"],
        "errors": result["errors"],
        "active_agent": result["active_agent"],
        "status": result["status"],
    }


# ============================================================
# BOX 29 — VALIDATION
# ============================================================

assert callable(investment_risk_agent)
assert callable(investment_risk_agent_node)

box29_test_state = deepcopy(unified_state)
box29_test_state["requested_parameters"] = {}
box29_result = investment_risk_agent(box29_test_state)

assert box29_result["status"] == "investment_ready", (
    "Existing investment results must be available and compatible."
)
assert box29_result["agent_results"]["investment_risk_agent"]["bootstrap_reused"] is True
assert box29_result["portfolio_scenario_result"] is box29_test_state["portfolio_scenario_result"]
assert box29_result["portfolio_scenario_paths"] is box29_test_state["portfolio_scenario_paths"]
assert box29_result["portfolio_optimization_result"] is box29_test_state["portfolio_optimization_result"]
assert box29_result["agent_results"]["investment_risk_agent"]["portfolio_weights"] is box29_test_state["portfolio_weights"]
assert box29_result["agent_results"]["investment_risk_agent"]["initial_investment_eur"] == box29_test_state["initial_investment_eur"]

print("BOX 29 — INVESTMENT RISK AGENT — PASSED")
print("STARTING POINT: net savings frozen after 12 months; 50% invested and 50% retained as personal liquidity. The portfolio evolves separately according to market performance.")
print("Equity ranking: REUSED")
print("Optimized portfolio and weights: REUSED")
print("Rolling portfolio bootstrap: REUSED")
print("Investment results: UNCHANGED")
```

```python
# ============================================================
# BOX 30 — AG-5: FINANCIAL ASSESSMENT AGENT
#
# Integrates Personal Finance and Investment Risk analyses
# to assess overall financial resilience and sustainability.
#
# Combines 5 Personal Finance conditions with 3 investment
# scenarios to generate 15 integrated financial assessments.
#
# Evaluates liquidity, portfolio performance, total financial
# wealth, and potential funding requirements across
# different time horizons.
# ============================================================


def financial_assessment_agent(state: UnifiedAgentState) -> UnifiedAgentState:
    """Integrate the original monthly PF and investment paths; no upstream recalculation."""
    updated = dict(state)
    updated["agent_results"] = dict(state.get("agent_results") or {})
    updated["errors"] = list(state.get("errors") or [])
    updated["active_agent"] = "financial_assessment_agent"

    pf = updated["agent_results"].get("personal_finance_agent")
    inv = updated["agent_results"].get("investment_risk_agent")
    if pf is None or inv is None:
        raise ValueError("Execute Box 28 and Box 29 agents before Box 30.")
    if not state.get("personal_finance_results") or not inv.get("bootstrap_reused"):
        raise ValueError("Validated Personal Finance and Investment results are required.")

    # Preserve original objects and names; retrieve monthly trajectories from PF-4.
    original_pf = state["personal_finance_results"]
    pf_paths = original_pf["monthly_trajectories"]
    investment_paths = state["portfolio_scenario_paths"]
    invested = float(state["initial_investment_eur"])
    initial_cash = float(state["initial_liquidity_eur"])
    remaining_cash = float(state["remaining_liquidity_eur"])
    months = int(state.get("rolling_months", 24))

    # Source labels from Box 25, without renaming its columns.
    investment_columns = {
        "Unfavorable P5": "Unfavorable P5 Value (EUR)",
        "Central P50": "Central P50 Value (EUR)",
        "Favorable P95": "Favorable P95 Value (EUR)",
    }
    # The original PF-4 scenarios use S1–S4 internally; present them as 1–4.
    pf_conditions = {"baseline": None}
    for number in range(1, 5):
        matches = [key for key in pf_paths if key.split(" — ")[0] == f"S{number}"]
        if len(matches) != 1:
            raise ValueError(f"Original Personal Finance scenario {number} unavailable.")
        pf_conditions[str(number)] = matches[0]

    # Box 25 paths contain trading-day points, including exact integer months.
    investment_monthly = investment_paths.loc[
        np.isclose(investment_paths["Month"].astype(float) % 1, 0, atol=1e-8)
    ].copy()
    investment_monthly["Month"] = investment_monthly["Month"].round().astype(int)
    investment_monthly = investment_monthly.set_index("Month")
    required_months = set(range(1, months + 1))
    if not required_months.issubset(set(investment_monthly.index)):
        raise ValueError("The existing bootstrap does not contain all requested months.")

    results = []
    monthly_results = {}
    for pf_label, original_name in pf_conditions.items():
        if original_name is None:
            # Baseline is the original PF-4 no-stress path (identical across all four runs).
            reference = next(iter(pf_paths.values()))
            pf_monthly = reference[["Month", "No_Stress_Liquidity_EUR"]].copy()
            pf_cash_column = "No_Stress_Liquidity_EUR"
        else:
            pf_monthly = pf_paths[original_name][["Month", "Stress_Liquidity_EUR"]].copy()
            pf_cash_column = "Stress_Liquidity_EUR"
        pf_monthly = pf_monthly.set_index("Month")
        if not required_months.issubset(set(pf_monthly.index.astype(int))):
            raise ValueError(f"Missing monthly Personal Finance values for {pf_label}.")

        # PF-4 cash includes initial cash. Deduct investment exactly once at month 0.
        # Investment gains/losses never flow automatically into spendable cash.
        liquidity = pf_monthly.loc[range(1, months + 1), pf_cash_column].astype(float) - invested
        baseline_cash = (next(iter(pf_paths.values()))
                         .set_index("Month").loc[range(1, months + 1), "No_Stress_Liquidity_EUR"]
                         .astype(float) - invested)
        for investment_label, value_column in investment_columns.items():
            value = investment_monthly.loc[range(1, months + 1), value_column].astype(float)
            frame = pd.DataFrame({
                "Month": list(range(1, months + 1)),
                "liquidity_eur": liquidity.to_numpy(),
                "investment_value_eur": value.to_numpy(),
            })
            frame["total_financial_wealth_eur"] = (
                frame["liquidity_eur"] + frame["investment_value_eur"]
            )
            frame["additional_funding_required_eur"] = (
                -frame["liquidity_eur"].cummin().clip(upper=0)
            )
            # Funding is a separately reported requirement, not fictional new wealth.
            min_cash = min(remaining_cash, float(frame["liquidity_eur"].min()))
            max_funding = max(0.0, -min_cash)
            final_cash = float(frame["liquidity_eur"].iloc[-1])
            final_investment = float(frame["investment_value_eur"].iloc[-1])
            baseline_final_cash = float(baseline_cash.iloc[-1])
            scenario_id = f"{pf_label} | {investment_label}"
            monthly_results[scenario_id] = frame
            results.append({
                "personal_finance_scenario": pf_label,
                "investment_scenario": investment_label,
                "initial_liquidity_eur": initial_cash,
                "initial_investment_eur": invested,
                "remaining_liquidity_eur": remaining_cash,
                "minimum_liquidity_eur": min_cash,
                "additional_funding_required_eur": max_funding,
                "months_negative_liquidity": int((frame["liquidity_eur"] < 0).sum()),
                "final_liquidity_eur": final_cash,
                "final_investment_value_eur": final_investment,
                "investment_gain_loss_eur": final_investment - invested,
                "final_total_financial_wealth_eur": final_cash + final_investment,
                "personal_finance_impact_vs_baseline_eur": final_cash - baseline_final_cash,
                "horizon_months": months,
            })

    summary = pd.DataFrame(results)
    worst_liquidity = summary.loc[summary["minimum_liquidity_eur"].idxmin()].to_dict()
    worst_wealth = summary.loc[summary["final_total_financial_wealth_eur"].idxmin()].to_dict()
    integrated = {
        "scenario_summary": summary,
        "monthly_scenarios": monthly_results,
        "most_critical_liquidity": worst_liquidity,
        "most_critical_total_wealth": worst_wealth,
        "personal_finance_conditions": 5,
        "investment_scenarios": 3,
        "integrated_combinations": 15,
        "horizon_months": months,
        "initial_investment_eur": invested,
        "remaining_liquidity_eur": remaining_cash,
        "methodology": (
            "Existing PF-4 monthly liquidity less the initial investment once; "
            "plus existing Box 25 portfolio values. Investments are not assumed liquidated; "
            "funding needs are reported separately."
        ),
    }
    updated["agent_results"]["financial_assessment_agent"] = integrated
    updated["integrated_assessment_result"] = integrated
    updated["status"] = "integrated_assessment_ready"
    return updated


def financial_assessment_agent_node(state: UnifiedAgentState) -> dict:
    result = financial_assessment_agent(state)
    return {
        "agent_results": result["agent_results"],
        "integrated_assessment_result": result["integrated_assessment_result"],
        "errors": result["errors"],
        "active_agent": result["active_agent"],
        "status": result["status"],
    }




def box30_income_duration_path(current_assessment, market, duration_months=None, reduction_pct=0.30):
    """Personal income stress with requested duration; original portfolio paths unchanged."""
    if not isinstance(duration_months, int) or not 1 <= duration_months <= 24:
        raise ValueError("Required shock duration: an integer from 1 to 24 months.")
    baseline = current_assessment["monthly_scenarios"][f"baseline | {market}"]
    frame = baseline.copy(deep=True)
    # Use actual employment income from Box 7's verified historical category.
    if "pf4_results" not in globals() or not isinstance(pf4_results, dict):
        raise RuntimeError("Box 7 results are unavailable for the dynamic shock.")
    inputs = pf4_results.get("inputs", {})
    employment = inputs.get("Average monthly Employment income EUR")
    if employment is None:
        raise RuntimeError("Monthly employment income is unavailable in Box 7 results.")
    reduction = float(employment) * float(reduction_pct)
    frame["liquidity_eur"] = (
        baseline["liquidity_eur"].astype(float)
        - frame["Month"].astype(int).clip(upper=duration_months) * reduction
    )
    frame["total_financial_wealth_eur"] = frame["liquidity_eur"] + frame["investment_value_eur"]
    frame["additional_funding_required_eur"] = -frame["liquidity_eur"].cummin().clip(upper=0)
    return frame

# ============================================================
# BOX 30 — INTEGRATED QUESTION, 5 PERSONAL FINANCE CONDITIONS x 3 MARKET SCENARIOS
# Four test horizons; rolling results can be queried from 1 to 24 months.
# Original results and agent keys remain unchanged.
# ============================================================

box30_state = dict(unified_state)
box30_state["agent_results"] = {}
box30_state = personal_finance_agent(box30_state)
box30_state = investment_risk_agent(box30_state)
box30_state = financial_assessment_agent(box30_state)
assessment = box30_state["integrated_assessment_result"]
monthly_tables = assessment["monthly_scenarios"]

_pf_descriptions = {
    "baseline": "Baseline: savings follow the historical trend, with no shock",
    "1": "Employment income reduced by 30% for 6 months",
    "2": "One-off expense of €10,000 in the first month",
    "3": "One-off expense of €20,000 in the first month",
    "4": "Recurring expenses increased by 15% for 12 months",
}
_market_descriptions = {
    "Unfavorable P5": "Unfavorable market (worst of the three scenarios considered)",
    "Central P50": "Central market (intermediate performance)",
    "Favorable P95": "Favorable market (best of the three scenarios considered)",
}


def _box30_eur(value):
    return f"{float(value):,.2f} €".replace(",", "§").replace(".", ",").replace("§", ".")


def _box30_signed(value):
    return ("+" if value >= 0 else "−") + _box30_eur(abs(value))


def _box30_pct(value):
    return ("+" if value >= 0 else "−") + f"{abs(value):.2f}%".replace(".", ",")


def _box30_snapshot(month, pf_condition, investment_scenario):
    """Read existing month from Box 30 integration, without recomputing upstream."""
    if not isinstance(month, int) or not 1 <= month <= assessment["horizon_months"]:
        raise ValueError("Month unavailable in the original rolling trajectories.")
    key = f"{pf_condition} | {investment_scenario}"
    frame = monthly_tables[key]
    match = frame.loc[frame["Month"] == month]
    if len(match) != 1:
        raise ValueError(f"Month {month} unavailable for {key}.")
    return match.iloc[0]


def _box30_answer(month, pf_condition, investment_scenario):
    """Natural-language answer with the cash and investment outcomes kept separate."""
    point = _box30_snapshot(month, pf_condition, investment_scenario)
    baseline = _box30_snapshot(month, "baseline", investment_scenario)
    initial_cash = float(box30_state["initial_liquidity_eur"])
    invested = float(assessment["initial_investment_eur"])
    cash_after_investment = float(assessment["remaining_liquidity_eur"])
    cash = float(point["liquidity_eur"])
    portfolio = float(point["investment_value_eur"])
    gain = portfolio - invested
    pct = 100.0 * gain / invested if invested else 0.0
    pf_impact = cash - float(baseline["liquidity_eur"])
    net_cash_change = cash - cash_after_investment
    market_outcome = (
        f"gain of {_box30_eur(gain)} ({_box30_pct(pct)})" if gain > 0
        else f"loss of {_box30_eur(-gain)} ({_box30_pct(pct)})" if gain < 0
        else "no change relative to invested capital (0.00%)"
    )
    print(f"\nPERIOD: {month} months")
    print(f"PERSONAL FINANCE: {_pf_descriptions[pf_condition]}")
    print(f"INVESTMENTS: {_market_descriptions[investment_scenario]}")
    print(f"Initial bank liquidity before investment: {_box30_eur(initial_cash)}.")
    print(f"Capital allocated to the portfolio: {_box30_eur(invested)}.")
    print(f"Initial liquidity after investment: {_box30_eur(cash_after_investment)}.")
    print(f"After {month} months, projected personal liquidity is {_box30_eur(cash)} "
          f"(change relative to liquidity after investment: {_box30_signed(net_cash_change)}).")
    if pf_condition == "baseline":
        print("No personal finance shock is applied: baseline savings are projected.")
    else:
        print(f"Impact of the personal finance shock relative to the no-shock baseline "
              f"in the same month: {_box30_signed(pf_impact)} in liquidity.")
    print(f"Without assuming sales or withdrawals, the portfolio is worth {_box30_eur(portfolio)}: "
          f"{market_outcome} after {month} months.")
    if cash < 0:
        print(f"Projected liquidity falls short by {_box30_eur(-cash)} "
              "at this horizon; investments remain separate.")
    else:
        print(f"Personal liquidity available at this horizon is {_box30_eur(cash)}.")
    return {
        "month": month, "personal_finance_scenario": pf_condition,
        "investment_scenario": investment_scenario,
        "liquidity_eur": cash, "investment_value_eur": portfolio,
        "investment_gain_loss_eur": gain, "investment_return_pct": pct,
        "personal_finance_impact_vs_baseline_eur": pf_impact,
    }


QUESTION = (
    "How will my personal liquidity and the portfolio I am about to "
    "build evolve, considering the baseline or one of the specified "
    "personal finance shocks and an unfavorable, central, or favorable market? "
    "How much will I save or lose, and how much will the investment gain or lose?"
)

print("BOX 30 — INTEGRATED FINANCIAL ASSESSMENT")
print("QUESTION:", QUESTION)
print("Methodology — Savings freeze and initial allocation")
print("At the end of the 12-month observation period, accumulated net savings amount to €6,264.52. Of this, 50% (€3,132.26) is allocated to the investment portfolio, while the remaining 50% (€3,132.26) remains available as personal liquidity.")
print("STARTING POINT: after 12 months of observation, accumulated net savings are frozen; 50% is allocated to the investment portfolio and the remaining 50% stays as personal liquidity. From the following month, the two components evolve separately according to personal finance and market scenarios.")
print(f"Net savings after 12 months: EUR {float(box30_state['net_surplus_12m_eur']):,.2f}; initial portfolio: EUR {float(box30_state['initial_investment_eur']):,.2f}; initial personal liquidity: EUR {float(box30_state['remaining_liquidity_eur']):,.2f}.")
print(
    "MARKET SCENARIOS: the three cases — unfavorable, central, and "
    "favorable — derive from bootstrap analysis of the aggregate portfolio "
    "of selected assets weighted according to the optimized portfolio. "
    "The simulations produce trajectories of the total portfolio value "
    "and allow three gain-or-loss outcomes to be estimated for each rolling "
    "period: unfavorable, central, and favorable. "
    "The reported percentage changes refer to the entire period "
    "considered, not to an individual month."
)
print("Changes in personal liquidity and portfolio value are reported separately.")
print("Each answer specifies the period, shock, capital, and percentage gain/loss.")

# There is only one question; the conditions change, not its meaning.
# 5 personal finance conditions x 3 market scenarios, repeated at 3, 6, 12, and 24 months.
box30_test_answers = []
for month in (3, 6, 12, 24):
    print("\n" + "=" * 88)
    print(f"ASSESSMENT AFTER {month} MONTHS")
    print("=" * 88)
    for pf_condition in ("baseline", "1", "2", "3", "4"):
        for investment_scenario in (
            "Unfavorable P5", "Central P50", "Favorable P95"
        ):
            print("\n" + "-" * 88)
            box30_test_answers.append(
                _box30_answer(month, pf_condition, investment_scenario)
            )

# Rolling results can be queried for any available month: _box30_answer(10, "3", "Central P50")
# The test only checks that each period can be queried; it does not revalidate
# the quantitative data already verified in previous boxes.
assert all(
    len(monthly_tables[f"{pf} | {inv}"].loc[
        monthly_tables[f"{pf} | {inv}"]["Month"] == month
    ]) == 1
    for month in range(1, assessment["horizon_months"] + 1)
    for pf in ("baseline", "1", "2", "3", "4")
    for inv in ("Unfavorable P5", "Central P50", "Favorable P95")
)
assert len(box30_test_answers) == 60
print("\nBOX 30 — 60 ANSWERS (15 COMBINATIONS x 4 PERIODS) — COMPLETED")
```

```python
# ============================================================
# BOX 31 — AG-6: RESPONSE GENERATION
#
# Interprets user questions and generates contextualized
# financial assessments based on the integrated analysis.
#
# Combines Personal Finance and Investment Risk insights
# to provide clear, evidence-based financial explanations.
#
# Supports flexible time horizons from 1 to 24 months,
# adapting responses to the requested financial conditions
# and investment scenarios.
# ============================================================


# Exclusive source: results already produced by Box 30.
if "box30_state" not in globals() or "assessment" not in globals():
    raise RuntimeError("Run Box 30 first: integrated results are missing.")

_BOX31_PF = {
    "baseline": "ordinary financial conditions, without shocks",
    "1": "30% reduction in income for 6 months",
    "2": "one-off expense of €10,000 in the first month",
    "3": "one-off expense of €20,000 in the first month",
    "4": "15% increase in recurring expenses for 12 months",
}
_BOX31_MARKETS = {
    "Unfavorable P5": "unfavorable",
    "Central P50": "central",
    "Favorable P95": "favorable",
}
_BOX31_METHOD = (
    "Methodology — Allocation of half of accumulated savings to investments and initial capital allocation\n"
    "At the end of the 12-month observation period, accumulated net savings amount to €6,264.52. "
    "50%, corresponding to €3,132.26, is allocated to the investment portfolio; "
    "the remaining 50%, equal to €3,132.26, is retained as available personal liquidity."
)


def _box31_normalize(value):
    value = unicodedata.normalize("NFKD", str(value).lower())
    return "".join(c for c in value if not unicodedata.combining(c))


def _box31_interpret(question):
    """Identifies the month, personal scenario, and market scenario; flags ambiguity rather than inventing results."""
    q = _box31_normalize(question)
    hits = re.findall(r"\b(\d{1,2})\s*(?:months|month)\b", q)
    if len(set(hits)) != 1:
        raise ValueError("Specify a single time horizon in months (from 1 to 24).")
    month = int(hits[0])
    if not 1 <= month <= min(24, int(assessment["horizon_months"])):
        raise ValueError("Time horizon unavailable: only 1 to 24 months are supported.")

    pf_matches = []
    if re.search(r"\bordinary\b|\bnormal\b|\bwithout shocks?\b|\bwithout unexpected expenses\b|\bbaseline\b|\bno shocks?\b", q):
        pf_matches.append("baseline")
    if ("income" in q or "salary" in q or "earnings" in q) and ("30%" in q or "30 percent" in q):
        pf_matches.append("1")
    if "10,000" in q or "10000" in q or "ten thousand" in q:
        pf_matches.append("2")
    if "20,000" in q or "20000" in q or "twenty thousand" in q:
        pf_matches.append("3")
    if ("expenses" in q or "costs" in q) and ("15%" in q or "15 percent" in q):
        pf_matches.append("4")
    if len(set(pf_matches)) != 1:
        raise ValueError("Specify one personal financial condition: baseline, income -30%, one-off expense €10,000/€20,000, or recurring expenses +15%.")

    markets = []
    if any(t in q for t in ("unfavorab", "negative", "pessimist", "bearish", "downturn", "declining market", "market decline", "p5")):
        markets.append("Unfavorable P5")
    if any(t in q for t in ("central", "intermediate", "median", "moderate returns", "modest returns", "p50")):
        markets.append("Central P50")
    if ("favorab" in q and "unfavorab" not in q) or any(t in q for t in ("positive", "optimist", "bullish", "rising market", "market growth", "p95")):
        markets.append("Favorable P95")
    if len(set(markets)) != 1:
        raise ValueError("Specify one market scenario: unfavorable, central, or favorable.")
    return month, pf_matches[0], markets[0]


def box31_respond(question, show_methodology=True, print_answer=True):
    """Traceable narrative response based on the monthly trajectories already validated in Box 30."""
    month, pf, market = _box31_interpret(question)
    frame = assessment["monthly_scenarios"][f"{pf} | {market}"]
    point = frame.loc[frame["Month"] == month]
    if len(point) != 1:
        raise ValueError("The requested month is missing from the original Box 30 results.")
    row = point.iloc[0]
    baseline = assessment["monthly_scenarios"][f"baseline | {market}"]
    baseline_row = baseline.loc[baseline["Month"] == month].iloc[0]
    invested = float(assessment["initial_investment_eur"])
    starting_cash = float(assessment["remaining_liquidity_eur"])
    cash = float(row["liquidity_eur"])
    portfolio = float(row["investment_value_eur"])
    gain = portfolio - invested
    pct = 100 * gain / invested if invested else 0.0
    impact = cash - float(baseline_row["liquidity_eur"])
    history = frame.loc[frame["Month"] <= month, "liquidity_eur"].astype(float)
    funding = max(0.0, -min(starting_cash, float(history.min())))

    eur = _box30_eur  # Reuses the original Box 30 formatting.
    signed = _box30_signed
    percent = _box30_pct
    lines = []
    if show_methodology:
        lines.append(_BOX31_METHOD)
    lines += [
        f"QUESTION: {question}",
        f"ROLLING HORIZON: {month} months following the investment of half of the savings accumulated during the preceding 12 months.",
        f"PERSONAL FINANCIAL CONDITION: {_BOX31_PF[pf]}.",
        f"MARKET: {_BOX31_MARKETS[market]}; cumulative portfolio return {percent(pct)} "
        f"({'gain' if gain > 0 else 'loss' if gain < 0 else 'no change'}), "
        "derived from the optimized portfolio bootstrap simulation.",
        f"INITIAL POSITION: personal liquidity {eur(starting_cash)}; invested portfolio {eur(invested)}.",
        f"AFTER {month} MONTHS — PERSONAL LIQUIDITY: {eur(cash)} "
        f"(change from initial liquidity {signed(cash-starting_cash)}).",
        f"AFTER {month} MONTHS — INVESTMENT PORTFOLIO: {eur(portfolio)} "
        f"({'gain' if gain > 0 else 'loss' if gain < 0 else 'change'} "
        f"{eur(abs(gain))}, cumulative return {percent(pct)}).",
    ]
    if pf != "baseline":
        lines.append(f"SHOCK IMPACT on liquidity relative to baseline conditions at the same month: {signed(impact)}.")
    else:
        lines.append("No personal financial shock applied.")
    if funding > 0:
        lines.append(f"WARNING: an uncovered liquidity requirement of up to {eur(funding)} arises during the period; investment liquidation is not assumed to finance it.")
    else:
        lines.append("No uncovered liquidity requirement arises during the period considered.")
    lines.append("Personal financial results are derived from Personal Finance trajectories; portfolio results from bootstrap trajectories. The two components are neither combined nor automatically offset.")
    result = {"question":question,"month":month,"personal_finance_scenario":pf,
              "investment_scenario":market,"liquidity_eur":cash,"investment_value_eur":portfolio,
              "investment_gain_loss_eur":gain,"investment_return_pct":pct,
              "funding_required_eur":funding,"text":"\n".join(lines)}
    if print_answer:
        print(result["text"])
    return result


def box31_respond_many(questions, print_answers=True):
    """N independent questions; no artificial limit on the number of requests."""
    outputs = []
    for i, question in enumerate(questions, 1):
        if print_answers:
            print("\n" + "=" * 84 + f"\nRESPONSE {i}\n" + "=" * 84)
        outputs.append(box31_respond(question, print_answer=print_answers))
    return outputs


# SEMANTIC TESTS: months not included in the Box 30 demonstration.
# Validate interpretation, point extraction, numerical accuracy, and language.
BOX31_TEST_QUESTIONS = [
    "After 5 months, with income reduced by 30% and an unfavorable market, how do liquidity and investments evolve?",
    "After 8 months, if I incur an extraordinary expense of €10,000 and the market is favorable, what happens?",
    "After 11 months, with a €20,000 expense and a central market, what is my funding shortfall and portfolio value?",
    "After 17 months, under ordinary financial conditions and a negative market, how much liquidity and investment capital do I have?",
    "After 23 months, with expenses increased by 15% and a positive market, what are the results?",
    "After 1 month, without shocks and with a central market, what are the values of the two components?",
    "After 24 months, with income reduced by 30% and a favorable market, what is the financial assessment?",
]

box31_test_answers = box31_respond_many(BOX31_TEST_QUESTIONS)
for answer in box31_test_answers:
    m, pf, market = _box31_interpret(answer["question"])
    original = assessment["monthly_scenarios"][f"{pf} | {market}"]
    original = original.loc[original["Month"] == m].iloc[0]
    assert answer["month"] == m and answer["personal_finance_scenario"] == pf
    assert answer["investment_scenario"] == market
    assert abs(answer["liquidity_eur"] - float(original["liquidity_eur"])) < 1e-8
    assert abs(answer["investment_value_eur"] - float(original["investment_value_eur"])) < 1e-8
    assert "PERSONAL LIQUIDITY" in answer["text"] and "INVESTMENT PORTFOLIO" in answer["text"]
    assert "cumulative return" in answer["text"] and "INITIAL POSITION" in answer["text"]
    assert "(P5)" not in answer["text"] and "(P50)" not in answer["text"] and "(P95)" not in answer["text"]
    assert f"MARKET: {_BOX31_MARKETS[market]}; cumulative portfolio return {_box30_pct(answer['investment_return_pct'])}" in answer["text"]
assert len(box31_test_answers) == len(BOX31_TEST_QUESTIONS)
for invalid in ("After 25 months, without shocks and with a central market?", "After 5 months, how is my financial position?"):
    try:
        box31_respond(invalid, print_answer=False)
    except ValueError:
        pass
    else:
        raise AssertionError("An ambiguous or out-of-range question must not generate invented figures.")
print(f"\nBOX 31 — {len(box31_test_answers)} SEMANTIC TESTS AND 2 VALIDATION TESTS COMPLETED")

# Single question: box31_respond("After 8 months, with income reduced by 30% and a central market, what happens?")
# Multiple questions: box31_respond_many(["After 5 months, without shocks and with a central market?", ...])


# ============================================================
# BOX 31 — UNIFIED RESPONSE ENTRYPOINT (extension)
# Preserves box31_respond and its verified integrated answers.
# Uses Box 27 orchestrator for routing; reuses original PF LangGraph.
# ============================================================

def box31_answer(question, rolling_months=24, print_answer=True):
    question = str(question or "").strip()
    if not question:
        return {"status": "clarification_required", "domain": "unknown", "text": "Please enter a question."}
    if "financial_orchestrator" not in globals():
        raise RuntimeError("Box 27 not loaded: Financial Orchestrator is missing.")
    state = dict(unified_state)
    state["query"] = question
    plan = financial_orchestrator(state)
    domain = plan.get("query_type", "unknown")
    if "bootstrap" in question.lower() and "scenar" in question.lower():
        # Returns the three previously calculated bootstrap scenarios without running new simulations.
        match = re.search(r"\b(?:after\s+)?(\d+)\s+months?\b", question.lower())
        paths = unified_state.get("portfolio_scenario_paths")
        invested = unified_state.get("initial_investment_eur")
        if match is None:
            output = {"status": "clarification_required", "domain": "investment",
                      "text": "Specify the number of months for the three bootstrap scenarios."}
        elif paths is None or invested is None:
            output = {"status": "unsupported", "domain": "investment",
                      "text": "Bootstrap results or initial capital are unavailable in the unified state."}
        else:
            month = int(match.group(1))
            matching = paths.loc[np.isclose(paths["Month"].astype(float), month)]
            if matching.empty:
                output = {"status": "unsupported", "domain": "investment",
                          "text": f"Bootstrap results are unavailable after {month} months."}
            else:
                point = matching.iloc[0]
                capital = float(invested)
                def _eur(value):
                    return f"€ {value:,.2f}".replace(",", "_").replace(".", ",").replace("_", ".")
                lines = [f"Portfolio after {month} months — three bootstrap scenarios:",
                         f"Initial invested capital: {_eur(capital)}", "",
                         "Scenario | Cumulative Return | Portfolio Value"]
                for label, key in (("Unfavorable", "Unfavorable P5"),
                                   ("Central", "Central P50"),
                                   ("Favorable", "Favorable P95")):
                    pct = float(point[f"{key} (%)"])
                    value = float(point[f"{key} Value (EUR)"])
                    lines.append(f"{label} | {pct:+.2f}% | {_eur(value)}")
                output = {"status": "response_ready", "domain": "investment",
                          "text": "\n".join(lines)}
    elif domain == "investment" and re.search(r"\b(?:portfolio|investments?|invested capital)\b", question.lower()) and not re.search(r"\b(?:optimi[sz]\w*|monte carlo|ranking|top[ -]?25|sharpe)\b", question.lower()):
        # Queries existing bootstrap trajectories for the requested scenario and month only.
        q = _box31_normalize(question)
        months = re.findall(r"\b(\d{1,2})\s*months?\b", q)
        scenarios = []
        if any(term in q for term in ("unfavorab", "negative", "pessimist", "bearish", "downturn", "declining market", "market decline")):
            scenarios.append(("Unfavorable", "Unfavorable P5"))
        if any(term in q for term in ("intermediate", "central", "median", "moderate returns", "modest returns")):
            scenarios.append(("Central", "Central P50"))
        if ("favorab" in q and "unfavorab" not in q) or any(term in q for term in ("positive", "optimist", "bullish", "rising market", "market growth")):
            scenarios.append(("Favorable", "Favorable P95"))
        if len(set(months)) != 1 or not 1 <= int(months[0]) <= 24 or len(scenarios) != 1:
            output = {"status": "clarification_required", "domain": domain,
                      "text": "Specify a month from 1 to 24 and one market scenario: unfavorable, central, or favorable."}
        else:
            month = int(months[0])
            paths = unified_state.get("portfolio_scenario_paths")
            invested = unified_state.get("initial_investment_eur")
            matching = (paths.loc[np.isclose(paths["Month"].astype(float), month)]
                        if isinstance(paths, pd.DataFrame) and "Month" in paths.columns else None)
            label, key = scenarios[0]
            if matching is None or matching.empty or invested is None:
                output = {"status": "unsupported", "domain": domain,
                          "text": f"Bootstrap results are unavailable after {month} months."}
            else:
                point = matching.iloc[0]
                pct = float(point[f"{key} (%)"])
                value = float(point[f"{key} Value (EUR)"])
                def _eur(value):
                    return f"€ {value:,.2f}".replace(",", "_").replace(".", ",").replace("_", ".")
                output = {"status": "response_ready", "domain": domain,
                          "text": (f"Portfolio after {month} months — {label.lower()} scenario:\n"
                                   f"Initial invested capital: {_eur(float(invested))}\n"
                                   f"Cumulative return: {pct:+.2f}%\n"
                                   f"Portfolio value: {_eur(value)}")}
    elif domain == "integrated":
        # This is the ORIGINAL tested semantic response, with no parallel recomputation.
        try:
            result = box31_respond(question, print_answer=False)
            output = {**result, "domain": domain, "status": "response_ready"}
        except ValueError as exc:
            output = {"status": "clarification_required", "domain": domain, "text": str(exc)}
    elif domain == "personal_finance":
        if "app" not in globals() or not hasattr(app, "invoke"):
            raise RuntimeError("The original Personal Finance graph (Boxes 9–12) is unavailable.")
        try:
            raw = app.invoke({"query": re.sub(r"performance[\s-]+related payment", "performance award", question, flags=re.IGNORECASE), "result": "", "rolling_months": int(rolling_months)})
            answer = raw.get("result", "")
            if not isinstance(answer, str) and "format_financial_response" in globals():
                answer = format_financial_response(answer, raw.get("analysis_type", ""), question)
            output = {"status": "response_ready" if str(answer).strip() else "incomplete",
                      "domain": domain, "text": str(answer), "source_result": raw}
        except Exception as exc:
            output = {"status": "error", "domain": domain, "text": f"Original graph error: {exc}"}
    elif domain == "investment":
        # Dedicated presentation of the STOXX Europe 50 ranking, without exposing raw time series or dictionaries.
        ranking_question = ("ranking" in question.lower() and
                            ("stoxx" in question.lower() or "top 25" in question.lower()))
        optimization_question = (
            "monte carlo" in question.lower()
            or "optimiz" in question.lower()
            or "optimis" in question.lower()
            or "final portfolio" in question.lower()
            or "optimized portfolio" in question.lower()
            or "optimised portfolio" in question.lower()
        )
        if optimization_question:
            # Presents the final portfolio already calculated by the Monte Carlo optimization engine.
            result = globals().get("portfolio_optimization_result")
            if not isinstance(result, dict) or not all(
                key in result for key in ("portfolio_composition", "optimal_sharpe")
            ):
                output = {"status": "unsupported", "domain": domain,
                          "text": "Monte Carlo optimization results are unavailable: run the Portfolio Optimization engine box."}
            else:
                composition = result["portfolio_composition"]
                lines = [
                    "Final portfolio optimized through Monte Carlo simulation (maximum Sharpe Ratio):",
                    "",
                    "Ticker | Company | Weight (%)",
                ]
                for _, item in composition.iterrows():
                    lines.append(
                        f"{item['Ticker']} | {item['Company']} | {item['Weight (%)']:.2f}%"
                    )
                lines.extend([
                    "",
                    f"Portfolio Sharpe Ratio: {result['optimal_sharpe']:.4f}",
                ])
                output = {"status": "response_ready", "domain": domain,
                          "text": "\n".join(lines)}
        elif ranking_question:
            ranking_result = calculate_equity_ranking(prices=df2, metadata=stock_metadata)
            ranking_table = ranking_result["ranking_df"]
            lines = [
                "Ranking developed to select 25 securities from an initial universe of 50 STOXX Europe 50 constituents, based on five years of historical price data.",
                "",
                "The selection of the 25 securities precedes portfolio and weight optimization.",
                "",
                "The Combined Score is the average of the positions obtained in two rankings, rewarding higher annualized returns and lower volatility, respectively. Since first place represents the best result, a lower combined score indicates a more favorable assessment.",
                "",
                "Final ranking of the 25 selected securities (equal scores share the same rank):",
                "Rank | Ticker | Company | Annualized Return | Annualized Volatility | Combined Score",
            ]
            for idx, (ticker, item) in enumerate(ranking_table.head(25).iterrows()):
                pos = int((ranking_table["Combined Score"] < item["Combined Score"]).sum()) + 1
                lines.append(
                    f"{pos} | {ticker} | {item['Company']} | "
                    f"{item['Annualized Return']:.2%} | {item['Annualized Volatility']:.2%} | "
                    f"{item['Combined Score']:.2f}"
                )
            output = {"status": "response_ready", "domain": domain, "text": "\n".join(lines)}
        else:
            # Existing investment domain response from Box 32 (preserved below as helper).
            if "box31_investment_existing" not in globals():
                output = {"status": "unsupported", "domain": domain,
                          "text": "Investment formatter is unavailable for this question."}
            else:
                output = box31_investment_existing(question)
    else:
        output = {"status": "clarification_required", "domain": domain,
                  "text": "Question not recognized by the existing orchestrator."}
    output["orchestrator_plan"] = plan.get("execution_plan", [])
    if print_answer:
        print(output["text"])
    return output
```

```python
# ============================================================
# BOX 32
# GRADIO — UNIFIED FINANCIAL DECISION INTELLIGENCE
# Three-column interface: original Personal Finance analysis,
# original Personal Finance decisions, Investment and integrated analysis.
# Single backend: box31_ask() from the final LangGraph (Box 31).
# No calls to the initial provisional LangGraph.
# ============================================================

# Questions for the three financial analysis tools.
ANALYSIS_QUESTIONS = [
    "Show me the monthly totals",
    "Show me the top 3 income and expense items",
    "What is the average monthly savings amount?",
    "Show me the expense categories",
    "Show me the income categories",
    "Which months have a deficit?",
    "Show me the annual summary",
    "Show me the overall summary",
    "Analyze the main expenses during deficit months",
    "Compare 2026-01 and 2026-02",
    "Show me the annual comparison between income and expenses",
    "Show me the monthly comparison between income and expenses",
    "Show me the expense-to-income ratio",
    "Show me the savings rate",
    "Show me the overall comparison between income and expenses",
]

# Questions formulated as practical everyday financial problems.
# Each question contains terms recognized by the decision agent.
# Numerical assumptions are those of the existing simulations.
DECISION_QUESTIONS = [
    "Are my income and expenses sufficiently regular, or am I at risk of liquidity problems?",
    "Are there months when my cash flows become particularly critical?",
    "How stable is my cash flow, and what should I monitor?",
    "With my current income and expenses, how much can I save?",
    "If I continue saving at the current rate, how might my future savings grow?",
    "What does the savings simulation reveal about my ability to set money aside?",
    "If I faced a difficult financial period, how many months of financial autonomy would I have?",
    "Is my emergency fund sufficient to cover essential expenses?",
    "How resilient are my finances if my income decreases?",
    "If I did not receive my usual bonus or performance-related payment and my employment income fell by 30% for six months, what would happen to my liquidity?",
    "If I suddenly had to pay an extraordinary expense of €10,000, could I afford it?",
    "And if I faced an extraordinary expense of €20,000, how severely would it affect my savings?",
    "If my recurring expenses increased by 15% for one year, how much would my future liquidity decrease?",
    "Between an income reduction, an unexpected expense, and an increase in recurring expenses, which scenario would put my financial safety reserve at the greatest risk?",
    "After an unexpected financial event, how long would it take for my liquidity to recover?",
]

# Third column: seven integrated questions validated in Box 31,
# plus standalone investment questions handled by the Box 29 agent.
# Original questions are neither replaced nor modified beyond translation.
INTEGRATED_INVESTMENT_QUESTIONS = [
    "After 3 months, with income reduced by 30% for 3 months and an unfavorable market, how do liquidity and investments evolve?",
    "After 8 months, if I incur an extraordinary expense of €10,000 and the market is favorable, what happens?",
    "After 11 months, with an expense of €20,000 and a market with moderate returns, what is my funding shortfall and portfolio value?",
    "After 7 months, under ordinary financial conditions and a negative market, how much liquidity and investment capital do I have?",
    "After 1 month, without shocks and with a market with moderate returns, what are the values of the two components?",
    "After 12 months, with a reduced performance bonus (income reduced by 30% for one month) and a market with moderate returns, what is the impact on liquidity and investments?",
    "Show me the STOXX Europe 50 equity ranking and the Top 25 selection.",
    "Show me the Monte Carlo portfolio optimization and Sharpe Ratio.",
    "What are the weights and composition of the optimized portfolio?",
    "After 6 months, what are the portfolio values under the three bootstrap scenarios?",
    "After 12 months, what is the portfolio value under unfavorable market conditions?",
]


# ============================================================
# 2. PREPARE FINANCIAL DATA FOR CHARTS
# ============================================================

def prepare_financial_data():

    data = get_rolling_data(24)

    data["Month"] = pd.to_datetime(data["Month"])

    monthly = (
        data.groupby(["Month", "Transaction Type"])["Amount"]
        .sum()
        .unstack(fill_value=0)
        .sort_index()
    )

    for col in ["credit", "debit"]:
        if col not in monthly.columns:
            monthly[col] = 0.0

    monthly["Balance"] = monthly["credit"] - monthly["debit"]

    return data, monthly


# ============================================================
# 3. CHART 1 — MONTHLY INCOME, EXPENSES AND BALANCE
# ============================================================

def create_monthly_chart(monthly):

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=monthly.index,
            y=monthly["credit"],
            mode="lines+markers",
            name="Income",
            line=dict(color="#2196F3", width=3)
        )
    )

    fig.add_trace(
        go.Scatter(
            x=monthly.index,
            y=monthly["debit"],
            mode="lines+markers",
            name="Expenses",
            line=dict(color="#FF9800", width=3)
        )
    )

    fig.add_trace(
        go.Scatter(
            x=monthly.index,
            y=monthly["Balance"],
            mode="lines+markers",
            name="Balance",
            line=dict(color="#4CAF50", width=3)
        )
    )

    fig.update_layout(
        title="Monthly Income, Expenses and Balance",
        xaxis_title="Month",
        yaxis_title="EUR",
        hovermode="x unified",
        template="plotly_white",
        height=420,
        margin=dict(l=40, r=30, t=60, b=40)
    )

    return fig


# ============================================================
# 4. CHART 2 — EXPENSE DISTRIBUTION BY CATEGORY
# ============================================================

def create_expense_chart(data):

    expenses = data[
        data["Transaction Type"] == "debit"
    ]

    category_totals = (
        expenses.groupby("Category")["Amount"]
        .sum()
        .sort_values(ascending=False)
        .reset_index()
    )

    fig = px.bar(
        category_totals,
        x="Amount",
        y="Category",
        orientation="h",
        title="Expense Distribution by Category",
        labels={
            "Amount": "Total Expenses (EUR)",
            "Category": "Expense Category"
        },
        color="Amount",
        color_continuous_scale="Blues"
    )

    fig.update_layout(
        template="plotly_white",
        height=450,
        yaxis=dict(autorange="reversed"),
        showlegend=False,
        coloraxis_showscale=False,
        margin=dict(l=40, r=30, t=60, b=40)
    )

    return fig


# ============================================================
# 5. CHART 3 — MONTHLY SURPLUS AND DEFICIT
# ============================================================

def create_balance_chart(monthly):

    colors = [
        "#4CAF50" if value >= 0 else "#F44336"
        for value in monthly["Balance"]
    ]

    fig = go.Figure()

    fig.add_trace(
        go.Bar(
            x=monthly.index,
            y=monthly["Balance"],
            marker_color=colors,
            name="Monthly Balance",
            hovertemplate=(
                "Month: %{x|%Y-%m}<br>"
                "Balance: €%{y:,.2f}<extra></extra>"
            )
        )
    )

    fig.add_hline(
        y=0,
        line_color="gray",
        line_width=1
    )

    fig.update_layout(
        title="Monthly Surplus and Deficit",
        xaxis_title="Month",
        yaxis_title="Balance (EUR)",
        template="plotly_white",
        height=420,
        margin=dict(l=40, r=30, t=60, b=40)
    )

    return fig


# ============================================================
# 6. EXECUTE LANGGRAPH QUERY
# ============================================================

def run_financial_analysis(question, rolling_months):
    """Queries only the final LangGraph from Box 31."""
    if not question or not question.strip():
        return "Please enter a question."
    try:
        result = box31_answer(
            question.strip(),
            rolling_months=int(rolling_months),
            print_answer=False,
        )
        if result.get("status") != "response_ready":
            return ("Request not resolved by the final graph: "
                    + str(result.get("text") or result.get("error") or "no result"))
        answer = result.get("text")
        return str(answer).lstrip("\n") if answer is not None else "No result available."
    except Exception as exc:
        return f"Error during analysis through Box 31: {exc}"


# ============================================================
# 7. GENERATE ALL THREE CHARTS
# ============================================================

def generate_charts():

    data, monthly = prepare_financial_data()

    return (
        create_monthly_chart(monthly),
        create_expense_chart(data),
        create_balance_chart(monthly)
    )


# ============================================================
# 8. GRADIO INTERFACE
# ============================================================

CUSTOM_CSS = """
.gradio-container { max-width: 1680px !important; margin: auto !important; }
#finance-header { padding: 16px 2px 10px; }
#finance-header h1 { font-size: 2rem; letter-spacing: -0.035em; margin-bottom: 4px; }
#finance-header p { color: #64748b; }
.question-example {
    position: relative !important;
    text-align: left !important;
    justify-content: flex-start !important;
    white-space: normal !important;
    min-height: 51px !important;
    padding: 11px 35px 11px 14px !important;
    line-height: 1.35 !important;
    font-weight: 450 !important;
}
.question-example::after {
    content: "+";
    position: absolute;
    right: 14px;
    top: 50%;
    transform: translateY(-50%);
    font-size: 20px;
    pointer-events: none;
    opacity: .65;
}
#analysis-panel, #decision-panel, #integrated-panel { min-width: 0 !important; }
#answer-panel textarea { font-family: inherit !important; line-height: 1.55 !important; }
"""

with gr.Blocks(title="Financial Decision Intelligence — Simulation Framework", css=CUSTOM_CSS) as demo_box32:
    gr.HTML('<section style="max-width:1100px;margin:0 auto 20px;font-family:inherit;line-height:1.65;color:inherit;">\n  <header style="padding:8px 0 18px;border-bottom:1px solid #9aa3af55;">\n    <h1 style="font-size:clamp(23px,3vw,32px);line-height:1.2;margin:0 0 10px;">Financial Decision Intelligence — Simulation Framework</h1>\n    <p style="margin:0 0 10px;">Personal budgeting, investment risks, and integrated financial assessments</p>\n    <p style="font-weight:650;margin:0;">12-month observation period + 24-month projection horizon</p>\n  </header>\n  <h2 style="font-size:20px;margin:23px 0 12px;">METHODOLOGY</h2>\n  <p>At the end of the <strong>12-month personal budget observation period</strong>, accumulated net savings amount to <strong>€6,264.52</strong>.</p>\n  <p>We assume that the user decides to allocate 50% to an investment portfolio, retaining the remaining 50% as available personal liquidity.</p>\n  <div style="display:grid;grid-template-columns:repeat(auto-fit,minmax(220px,1fr));gap:12px;margin:16px 0 20px;">\n    <div style="padding:15px 18px;border:1px solid #9aa3af66;border-radius:10px;">\n      <div style="font-size:13px;opacity:.8;">50% · Investment Portfolio</div>\n      <div style="font-size:23px;font-weight:700;">€3,132.26</div>\n    </div>\n    <div style="padding:15px 18px;border:1px solid #9aa3af66;border-radius:10px;">\n      <div style="font-size:13px;opacity:.8;">50% · Available Personal Liquidity</div>\n      <div style="font-size:23px;font-weight:700;">€3,132.26</div>\n    </div>\n  </div>\n  <p>User queries are positioned at <strong>month zero</strong>, representing the initial simulation point after the 12-month observation period and initial savings allocation. From this point, the system can <strong>project the evolution of the personal budget and investment portfolio over a time horizon ranging from 1 to 24 months</strong>.</p>\n  <div style="border:1px solid #9aa3af66;border-radius:10px;padding:4px 18px;margin:16px 0;">\n    <h3 style="font-size:16px;margin:12px 0 6px;">Personal Budget Projection</h3>\n    <p style="margin:0 0 13px;">The personal budget is projected using historical monthly averages of income and expenses, with potential <strong>financial shocks</strong> applied to assess their impact.</p>\n    <div style="border-top:1px solid #9aa3af55;"></div>\n    <h3 style="font-size:16px;margin:12px 0 6px;">Investment Portfolio Projection</h3>\n    <p style="margin:0 0 13px;">The future evolution of the investment portfolio is simulated using <strong>historical return bootstrapping</strong> of the portfolio assets, accounting for the volatility characteristics and historical price fluctuations of individual securities.</p>\n  </div>\n  <p>The system <strong>integrates both projections</strong>, providing, for the requested time horizon, an estimate of <strong>personal liquidity, portfolio value, and overall financial wealth</strong> under the selected scenarios.</p>\n  <h2 style="font-size:20px;margin:24px 0 12px;padding-top:18px;border-top:1px solid #9aa3af55;">OBJECTIVE</h2>\n  <p>The platform aims to <strong>optimize personal financial management through integrated planning of budgeting, savings, and investments</strong>.</p>\n  <p>A detailed analysis of income and expense categories helps identify liquidity requirements, assess potential financial imbalances, and plan for the coverage of funding gaps. At the same time, it helps determine the portion of savings effectively available for investment, allocating it consistently with current and future financial needs.</p>\n  <p>The objective is therefore to <strong>support appropriate savings allocation, preserve the liquidity required to cover expenses, and make effective use of available financial resources</strong>, ensuring that investments contribute to the overall balance and efficiency of personal financial management.</p>\n  <p style="font-size:12px;opacity:.75;padding-top:14px;border-top:1px solid #9aa3af55;">Results represent projections and simulations conditional on the underlying assumptions, rather than certain predictions of future personal financial conditions or market performance.</p>\n</section>')

    with gr.Row():
        rolling_selector = gr.Radio(
            choices=[("3 months", 3), ("6 months", 6), ("12 months", 12), ("2 years", 24)],
            value=12,
            label="Analysis Period",
            interactive=True,
        )

    question_input = gr.Textbox(
        label="Your Question",
        placeholder="Enter a question or select one of the examples in the three columns…",
        lines=2,
    )

    with gr.Row():
        analyze_button = gr.Button("Analyze Question", variant="primary", scale=3)
        clear_button = gr.Button("Clear", variant="secondary", scale=1)

    gr.Markdown("### Response")
    answer_output = gr.Textbox(
        label="Agent Response",
        placeholder="The response from the final LangGraph (Box 31) will appear here.",
        lines=10,
        max_lines=28,
        interactive=False,
        elem_id="answer-panel",
    )

    analyze_button.click(
        fn=run_financial_analysis,
        inputs=[question_input, rolling_selector],
        outputs=answer_output,
    )
    question_input.submit(
        fn=run_financial_analysis,
        inputs=[question_input, rolling_selector],
        outputs=answer_output,
    )
    clear_button.click(fn=lambda: ("", ""), inputs=[], outputs=[question_input, answer_output])

    gr.Markdown("### Choose a Question")
    gr.Markdown("Select an example question. Choose a situation of interest: the question will appear in the field above. You can edit it before clicking **Analyze Question**.")

    with gr.Row(equal_height=False):
        with gr.Column(scale=1, min_width=310, elem_id="analysis-panel"):
            gr.Markdown("#### Financial Analysis")
            gr.Markdown("Income, expenses, deficits, categories, and historical comparisons.")
            for example in ANALYSIS_QUESTIONS:
                button = gr.Button(example, variant="secondary", elem_classes=["question-example"])
                button.click(fn=lambda q=example: q, inputs=[], outputs=question_input)

        with gr.Column(scale=1, min_width=310, elem_id="decision-panel"):
            gr.Markdown("#### Financial Decision Intelligence")
            gr.Markdown("Practical questions about savings, unexpected events, financial security, and future decisions.")
            for example in DECISION_QUESTIONS:
                button = gr.Button(example, variant="secondary", elem_classes=["question-example"])
                button.click(fn=lambda q=example: q, inputs=[], outputs=question_input)

        with gr.Column(scale=1, min_width=310, elem_id="integrated-panel"):
            gr.Markdown("#### Investment & Integrated Financial Intelligence")
            gr.Markdown("Portfolio, ranking, Monte Carlo, bootstrap, and combined liquidity and investment scenarios.")
            for example in INTEGRATED_INVESTMENT_QUESTIONS:
                button = gr.Button(example, variant="secondary", elem_classes=["question-example"])
                button.click(fn=lambda q=example: q, inputs=[], outputs=question_input)

    gr.Markdown(
        """---
        **Methodological note.** The Expense-to-Income Ratio measures the proportion of income absorbed by expenses.
        Simulations use predefined assumptions: a 30% reduction in employment income for six months is not necessarily equivalent to losing an annual bonus. Initial liquidity estimated
        from historical cash flows does not represent a verified bank balance. The selected period
        is passed to the agent; previously completed analyses retain their original calculation horizons.
        """
    )

# ============================================================
# 9. LAUNCH GRADIO
# ============================================================

demo_box32.launch(share=True, debug=False, quiet=True, show_error=False)
```
